## Part A — Python Fundamentals

<text style="color:#64748b">20 questions · Expand each answer when ready</text>

<details>
<summary><b style="color:#2563eb;">Q1. What is Python, and why is it widely used?</b></summary>

Python is a high-level, general-purpose language with readable syntax and a large ecosystem. It is widely used in automation, APIs, data engineering, scientific computing, ML and AI because libraries such as NumPy, pandas, PyTorch and FastAPI reduce implementation time. Python itself may be slower than compiled languages for CPU-bound loops, so production systems often move hot paths to optimized libraries, native extensions, or another service.

</details>

<details>
<summary><b style="color:#2563eb;">Q2. Is Python interpreted or compiled?</b></summary>

Python is commonly described as interpreted, but CPython first compiles source into bytecode, then its virtual machine executes that bytecode. The exact implementation matters: CPython, PyPy and other implementations can behave differently in performance. Avoid saying Python never compiles.

</details>

<details>
<summary><b style="color:#2563eb;">Q3. What is dynamic typing?</b></summary>

A variable name can refer to objects of different types over its lifetime; types belong to objects, not permanently to variable names. This enables flexible code but can defer type-related failures until runtime. Type hints, validation, tests and static analysis help catch errors earlier.

</details>

<details>
<summary><b style="color:#2563eb;">Q4. What is the difference between mutable and immutable objects?</b></summary>

Mutable objects can change in place; lists, dictionaries and sets are examples. Immutable objects cannot be changed after creation; integers, strings and tuples (though a tuple can contain mutable objects) are examples. This distinction affects aliasing, caching, hashability, function defaults and concurrency safety.

</details>

<details>
<summary><b style="color:#2563eb;">Q5. What is the difference between `==` and `is`?</b></summary>

`==` checks value equality through an object's equality implementation. `is` checks object identity—whether two references point to the same object. Use `is None` to test for `None`; do not use `is` to compare strings or integers by value because interning is an implementation detail.

</details>

<details>
<summary><b style="color:#2563eb;">Q6. What are truthy and falsy values?</b></summary>

Falsy values include `False`, `None`, numeric zero, and empty containers such as `[]`, `{}`, `set()` and `''`. Most other objects are truthy unless their type defines otherwise. Use `if not items:` when checking emptiness; use explicit comparisons when `0` and `None` have different business meanings.

</details>

<details>
<summary><b style="color:#2563eb;">Q7. What is the difference between `None`, `False`, and an empty list?</b></summary>

`None` commonly means missing or absent; `False` is a Boolean value; `[]` is a present collection with zero elements. Treating them as interchangeable can hide API or data-quality bugs. Define clear contracts for optional values.

</details>

<details>
<summary><b style="color:#2563eb;">Q8. Explain Python's basic numeric types.</b></summary>

`int` represents integers with arbitrary precision in normal Python implementations, `float` is typically IEEE-754 double precision, and `complex` stores real and imaginary components. Floating-point arithmetic can introduce rounding error; use `decimal.Decimal` for many decimal financial calculations and integer minor units (such as paise) where appropriate.

</details>

<details>
<summary><b style="color:#2563eb;">Q9. Why can `0.1 + 0.2` differ from `0.3`?</b></summary>

Many decimal fractions cannot be represented exactly in binary floating point, so the stored approximations are added. Compare with a tolerance using `math.isclose()` for numerical work, or use decimal arithmetic when exact decimal semantics are required.

</details>

<details>
<summary><b style="color:#2563eb;">Q10. What is a namespace?</b></summary>

A namespace maps names to objects. Local, enclosing, global and built-in namespaces are searched according to Python's LEGB rule. Understanding namespaces helps explain shadowing, closures, imports and why assigning to a name inside a function can make it local.

</details>

<details>
<summary><b style="color:#2563eb;">Q11. What is LEGB scope resolution?</b></summary>

Python resolves a name in Local, Enclosing, Global, then Built-in scope. A local assignment generally makes the name local throughout that function unless declared `global` or `nonlocal`. Avoid unnecessary global state because it makes tests, reuse and concurrent execution harder.

</details>

<details>
<summary><b style="color:#2563eb;">Q12. What is the difference between a script, module and package?</b></summary>

A script is typically executed as a program; a module is an importable Python file; a package organizes modules under a package namespace. Packages help separate concerns and support reusable applications. Keep import side effects minimal so importing a module does not unexpectedly start servers, jobs or downloads.

</details>

<details>
<summary><b style="color:#2563eb;">Q13. What does `if __name__ == '__main__':` do?</b></summary>

It runs a block only when the file is executed as the main program, not when imported as a module. Put command-line entry-point logic there to keep functions importable and testable.

</details>

<details>
<summary><b style="color:#2563eb;">Q14. What are comments, docstrings and type hints?</b></summary>

Comments explain reasoning; docstrings document a module, class or function; type hints express expected types for readers and static tools. Type hints are not generally runtime enforcement by themselves. Use all three deliberately and keep them synchronized with actual behavior.

</details>

<details>
<summary><b style="color:#2563eb;">Q15. Why is indentation significant in Python?</b></summary>

Indentation defines suites and blocks instead of braces. Inconsistent indentation can cause syntax errors or unintended logic. Use four spaces per level and configure the editor to avoid mixing tabs and spaces.

</details>

<details>
<summary><b style="color:#2563eb;">Q16. What is the difference between `break`, `continue` and `pass`?</b></summary>

`break` exits the nearest loop, `continue` skips to the next iteration, and `pass` is a no-op placeholder. Choose the one that communicates the intended control flow; avoid complicated nested loops where early returns or extracted functions improve readability.

</details>

<details>
<summary><b style="color:#2563eb;">Q17. What is the difference between `for` and `while`?</b></summary>

`for` iterates over an iterable and is usually preferred when processing a sequence or stream. `while` repeats while a condition remains true and is useful when the number of iterations is not known in advance. Ensure `while` loops have a reachable termination condition.

</details>

<details>
<summary><b style="color:#2563eb;">Q18. What are f-strings?</b></summary>

F-strings embed expressions in string literals, such as `f'User: {user_id}'`. They are convenient for formatting, but never build SQL queries or shell commands by interpolating untrusted input; use parameterized APIs instead.

</details>

<details>
<summary><b style="color:#2563eb;">Q19. How do slicing and negative indices work?</b></summary>

A slice `sequence[start:stop:step]` excludes `stop`. Negative indices count backward from the end. Slicing many built-in sequences creates a new object, so repeatedly slicing large data can use extra memory and time.

</details>

<details>
<summary><b style="color:#2563eb;">Q20. How do you read command-line arguments and environment variables?</b></summary>

Use `argparse` for command-line interfaces and `os.environ` or a configuration library for environment variables. Validate values at startup, define safe defaults, and keep secrets out of source code and logs.

</details>

## Part B — Collections and Complexity

<text style="color:#64748b">15 questions · Expand each answer when ready</text>

<details>
<summary><b style="color:#2563eb;">Q21. Compare list, tuple, set and dictionary.</b></summary>

A list is ordered and mutable; a tuple is ordered and immutable at the container level; a set stores unique hashable elements; a dictionary maps hashable keys to values and preserves insertion order in modern Python. Choose based on semantics, mutation needs and lookup patterns.

</details>

<details>
<summary><b style="color:#2563eb;">Q22. What is the average complexity of list append and membership?</b></summary>

Appending to a list is amortized O(1), though occasional resizing costs more. Membership (`x in list`) is O(n) because elements are checked sequentially. If many membership checks are needed, a set may offer average O(1) lookup when values are hashable.

</details>

<details>
<summary><b style="color:#2563eb;">Q23. Why are dictionary lookups usually fast?</b></summary>

Dictionaries use a hash table. Average lookup and insertion are O(1), but collisions and pathological inputs can worsen performance. Keys must be hashable, and equality/hash behavior must remain consistent.

</details>

<details>
<summary><b style="color:#2563eb;">Q24. What does hashable mean?</b></summary>

An object is hashable if it has a stable hash during its lifetime and equality-compatible behavior. Hashable objects can be dictionary keys or set elements. Immutable built-ins such as strings and integers are hashable; lists and dictionaries are not.

</details>

<details>
<summary><b style="color:#2563eb;">Q25. Why can't a list be a dictionary key?</b></summary>

A list is mutable and unhashable. If a key's hash changed after insertion, the dictionary could no longer reliably locate it. Use a tuple of hashable elements when a composite immutable key is appropriate.

</details>

<details>
<summary><b style="color:#2563eb;">Q26. How do you remove duplicates while preserving order?</b></summary>

For hashable values, use `list(dict.fromkeys(items))` in modern Python. If elements are unhashable, use an explicit comparison strategy or normalize them to a stable key. State whether order and original representations must be preserved.

</details>

<details>
<summary><b style="color:#2563eb;">Q27. What is a shallow copy versus a deep copy?</b></summary>

A shallow copy creates a new outer container but retains references to nested objects. A deep copy recursively copies objects where supported. Deep copying can be expensive and may be inappropriate for clients, locks, file handles or large tensors; often explicit construction is clearer.

</details>

<details>
<summary><b style="color:#2563eb;">Q28. What is aliasing, and why is it dangerous?</b></summary>

Aliasing means multiple names refer to the same mutable object. A mutation through one reference appears through the other. This can cause surprising side effects in shared configuration, cached objects and function arguments; copy deliberately when independent ownership is required.

</details>

<details>
<summary><b style="color:#2563eb;">Q29. Explain list comprehensions and their trade-offs.</b></summary>

A list comprehension builds a list from an iterable, optionally with a filter. It is concise and often fast, but a huge comprehension materializes all results in memory. Use a generator expression for lazy processing when the consumer can stream results.

</details>

<details>
<summary><b style="color:#2563eb;">Q30. What is a generator expression?</b></summary>

It creates a lazy iterator, for example `(x*x for x in values)`. Values are produced on demand, which can reduce memory use. A generator is usually single-pass, so do not expect to iterate over it repeatedly without recreating it.

</details>

<details>
<summary><b style="color:#2563eb;">Q31. How does `enumerate()` help?</b></summary>

`enumerate(iterable, start=0)` yields index-value pairs without manually managing a counter. It improves readability and reduces off-by-one errors.

</details>

<details>
<summary><b style="color:#2563eb;">Q32. What does `zip()` do, and what is a common pitfall?</b></summary>

`zip()` combines elements from iterables position by position and stops at the shortest input by default. If unequal lengths indicate bad data, use `zip(..., strict=True)` where supported or validate lengths explicitly.

</details>

<details>
<summary><b style="color:#2563eb;">Q33. What is dictionary unpacking?</b></summary>

`{**a, **b}` combines mappings, with later keys overriding earlier ones. This is useful for layered configuration, but implicit overrides can hide mistakes. Validate the final configuration and document precedence.

</details>

<details>
<summary><b style="color:#2563eb;">Q34. What are `collections.Counter`, `defaultdict` and `deque` useful for?</b></summary>

`Counter` counts hashable items, `defaultdict` supplies default values for missing keys, and `deque` supports efficient appends/pops at both ends. A deque is usually better than repeatedly removing index zero from a list.

</details>

<details>
<summary><b style="color:#2563eb;">Q35. How do you choose a data structure for a production workload?</b></summary>

Start with access patterns: membership, ordering, uniqueness, append/pop location, key lookup and memory limits. Estimate time and space complexity, then benchmark representative data. Avoid premature optimization, but do not use an O(n) list scan for millions of repeated membership checks without a reason.

</details>

## Part C — Functions and Functional Patterns

<text style="color:#64748b">15 questions · Expand each answer when ready</text>

<details>
<summary><b style="color:#2563eb;">Q36. What makes a function well-designed?</b></summary>

A function should have a clear responsibility, explicit inputs and outputs, predictable side effects, and meaningful names. Smaller functions are easier to test and reuse. Avoid splitting code into so many trivial functions that control flow becomes harder to follow.

</details>

<details>
<summary><b style="color:#2563eb;">Q37. Explain positional-only, positional-or-keyword and keyword-only arguments.</b></summary>

Python functions can define `/` to mark positional-only parameters and `*` to make later parameters keyword-only. This can make public APIs clearer and preserve flexibility as a library evolves. Keyword-only options are especially useful for flags and configuration.

</details>

<details>
<summary><b style="color:#2563eb;">Q38. What are `*args` and `**kwargs`?</b></summary>

`*args` collects extra positional arguments into a tuple; `**kwargs` collects extra keyword arguments into a dictionary. They are useful for wrappers and flexible APIs, but accepting everything without validation can make errors harder to detect.

</details>

<details>
<summary><b style="color:#2563eb;">Q39. Why are mutable default arguments a classic bug?</b></summary>

Default argument expressions are evaluated once when the function is defined. A default list or dictionary is therefore reused across calls. Use `None` as the default and create a fresh object inside the function.

</details>

<details>
<summary><b style="color:#2563eb;">Q40. What is a lambda?</b></summary>

A lambda is a small anonymous function expression, often used as a short key function or callback. Use `def` when logic is multi-step or needs documentation; nested lambdas usually reduce readability.

</details>

<details>
<summary><b style="color:#2563eb;">Q41. What is a closure?</b></summary>

A closure is a function that retains access to variables from its enclosing scope. Closures are useful for small factories and decorators. Be careful with late binding in loops and with hidden mutable state.

</details>

<details>
<summary><b style="color:#2563eb;">Q42. What is a decorator?</b></summary>

A decorator wraps or transforms a function or class, often for logging, timing, authorization or retries. Use `functools.wraps` to preserve metadata. Decorators should make behavior more consistent without hiding important control flow.

</details>

<details>
<summary><b style="color:#2563eb;">Q43. What is recursion, and what are its risks?</b></summary>

Recursion solves a problem by calling a function on smaller subproblems. It can be elegant for trees and divide-and-conquer algorithms, but Python has a recursion limit and function calls have overhead. Iterative solutions may be safer for deep inputs.

</details>

<details>
<summary><b style="color:#2563eb;">Q44. What is a higher-order function?</b></summary>

A higher-order function accepts a function as an argument or returns one. Examples include `map`, `filter`, sorting with `key=`, and decorators. This pattern supports reusable behavior but should remain readable to the team.

</details>

<details>
<summary><b style="color:#2563eb;">Q45. Explain `map`, `filter` and `sorted(key=...)`.</b></summary>

`map` transforms values, `filter` selects values based on a predicate, and `sorted(key=...)` sorts by a derived key. Comprehensions are often easier to read for simple transformations; choose the clearest form.

</details>

<details>
<summary><b style="color:#2563eb;">Q46. What is the difference between `return` and `yield`?</b></summary>

`return` ends a function and gives back a result. `yield` turns a function into a generator that can suspend and resume, producing values incrementally. Generators are useful for streaming files, records and model outputs.

</details>

<details>
<summary><b style="color:#2563eb;">Q47. What are type hints and when should you use them?</b></summary>

Type hints communicate expected types and enable tools such as mypy or pyright to catch many mistakes before runtime. They are valuable at module boundaries, public APIs and complex data transformations. Pair them with runtime validation when inputs are untrusted.

</details>

<details>
<summary><b style="color:#2563eb;">Q48. What are `Optional`, unions and generics?</b></summary>

A union expresses that a value may have one of several types; `str | None` means a string or `None`. Generics let reusable components preserve type relationships, such as a function that returns the same item type it receives. Use the syntax appropriate to the supported Python version.

</details>

<details>
<summary><b style="color:#2563eb;">Q49. What is dependency injection?</b></summary>

Dependency injection means passing dependencies—such as a database client, clock, model or HTTP client—into a component rather than constructing them internally. It improves testing, configuration and replacement. Avoid building a complex dependency framework when simple constructor arguments suffice.

</details>

<details>
<summary><b style="color:#2563eb;">Q50. How do you design a function that is safe to retry?</b></summary>

Prefer idempotent operations: repeating the same request should not create duplicate effects. Separate pure computation from side effects, use idempotency keys or unique constraints for writes, and retry only transient errors with bounded exponential backoff and jitter.

</details>

## Part D — Intermediate and Advanced Python

<text style="color:#64748b">20 questions · Expand each answer when ready</text>

<details>
<summary><b style="color:#2563eb;">Q51. What is an iterator?</b></summary>

An iterator produces values one at a time and implements `__iter__()` and `__next__()`. Iteration ends by raising `StopIteration`. Iterators enable uniform processing of lists, files, database cursors and streaming records.

</details>

<details>
<summary><b style="color:#2563eb;">Q52. What is the iterable versus iterator distinction?</b></summary>

An iterable can produce an iterator; an iterator is the stateful object that returns the next value. A list is reusable and iterable, while a generator object is an iterator that is typically consumed once.

</details>

<details>
<summary><b style="color:#2563eb;">Q53. What is a context manager?</b></summary>

A context manager manages setup and cleanup around a block, usually through `__enter__`/`__exit__` or a generator decorated with `contextmanager`. The `with` statement is used for files, locks, transactions and temporary resources so cleanup happens reliably, including when exceptions occur.

</details>

<details>
<summary><b style="color:#2563eb;">Q54. What is a descriptor?</b></summary>

A descriptor implements methods such as `__get__`, `__set__` or `__delete__` to customize attribute access. Properties, methods and many ORM field abstractions use descriptor behavior. Learn the concept for framework-level Python; do not create custom descriptors unless they simplify a real design.

</details>

<details>
<summary><b style="color:#2563eb;">Q55. What are dunder methods?</b></summary>

Special methods such as `__repr__`, `__len__`, `__iter__` and `__eq__` integrate custom objects with Python syntax and built-ins. Implement only behavior that has clear, consistent semantics; equality and hashing must be designed together.

</details>

<details>
<summary><b style="color:#2563eb;">Q56. What is a dataclass?</b></summary>

A dataclass generates common methods such as `__init__` and `__repr__` from annotated fields. It is useful for configuration and structured records. Use `frozen=True` when immutability is desired, and use `default_factory` for mutable defaults.

</details>

<details>
<summary><b style="color:#2563eb;">Q57. What is the difference between `__repr__` and `__str__`?</b></summary>

`__repr__` is the developer-oriented representation and ideally unambiguous; `__str__` is the user-friendly representation. Do not include credentials, personal data or sensitive payloads in either representation because logs and debugging tools may expose them.

</details>

<details>
<summary><b style="color:#2563eb;">Q58. What is the GIL in CPython?</b></summary>

The Global Interpreter Lock historically allows only one thread at a time to execute Python bytecode in a standard CPython process. Threads can still help with I/O-bound work, while CPU-bound Python code often needs multiprocessing, native code or external workers. GIL behavior is implementation- and version-dependent, so verify the runtime.

</details>

<details>
<summary><b style="color:#2563eb;">Q59. What is monkey patching?</b></summary>

Monkey patching changes an object or module at runtime. It can be useful in narrowly scoped tests, but production monkey patches create hidden dependencies and upgrade risks. Prefer explicit interfaces, dependency injection and supported extension points.

</details>

<details>
<summary><b style="color:#2563eb;">Q60. What are import side effects and circular imports?</b></summary>

Import side effects occur when importing a module performs work such as connecting to a service or launching a task. Circular imports happen when modules depend on each other during initialization. Keep module top levels declarative, move shared contracts to a lower-level module, and defer runtime setup to application startup.

</details>

<details>
<summary><b style="color:#2563eb;">Q61. What is a virtual environment?</b></summary>

A virtual environment isolates project dependencies from the system interpreter and other projects. Use a reproducible dependency file, pin critical versions where appropriate, and test clean installs. Isolation prevents one project's dependency upgrade from breaking another.

</details>

<details>
<summary><b style="color:#2563eb;">Q62. What is the difference between a dependency range and a lock file?</b></summary>

A dependency declaration describes acceptable versions; a lock file records a resolved environment for repeatable installs. Applications often benefit from locked deployments, while libraries need carefully chosen compatibility ranges. Rebuild and test environments rather than assuming local state is reproducible.

</details>

<details>
<summary><b style="color:#2563eb;">Q63. What are Python wheels and source distributions?</b></summary>

A wheel is a built distribution that can be installed without building from source in many cases. A source distribution contains source and build metadata. Wheels can improve deployment speed, but native extensions must match the target platform and Python ABI.

</details>

<details>
<summary><b style="color:#2563eb;">Q64. What is serialization? Compare JSON and pickle.</b></summary>

Serialization converts an object into a storable or transferable representation. JSON is broadly interoperable but supports a limited set of data types. Pickle can represent more Python objects but loading an untrusted pickle can execute arbitrary code; never unpickle untrusted data.

</details>

<details>
<summary><b style="color:#2563eb;">Q65. How does garbage collection work in Python?</b></summary>

CPython primarily uses reference counting and supplements it with cyclic garbage collection for reference cycles. Do not depend on immediate finalization for critical cleanup; use context managers for files, sockets and locks. Memory may not immediately return to the operating system after objects are freed.

</details>

<details>
<summary><b style="color:#2563eb;">Q66. What is a memory leak in a Python service?</b></summary>

A leak-like growth pattern can come from retained references, unbounded caches, queues, global collections, callbacks or native-library allocations. Compare snapshots, track object counts and process memory, and reproduce under realistic load. Distinguish live-object growth from allocator retention and external memory.

</details>

<details>
<summary><b style="color:#2563eb;">Q67. What are slots in a class?</b></summary>

`__slots__` can reduce per-instance memory and restrict dynamically added attributes, depending on inheritance and implementation details. It can complicate inheritance and weak references. Measure before using it; data model changes should be driven by memory evidence.

</details>

<details>
<summary><b style="color:#2563eb;">Q68. What is late binding in closures?</b></summary>

A closure usually looks up a free variable when called, not when the closure is created. Functions created in a loop can therefore all see the final loop value. Capture the current value through a default argument or a small factory function.

</details>

<details>
<summary><b style="color:#2563eb;">Q69. What is the difference between `@staticmethod` and `@classmethod`?</b></summary>

A static method receives no implicit instance or class; a class method receives the class as `cls`, which helps implement alternative constructors and subclass-aware factories. If a function does not need class or instance state, a module-level function may be simpler.

</details>

<details>
<summary><b style="color:#2563eb;">Q70. What are structural pattern matching and when is it useful?</b></summary>

The `match` statement can branch on value patterns and destructure structured data. It is useful for interpreters, protocol messages and state handling when patterns are clearer than long conditional chains. It is not a replacement for every `if` statement and requires a compatible Python version.

</details>

## Part E — Object-Oriented Programming and Design

<text style="color:#64748b">15 questions · Expand each answer when ready</text>

<details>
<summary><b style="color:#2563eb;">Q71. What are the four commonly taught OOP principles?</b></summary>

Encapsulation groups data and behavior; abstraction exposes essential operations while hiding implementation details; inheritance reuses or specializes behavior; polymorphism lets different implementations satisfy a common interface. These are design tools, not goals by themselves.

</details>

<details>
<summary><b style="color:#2563eb;">Q72. Inheritance versus composition: which should you prefer?</b></summary>

Inheritance models a genuine substitutable 'is-a' relationship; composition builds behavior from collaborators and often reduces coupling. Prefer composition when behavior needs to be swapped, tested independently or assembled in different combinations. Inheritance is appropriate when the subtype truly honors the base contract.

</details>

<details>
<summary><b style="color:#2563eb;">Q73. What is polymorphism in Python?</b></summary>

Polymorphism allows code to work with different objects through a shared protocol, often using duck typing. A caller can depend on `predict()` rather than a specific model class. Clear contracts and tests are still needed because dynamic dispatch does not guarantee compatible behavior.

</details>

<details>
<summary><b style="color:#2563eb;">Q74. What is an abstract base class?</b></summary>

An ABC defines an interface-like contract and can prevent instantiation until abstract methods are implemented. It is useful when explicit inheritance and runtime contract enforcement matter. For simpler systems, a `Protocol` or duck-typed interface may be less restrictive.

</details>

<details>
<summary><b style="color:#2563eb;">Q75. What is a Python `Protocol`?</b></summary>

A typing `Protocol` describes the methods and attributes an object must provide, enabling structural subtyping in static analysis. It is useful for interfaces such as a model with `predict()` or a storage adapter with `get()` and `put()`, without requiring a shared base class.

</details>

<details>
<summary><b style="color:#2563eb;">Q76. Explain SOLID in practical terms.</b></summary>

SOLID describes five object-oriented design principles: single responsibility, open/closed, Liskov substitution, interface segregation and dependency inversion. Apply them to reduce change impact and make components testable; do not turn every small script into dozens of classes.

</details>

<details>
<summary><b style="color:#2563eb;">Q77. What is the Liskov Substitution Principle?</b></summary>

A subtype should be usable wherever its base type is expected without breaking the caller's assumptions. A subtype that strengthens preconditions or weakens guarantees violates the principle. In production, this prevents surprising behavior when swapping implementations.

</details>

<details>
<summary><b style="color:#2563eb;">Q78. What is a factory pattern?</b></summary>

A factory centralizes object creation when construction depends on configuration, environment or requested implementation. For example, it can select a local model or remote inference client. Keep factories simple and explicit; avoid turning them into hidden service locators.

</details>

<details>
<summary><b style="color:#2563eb;">Q79. What is the strategy pattern?</b></summary>

The strategy pattern encapsulates interchangeable algorithms behind a common interface. It is useful for choosing ranking methods, retry policies, routing logic or model providers. Configuration should select among known strategies, with validation and observability.

</details>

<details>
<summary><b style="color:#2563eb;">Q80. How should configuration be represented in a Python service?</b></summary>

Use a typed configuration object or validated settings model, with a clear precedence such as defaults, file, environment, then explicit overrides. Validate at startup and fail fast on invalid values. Never silently replace a missing production secret with a development default.

</details>

<details>
<summary><b style="color:#2563eb;">Q81. How do you prevent a class from becoming a 'god object'?</b></summary>

Keep each class focused on one cohesive responsibility and move unrelated work into collaborators. If a class handles HTTP, persistence, model inference, retries and metrics, split those concerns behind small interfaces. Refactor around reasons to change, not arbitrary line counts.

</details>

<details>
<summary><b style="color:#2563eb;">Q82. What is coupling versus cohesion?</b></summary>

Coupling measures dependence between components; cohesion measures how closely a component's responsibilities belong together. Aim for high cohesion and controlled coupling so one change does not ripple across the application. Dependency injection and stable contracts help.

</details>

<details>
<summary><b style="color:#2563eb;">Q83. How do you make an object effectively immutable?</b></summary>

Avoid exposing mutable internals, use immutable field types where possible, and provide methods that return new values rather than mutating shared state. `frozen=True` dataclasses prevent field reassignment but do not recursively freeze nested objects.

</details>

<details>
<summary><b style="color:#2563eb;">Q84. How do you design a pluggable model interface?</b></summary>

Define a small contract such as `predict(inputs)` plus explicit metadata about input schema, output schema, and supported capabilities. Provide adapters for local and remote models, validate outputs, and test all implementations against the same contract.

</details>

<details>
<summary><b style="color:#2563eb;">Q85. When is functional programming preferable to OOP?</b></summary>

Pure functions are especially useful for transformations, validation and deterministic business rules because they are easy to test and parallelize. OOP is useful for stateful resources and interchangeable collaborators. Most production Python benefits from combining both.

</details>

## Part F — Errors, Testing, Debugging and Production

<text style="color:#64748b">20 questions · Expand each answer when ready</text>

<details>
<summary><b style="color:#2563eb;">Q86. How should exceptions be handled?</b></summary>

Catch exceptions where you can add context, recover, translate them to a meaningful boundary error, or clean up resources. Avoid broad `except Exception: pass`, which hides failures. Preserve the original cause with exception chaining when translating errors.

</details>

<details>
<summary><b style="color:#2563eb;">Q87. When should you create a custom exception?</b></summary>

Create a custom exception when callers need to distinguish a meaningful failure category, such as invalid input, unavailable dependency or model timeout. Keep the hierarchy small and include safe context without secrets or raw sensitive payloads.

</details>

<details>
<summary><b style="color:#2563eb;">Q88. What is the difference between an error and an exception?</b></summary>

In everyday Python discussion, 'error' is a broad term for failure; an exception is Python's runtime mechanism for propagating an abnormal condition. Some exceptions represent expected control outcomes, while others indicate defects or operational failures.

</details>

<details>
<summary><b style="color:#2563eb;">Q89. What is exception chaining?</b></summary>

Use `raise NewError(...) from exc` to preserve the original exception as the cause. This makes the boundary error understandable while retaining the lower-level diagnostic trail. Do not leak internal stack traces to end users.

</details>

<details>
<summary><b style="color:#2563eb;">Q90. What is unit, integration and end-to-end testing?</b></summary>

Unit tests exercise a small unit in isolation; integration tests verify interactions between components; end-to-end tests exercise a realistic user flow. Use many fast unit tests, targeted integration tests, and a smaller number of robust end-to-end tests.

</details>

<details>
<summary><b style="color:#2563eb;">Q91. What are mocks and fakes?</b></summary>

A mock verifies interactions or provides controlled behavior; a fake is a lightweight working substitute such as an in-memory repository. Prefer fakes when they make tests realistic and readable. Excessive mocking can make tests pass even when real integration behavior is broken.

</details>

<details>
<summary><b style="color:#2563eb;">Q92. What is test-driven development?</b></summary>

TDD cycles through a failing test, a minimal implementation, and refactoring. It can clarify contracts and edge cases, but the goal is reliable design—not maximizing test count or testing private implementation details.

</details>

<details>
<summary><b style="color:#2563eb;">Q93. What is a regression test?</b></summary>

A regression test captures a previously fixed bug so it does not return. Reproduce the failure with a minimal input, add the test, fix the defect, and ensure the test fails against the old behavior and passes against the fix.

</details>

<details>
<summary><b style="color:#2563eb;">Q94. How should logs differ from metrics and traces?</b></summary>

Logs are event records; metrics are aggregated numeric measurements over time; traces show a request's path across components. Together they help identify what failed, how often, and where latency was spent. Use correlation IDs and avoid logging secrets or full user payloads.

</details>

<details>
<summary><b style="color:#2563eb;">Q95. What should a production log contain?</b></summary>

Include timestamp, severity, service/version, request or trace ID, operation, outcome, duration and a safe error category. Prefer structured logs over brittle text parsing. Redact credentials, tokens, personal data and sensitive prompts.

</details>

<details>
<summary><b style="color:#2563eb;">Q96. What is observability?</b></summary>

Observability is the ability to infer internal system state from outputs such as logs, metrics and traces. For a Python API, monitor request rate, latency percentiles, error rates, saturation, dependency failures and business-level outcomes.

</details>

<details>
<summary><b style="color:#2563eb;">Q97. What is a health check?</b></summary>

A liveness check asks whether the process should be restarted; a readiness check asks whether it can accept traffic. Do not make liveness depend on every downstream dependency, or a temporary outage may trigger restart storms. Readiness should reflect critical dependencies and startup completion.

</details>

<details>
<summary><b style="color:#2563eb;">Q98. How should API input be validated?</b></summary>

Validate types, ranges, lengths, allowed values and cross-field constraints at the boundary. Reject malformed requests with actionable client errors; do not let invalid data travel deep into the system. Validation is not a substitute for parameterized SQL, authorization or output encoding.

</details>

<details>
<summary><b style="color:#2563eb;">Q99. How do you handle secrets in Python applications?</b></summary>

Load secrets from a secret manager or protected environment injection, rotate them, scope permissions, and never commit them or print them in logs. Avoid baking secrets into images or notebooks. If a secret is exposed, revoke and rotate it rather than only deleting the visible copy.

</details>

<details>
<summary><b style="color:#2563eb;">Q100. What is idempotency in an API?</b></summary>

An operation is idempotent if repeating it has the same intended effect as performing it once. It matters when clients retry after timeouts because they may not know whether the first request succeeded. Use idempotency keys, unique constraints and durable operation records for non-idempotent writes.

</details>

<details>
<summary><b style="color:#2563eb;">Q101. How should retries be implemented?</b></summary>

Retry only transient failures, set attempt and total-time limits, use exponential backoff with jitter, and respect server guidance such as `Retry-After`. Do not retry invalid input or permanent authorization failures. Retries can amplify load, so pair them with concurrency limits and circuit breakers.

</details>

<details>
<summary><b style="color:#2563eb;">Q102. What is a circuit breaker?</b></summary>

A circuit breaker temporarily stops calls to a failing dependency after a threshold, allows a recovery probe, and closes again when healthy. It prevents repeated slow failures from consuming all worker capacity. Configure thresholds based on dependency behavior and expose state transitions in metrics.

</details>

<details>
<summary><b style="color:#2563eb;">Q103. What is graceful degradation?</b></summary>

Graceful degradation preserves core functionality when a noncritical dependency fails—for example, returning a response without optional enrichment. Define what can be omitted, how stale data is labeled, and when the system must fail closed for safety or correctness.

</details>

<details>
<summary><b style="color:#2563eb;">Q104. How do you debug a memory or latency incident?</b></summary>

Establish the timeline and blast radius, inspect metrics and recent deployments, reproduce under controlled load, and profile before changing code. For memory, inspect object retention and native allocations; for latency, break down time across queueing, application work and dependencies. Roll back or mitigate first if customer impact is severe.

</details>

<details>
<summary><b style="color:#2563eb;">Q105. What is a postmortem, and what makes it useful?</b></summary>

A blameless postmortem records impact, timeline, contributing conditions, detection gaps and prioritized corrective actions with owners. Focus on system conditions and prevention, not individual blame. Verify that action items reduce recurrence and improve detection.

</details>

## Part G — Concurrency, Performance and Architecture

<text style="color:#64748b">15 questions · Expand each answer when ready</text>

<details>
<summary><b style="color:#2563eb;">Q106. Compare threading, multiprocessing and asyncio.</b></summary>

Threads are useful for blocking I/O and share process memory; processes can use multiple CPU cores for Python CPU-bound work but incur serialization and memory costs; `asyncio` supports many concurrent I/O operations when dependencies are async-compatible. Choose based on workload and operational complexity.

</details>

<details>
<summary><b style="color:#2563eb;">Q107. What is a race condition?</b></summary>

A race condition occurs when correctness depends on the timing of concurrent operations. Shared counters, caches and multi-step read-modify-write logic are common examples. Use locks, queues, atomic database operations or immutable message passing as appropriate.

</details>

<details>
<summary><b style="color:#2563eb;">Q108. What is a deadlock?</b></summary>

A deadlock occurs when tasks wait indefinitely for resources held by one another. Reduce risk by acquiring locks in a consistent order, holding locks briefly, using timeouts where appropriate, and avoiding network calls while holding critical locks.

</details>

<details>
<summary><b style="color:#2563eb;">Q109. What does `async`/`await` do?</b></summary>

`async def` defines a coroutine and `await` suspends it while an awaitable operation completes, allowing the event loop to run other tasks. It does not automatically make CPU-heavy work faster. Blocking calls inside an async handler can stall the entire event loop.

</details>

<details>
<summary><b style="color:#2563eb;">Q110. What is a common asyncio production failure?</b></summary>

Calling a blocking library such as synchronous network I/O inside an async request handler blocks the event loop. Use an async-compatible client or offload blocking work to a bounded thread pool. Measure event-loop lag and enforce timeouts.

</details>

<details>
<summary><b style="color:#2563eb;">Q111. Why is unbounded concurrency dangerous?</b></summary>

It can exhaust memory, file descriptors, connection pools or downstream capacity and cause cascading failures. Use semaphores, bounded queues, worker limits and backpressure. Load shedding is often safer than accepting work the system cannot complete.

</details>

<details>
<summary><b style="color:#2563eb;">Q112. What is backpressure?</b></summary>

Backpressure communicates that a downstream consumer cannot accept work at the current rate. It can be implemented with bounded queues, demand-aware streams, rate limits or rejected requests. Without it, queues grow until latency and memory become unacceptable.

</details>

<details>
<summary><b style="color:#2563eb;">Q113. What is caching and what can go wrong?</b></summary>

Caching avoids repeated expensive work. Common failures include stale values, incorrect cache keys, unbounded memory, stampedes, privacy leakage across users and invalidation errors. Define TTL, ownership, versioning, invalidation and observability before deploying a cache.

</details>

<details>
<summary><b style="color:#2563eb;">Q114. What is cache stampede, and how can you mitigate it?</b></summary>

Many callers may recompute the same expired key simultaneously. Use single-flight request coalescing, randomized TTLs, stale-while-revalidate, or a distributed lock when justified. Ensure the mitigation does not create a new single point of failure.

</details>

<details>
<summary><b style="color:#2563eb;">Q115. How do you profile Python code?</b></summary>

Use `cProfile` for function-level CPU profiling, `timeit` for focused microbenchmarks, and memory tools for allocations and retained objects. Profile representative workloads and include I/O and serialization costs. Optimize the measured bottleneck, not the most visually complex code.

</details>

<details>
<summary><b style="color:#2563eb;">Q116. What is Big-O, and what does it not tell you?</b></summary>

Big-O describes asymptotic growth in time or space as input size increases. It does not capture constants, cache locality, network latency, database indexes or allocation overhead. Use complexity analysis to reject poor approaches and benchmarks to compare realistic implementations.

</details>

<details>
<summary><b style="color:#2563eb;">Q117. How do you design a resilient Python worker?</b></summary>

Use a durable queue, bounded concurrency, timeouts, retry classification, idempotent processing, dead-letter handling and structured metrics. Track queue age as well as queue depth. Ensure shutdown stops intake, finishes or safely requeues in-flight work, and closes resources.

</details>

<details>
<summary><b style="color:#2563eb;">Q118. How should a Python application be packaged and deployed?</b></summary>

Separate application code from environment configuration, define dependencies, run tests and security checks, build a reproducible artifact, and deploy through a controlled pipeline. Add startup validation, health endpoints, logs and rollback capability. Avoid relying on files or packages present only on a developer's machine.

</details>

<details>
<summary><b style="color:#2563eb;">Q119. What is horizontal scaling versus vertical scaling?</b></summary>

Vertical scaling gives a machine more CPU or memory; horizontal scaling adds instances. Horizontal scaling requires careful handling of shared state, sessions, queues, idempotency and database capacity. Scale based on measured bottlenecks and service-level objectives rather than request count alone.

</details>

<details>
<summary><b style="color:#2563eb;">Q120. What is an SLO, and how does it guide engineering?</b></summary>

A service-level objective is a target for a reliability measure, such as a latency percentile or successful-request ratio over a time window. SLOs help balance feature delivery and reliability work. Define user-relevant indicators, error budgets and alerts that trigger action rather than noise.

</details>

```python
        if nums[mid] == target:
            return mid
        if nums[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return -1
```

<summary><b style="color:#7c3aed;">C20. Retry a transient operation</b> — Implement bounded retries with exponential backoff. This demonstration sleeps; production code should inject sleep/clock for tests.</summary>

**Approach and answer**

```python
import random
import time
from collections.abc import Callable

def retry(operation: Callable[[], object], attempts: int = 3, base_delay: float = 0.1):
    if attempts < 1:
        raise ValueError('attempts must be at least 1')
    for attempt in range(attempts):
        try:
            return operation()
        except (TimeoutError, ConnectionError):
            if attempt == attempts - 1:
                raise
            delay = base_delay * (2 ** attempt) + random.uniform(0, base_delay)
            time.sleep(delay)

# Production: set an overall deadline, classify retryable errors, cap delay,
# honor Retry-After where relevant, and ensure the operation is safe to retry.
```

<summary><b style="color:#7c3aed;">C21. Implement a simple LRU cache</b> — Use the standard library's OrderedDict to demonstrate least-recently-used eviction.</summary>

**Approach and answer**

```python
from collections import OrderedDict

class LRUCache:
    def __init__(self, capacity: int):
        if capacity < 1:
            raise ValueError('capacity must be positive')
        self.capacity = capacity
        self.data = OrderedDict()

    def get(self, key):
        if key not in self.data:
            return None
        self.data.move_to_end(key)
        return self.data[key]

    def put(self, key, value):
        if key in self.data:
            self.data.move_to_end(key)
        self.data[key] = value
        if len(self.data) > self.capacity:
            self.data.popitem(last=False)

cache = LRUCache(2)
cache.put('a', 1); cache.put('b', 2); cache.get('a'); cache.put('c', 3)
assert cache.get('b') is None
assert cache.get('a') == 1
# Average O(1) operations. This example is not thread-safe or distributed.
```

**Approach and answer**

```python
from concurrent.futures import ThreadPoolExecutor, TimeoutError as FutureTimeout

def run_with_timeout(operation, timeout_seconds: float):
    if timeout_seconds <= 0:
        raise ValueError('timeout must be positive')
    executor = ThreadPoolExecutor(max_workers=1)
    future = executor.submit(operation)
    try:
        return future.result(timeout=timeout_seconds)
    except FutureTimeout as exc:
        future.cancel()  # Only cancels if work has not started.
        raise TimeoutError('operation exceeded its deadline') from exc
    finally:
        # wait=False avoids waiting here, but running work may continue in the thread.
        executor.shutdown(wait=False, cancel_futures=True)

# For hard isolation, use process/service-level deadlines or client-library timeouts.
# Do not use this helper as a substitute for cancelling an external network request.
```

# Part J — Production Scenario and Architect-Level Questions

Use this answer pattern: **clarify impact → contain risk → diagnose → fix → prevent recurrence → measure recovery**.

<details>
<summary><b style="color:#ea580c;">S1. A Python API's p95 latency doubled after deployment. What do you do?</b></summary>

Confirm the change window and affected routes, compare p50/p95/p99 and error rates, and inspect traces to split time among application code, database, network and model/provider calls. Roll back if impact is material and rollback is safe. Profile the specific bottleneck, reproduce with representative load, then deploy a measured fix with a canary and regression test. Add an alert tied to the service objective.

</details>

<details>
<summary><b style="color:#ea580c;">S2. Memory grows steadily in a long-running worker.</b></summary>

Check whether growth is per request, per batch, or correlated with a specific input. Compare process RSS with Python allocation snapshots; inspect unbounded caches, retained task results, global lists, queues, reference cycles and native allocations. Mitigate with bounded queues/caches or worker recycling if needed, but find the retention path and add a load test plus memory-growth alert.

</details>

<details>
<summary><b style="color:#ea580c;">S3. A downstream API starts timing out and retries increase traffic.</b></summary>

Stop retry amplification by limiting concurrency and applying bounded retries with exponential backoff and jitter. Use circuit breaking and request deadlines; honor rate-limit guidance. Provide a clear degraded response if business rules allow it. Track timeout rate, retry count, dependency latency and queue age, and coordinate recovery with the dependency owner.

</details>

<details>
<summary><b style="color:#ea580c;">S4. Duplicate payments or jobs appear after client retries.</b></summary>

Assume a timeout may happen after the side effect committed. Introduce an idempotency key with durable storage and a uniqueness constraint, return the original result for duplicates, and make processing state transitions atomic. Add tests for retries before, during and after commit; reconcile historical duplicates if required.

</details>

<details>
<summary><b style="color:#ea580c;">S5. A batch job processes 50 million records and runs out of memory.</b></summary>

Avoid loading all records into one in-memory list. Stream input or read in bounded chunks, use columnar formats where useful, avoid unnecessary copies, and aggregate incrementally. Measure peak memory and throughput on representative data. If the job needs distributed execution, partition by a stable key and make writes restartable and idempotent.

</details>

<details>
<summary><b style="color:#ea580c;">S6. A model service returns occasional malformed output.</b></summary>

Validate the response schema at the boundary, record safe metadata about model/version and failure category, and reject or quarantine invalid output. If a fallback is safe, route to it under an explicit policy. Test adversarial and edge-case inputs, track schema failure rates, and avoid silently coercing values that change meaning.

</details>

<details>
<summary><b style="color:#ea580c;">S7. An async web service becomes unresponsive under load.</b></summary>

Look for blocking I/O inside the event loop, CPU-heavy work, unbounded tasks and saturated connection pools. Measure event-loop lag and queue depth. Replace blocking clients with async clients or bounded offloading, apply backpressure and timeouts, and load-test at the expected concurrency.

</details>

<details>
<summary><b style="color:#ea580c;">S8. A new dependency upgrade breaks production only.</b></summary>

Compare lock files, Python versions, OS/architecture, environment variables and native-library versions between staging and production. Reproduce using the production artifact and a clean environment. Roll back or pin a known-good version, then add compatibility tests and artifact-level deployment validation.

</details>

<details>
<summary><b style="color:#ea580c;">S9. Sensitive user data appears in logs.</b></summary>

Restrict access and stop the leaking path immediately. Determine what data, where, and for how long it was exposed; follow incident and legal processes. Redact at the logging boundary, use allowlisted structured fields, add automated tests that assert secrets are absent, and rotate credentials if any were logged.

</details>

<details>
<summary><b style="color:#ea580c;">S10. How would you design a Python inference service for 10x traffic growth?</b></summary>

Start with an SLO and workload profile: request rate, payload size, latency percentiles, model runtime and cost. Keep the API stateless where possible, reuse loaded models safely, bound concurrency, batch when latency permits, and scale horizontally behind a load balancer. Protect downstreams with quotas, timeouts, backpressure and circuit breakers. Use canaries, autoscaling signals, model-version rollbacks, traces and load tests. Confirm the database, artifact store and model-serving layer—not just the API CPU—can handle the target load.

</details>
