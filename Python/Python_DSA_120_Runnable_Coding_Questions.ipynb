### 04 · Maximum of three numbers
**Problem:** Find the maximum without using max().

**Expected example:** `12`  
**Complexity:** O(1).

```python
def maximum_of_three(a, b, c):
    result = a
    if b > result:
        result = b
    if c > result:
        result = c
    return result

print(maximum_of_three(4, 12, 7))
assert maximum_of_three(-5, -2, -9) == -2
```

### 05 · Factorial iteratively
**Problem:** Calculate n! and reject negative input.

**Expected example:** `120`  
**Complexity:** O(n) time, O(1) extra space.

```python
def factorial(n):
    if n < 0:
        raise ValueError("n must be non-negative")
    result = 1
    for value in range(2, n + 1):
        result *= value
    return result

print(factorial(5))
assert factorial(0) == 1
```

### 07 · Prime number check
**Problem:** Check if an integer is prime.

**Expected example:** `True False`  
**Complexity:** O(√n) time, O(1) space.

```python
def is_prime(n):
    if n < 2:
        return False
    divisor = 2
    while divisor * divisor <= n:
        if n % divisor == 0:
            return False
        divisor += 1
    return True

print(is_prime(29), is_prime(30))
assert is_prime(2) and not is_prime(1)
```

### 08 · List primes up to n
**Problem:** Return every prime from 2 through n using the sieve of Eratosthenes.

**Expected example:** `[2, 3, 5, 7, 11, 13, 17, 19]`  
**Complexity:** O(n log log n) time, O(n) space.

```python
def primes_up_to(n):
    if n < 2:
        return []
    sieve = [True] * (n + 1)
    sieve[0] = sieve[1] = False
    p = 2
    while p * p <= n:
        if sieve[p]:
            for multiple in range(p * p, n + 1, p):
                sieve[multiple] = False
        p += 1
    return [i for i, prime in enumerate(sieve) if prime]

print(primes_up_to(20))
assert primes_up_to(10) == [2, 3, 5, 7]
```

### 09 · Fibonacci sequence
**Problem:** Return the first n Fibonacci numbers iteratively.

**Expected example:** `[0, 1, 1, 2, 3, 5, 8, 13]`  
**Complexity:** O(n) time, O(n) output space.

```python
def fibonacci_sequence(n):
    if n < 0:
        raise ValueError("n must be non-negative")
    result, a, b = [], 0, 1
    for _ in range(n):
        result.append(a)
        a, b = b, a + b
    return result

print(fibonacci_sequence(8))
assert fibonacci_sequence(0) == []
```

### 10 · GCD of two integers
**Problem:** Compute the greatest common divisor using Euclid's algorithm.

**Expected example:** `6`  
**Complexity:** O(log(min(a,b))) time.

```python
def gcd(a, b):
    a, b = abs(a), abs(b)
    while b:
        a, b = b, a % b
    return a

print(gcd(48, 18))
assert gcd(0, 7) == 7
```

### 18 · Run-length encode
**Problem:** Encode repeated consecutive characters, e.g. aaabb -> a3b2.

**Expected example:** `a3b2c1`  
**Complexity:** O(n) time.

```python
def run_length_encode(text):
    if not text:
        return ""
    output, count = [], 1
    for i in range(1, len(text)):
        if text[i] == text[i - 1]:
            count += 1
        else:
            output.append(f"{text[i - 1]}{count}")
            count = 1
    output.append(f"{text[-1]}{count}")
    return ''.join(output)

print(run_length_encode("aaabbc"))
assert run_length_encode("") == ""
```

### 19 · Longest common prefix
**Problem:** Find the longest prefix shared by every string.

**Expected example:** `fl`  
**Complexity:** O(total characters) in the straightforward implementation.

```python
def longest_common_prefix(words):
    if not words:
        return ""
    prefix = words[0]
    for word in words[1:]:
        while not word.startswith(prefix):
            prefix = prefix[:-1]
            if not prefix:
                return ""
    return prefix

print(longest_common_prefix(["flower", "flow", "flight"]))
assert longest_common_prefix(["dog", "racecar"]) == ""
```

### 22 · Find minimum and maximum
**Problem:** Find min and max in one pass; return None for empty input.

**Expected example:** `(-2, 10)`  
**Complexity:** O(n) time, O(1) extra space.

```python
def min_max(values):
    if not values:
        return None
    low = high = values[0]
    for value in values[1:]:
        if value < low: low = value
        if value > high: high = value
    return low, high

print(min_max([8, -2, 10, 3]))
assert min_max([]) is None
```

### 24 · Second-largest distinct number
**Problem:** Return the second-largest distinct value, or None if unavailable.

**Expected example:** `7`  
**Complexity:** O(n) average time, O(n) space.

```python
def second_largest(values):
    distinct = set(values)
    if len(distinct) < 2:
        return None
    largest = max(distinct)
    return max(v for v in distinct if v != largest)

print(second_largest([4, 9, 9, 7, 1]))
assert second_largest([5, 5]) is None
```

### 25 · Rotate list right by k
**Problem:** Rotate a list right by k positions.

**Expected example:** `[4, 5, 1, 2, 3]`  
**Complexity:** O(n) time and space.

```python
def rotate_right(values, k):
    if not values:
        return []
    k %= len(values)
    return values[-k:] + values[:-k] if k else values[:]

print(rotate_right([1, 2, 3, 4, 5], 2))
assert rotate_right([1, 2], 0) == [1, 2]
```

### 26 · Move zeros to the end
**Problem:** Keep non-zero order unchanged.

**Expected example:** `[1, 3, 12, 0, 0]`  
**Complexity:** O(n) time and space.

```python
def move_zeros(values):
    nonzero = [x for x in values if x != 0]
    return nonzero + [0] * (len(values) - len(nonzero))

print(move_zeros([0, 1, 0, 3, 12]))
assert move_zeros([0, 0]) == [0, 0]
```

### 27 · Merge two sorted lists
**Problem:** Merge sorted lists in linear time.

**Expected example:** `[1, 2, 3, 4, 8, 9]`  
**Complexity:** O(n + m) time.

```python
def merge_sorted(a, b):
    i = j = 0
    result = []
    while i < len(a) and j < len(b):
        if a[i] <= b[j]:
            result.append(a[i]); i += 1
        else:
            result.append(b[j]); j += 1
    result.extend(a[i:]); result.extend(b[j:])
    return result

print(merge_sorted([1, 4, 8], [2, 3, 9]))
assert merge_sorted([], [1]) == [1]
```

### 28 · Two Sum
**Problem:** Return indices of two values adding to target, or None.

**Expected example:** `(0, 1)`  
**Complexity:** O(n) average time, O(n) space.

```python
def two_sum(values, target):
    seen = {}
    for i, value in enumerate(values):
        needed = target - value
        if needed in seen:
            return seen[needed], i
        seen[value] = i
    return None

print(two_sum([2, 7, 11, 15], 9))
assert two_sum([3, 3], 6) == (0, 1)
```

### 29 · Maximum subarray sum
**Problem:** Use Kadane's algorithm for a non-empty maximum-sum contiguous subarray.

**Expected example:** `6`  
**Complexity:** O(n) time, O(1) extra space.

```python
def max_subarray_sum(values):
    if not values:
        raise ValueError("values cannot be empty")
    current = best = values[0]
    for value in values[1:]:
        current = max(value, current + value)
        best = max(best, current)
    return best

print(max_subarray_sum([-2, 1, -3, 4, -1, 2, 1, -5, 4]))
assert max_subarray_sum([-3, -1]) == -1
```

### 30 · Product except self
**Problem:** Return each position's product excluding its own value, without division.

**Expected example:** `[24, 12, 8, 6]`  
**Complexity:** O(n) time, O(1) extra space excluding output.

```python
def product_except_self(values):
    result = [1] * len(values)
    prefix = 1
    for i, value in enumerate(values):
        result[i] = prefix
        prefix *= value
    suffix = 1
    for i in range(len(values) - 1, -1, -1):
        result[i] *= suffix
        suffix *= values[i]
    return result

print(product_except_self([1, 2, 3, 4]))
assert product_except_self([0, 2, 0]) == [0, 0, 0]
```

### 31 · Chunk a list
**Problem:** Yield list chunks of a positive size.

**Expected example:** `[[1, 2], [3, 4], [5]]`  
**Complexity:** O(n) overall.

```python
def chunks(values, size):
    if size <= 0:
        raise ValueError("size must be positive")
    for start in range(0, len(values), size):
        yield values[start:start + size]

print(list(chunks([1, 2, 3, 4, 5], 2)))
assert list(chunks([], 3)) == []
```

### 35 · Best time to buy and sell stock once
**Problem:** Maximize profit from one buy followed by one sell; return 0 if no profit.

**Expected example:** `5`  
**Complexity:** O(n) time, O(1) extra space.

```python
def max_stock_profit(prices):
    if not prices:
        return 0
    lowest = prices[0]
    best = 0
    for price in prices[1:]:
        best = max(best, price - lowest)
        lowest = min(lowest, price)
    return best

print(max_stock_profit([7, 1, 5, 3, 6, 4]))
assert max_stock_profit([7, 6, 4]) == 0
```

```python
from collections import defaultdict

def group_by_first_letter(words):
    groups = defaultdict(list)
    for word in words:
        if word:
            groups[word[0].casefold()].append(word)
    return dict(groups)

print(group_by_first_letter(["apple", "ant", "bat"]))
assert group_by_first_letter([""]) == {}
```

### 43 · Safe nested dictionary lookup
**Problem:** Get a nested value using a sequence of keys; return default if missing.

**Expected example:** `Asha`  
**Complexity:** O(depth).

```python
def nested_get(data, keys, default=None):
    current = data
    for key in keys:
        if not isinstance(current, dict) or key not in current:
            return default
        current = current[key]
    return current

payload = {"user": {"profile": {"name": "Asha"}}}
print(nested_get(payload, ["user", "profile", "name"]))
assert nested_get(payload, ["user", "missing"], "N/A") == "N/A"
```

### 44 · Build an index by ID
**Problem:** Index records by unique `id`, rejecting duplicates.

**Expected example:** `{'u1': {'id': 'u1', 'score': 4}, 'u2': {'id': 'u2', 'score': 9}}`  
**Complexity:** O(n) average.

```python
def index_by_id(records):
    result = {}
    for record in records:
        key = record["id"]
        if key in result:
            raise ValueError(f"duplicate id: {key}")
        result[key] = record
    return result

records = [{"id": "u1", "score": 4}, {"id": "u2", "score": 9}]
print(index_by_id(records))
assert len(index_by_id(records)) == 2
```

### 48 · Two-way mapping
**Problem:** Build forward and reverse mappings for unique pairs.

**Expected example:** `({'en': 'English', 'fr': 'French'}, {'English': 'en', 'French': 'fr'})`  
**Complexity:** O(n) average.

```python
def build_bimap(pairs):
    forward, reverse = {}, {}
    for left, right in pairs:
        if left in forward or right in reverse:
            raise ValueError("mapping values must be unique on both sides")
        forward[left] = right
        reverse[right] = left
    return forward, reverse

print(build_bimap([("en", "English"), ("fr", "French")]))
assert build_bimap([]) == ({}, {})
```

### 49 · LRU cache
**Problem:** Implement a small least-recently-used cache.

**Expected example:** `None 1 3`  
**Complexity:** Average O(1). This example is not thread-safe or distributed.

```python
from collections import OrderedDict

class LRUCache:
    def __init__(self, capacity):
        if capacity < 1:
            raise ValueError("capacity must be positive")
        self.capacity = capacity
        self.data = OrderedDict()
    def get(self, key):
        if key not in self.data:
            return None
        self.data.move_to_end(key)
        return self.data[key]
    def put(self, key, value):
        if key in self.data:
            self.data.move_to_end(key)
        self.data[key] = value
        if len(self.data) > self.capacity:
            self.data.popitem(last=False)

cache = LRUCache(2)
cache.put("a", 1); cache.put("b", 2); cache.get("a"); cache.put("c", 3)
print(cache.get("b"), cache.get("a"), cache.get("c"))
assert cache.get("b") is None
```

### 50 · Validate a configuration dictionary
**Problem:** Validate required keys and types, including rejecting bool as an integer timeout.

**Expected example:** `True`  
**Complexity:** O(number of fields).

```python
def validate_config(config):
    required = {"timeout_seconds", "environment"}
    missing = required - config.keys()
    if missing:
        raise ValueError(f"missing keys: {sorted(missing)}")
    timeout = config["timeout_seconds"]
    if isinstance(timeout, bool) or not isinstance(timeout, (int, float)) or timeout <= 0:
        raise ValueError("timeout_seconds must be a positive number")
    if config["environment"] not in {"dev", "test", "prod"}:
        raise ValueError("invalid environment")
    return True

print(validate_config({"timeout_seconds": 2.5, "environment": "prod"}))
assert validate_config({"timeout_seconds": 1, "environment": "test"})
```

### 51 · Stack with list
**Problem:** Implement push, pop and peek with empty-stack protection.

**Expected example:** `20 10`  
**Complexity:** Push/pop at end are amortized O(1).

```python
class Stack:
    def __init__(self):
        self._items = []
    def push(self, item):
        self._items.append(item)
    def pop(self):
        if not self._items:
            raise IndexError("pop from empty stack")
        return self._items.pop()
    def peek(self):
        if not self._items:
            raise IndexError("peek from empty stack")
        return self._items[-1]

stack = Stack()
stack.push(10); stack.push(20)
print(stack.pop(), stack.peek())
assert stack.pop() == 10
```

### 53 · Balanced brackets
**Problem:** Validate correctly nested parentheses, brackets and braces.

**Expected example:** `True False`  
**Complexity:** O(n) time and space.

```python
def balanced_brackets(text):
    pairs = {')': '(', ']': '[', '}': '{'}
    stack = []
    for ch in text:
        if ch in "([{":
            stack.append(ch)
        elif ch in pairs:
            if not stack or stack.pop() != pairs[ch]:
                return False
    return not stack

print(balanced_brackets("{[()]}"), balanced_brackets("([)]"))
assert balanced_brackets("")
```

### 54 · Min stack
**Problem:** Support push, pop and minimum in O(1) time.

**Expected example:** `1 1 2`  
**Complexity:** O(1) per operation, O(n) space.

```python
class MinStack:
    def __init__(self):
        self.items, self.mins = [], []
    def push(self, value):
        self.items.append(value)
        if not self.mins or value <= self.mins[-1]:
            self.mins.append(value)
    def pop(self):
        value = self.items.pop()
        if value == self.mins[-1]:
            self.mins.pop()
        return value
    def minimum(self):
        if not self.mins:
            raise IndexError("empty stack")
        return self.mins[-1]

s = MinStack()
for x in [4, 2, 5, 1]: s.push(x)
print(s.minimum(), s.pop(), s.minimum())
assert s.minimum() == 2
```

### 55 · Evaluate Reverse Polish Notation
**Problem:** Evaluate integer tokens in postfix notation using +, -, *, /; division truncates toward zero.

**Expected example:** `9`  
**Complexity:** O(n) time and space.

```python
def eval_rpn(tokens):
    stack = []
    for token in tokens:
        if token in {"+", "-", "*", "/"}:
            b, a = stack.pop(), stack.pop()
            if token == "+": value = a + b
            elif token == "-": value = a - b
            elif token == "*": value = a * b
            else:
                if b == 0: raise ZeroDivisionError
                value = int(a / b)
            stack.append(value)
        else:
            stack.append(int(token))
    if len(stack) != 1:
        raise ValueError("invalid expression")
    return stack[0]

print(eval_rpn(["2", "1", "+", "3", "*"]))
assert eval_rpn(["4", "13", "5", "/", "+"]) == 6
```

### 56 · Sliding-window maximum
**Problem:** Return the maximum of every window of size k using a monotonic deque.

**Expected example:** `[3, 3, 5, 5, 6, 7]`  
**Complexity:** O(n) time, O(k) space.

```python
from collections import deque

def sliding_max(nums, k):
    if k <= 0 or k > len(nums):
        raise ValueError("k must be between 1 and len(nums)")
    q, result = deque(), []
    for i, value in enumerate(nums):
        while q and q[0] <= i - k:
            q.popleft()
        while q and nums[q[-1]] <= value:
            q.pop()
        q.append(i)
        if i >= k - 1:
            result.append(nums[q[0]])
    return result

print(sliding_max([1, 3, -1, -3, 5, 3, 6, 7], 3))
assert sliding_max([4, 2, 12], 1) == [4, 2, 12]
```

### 57 · Reverse a linked list
**Problem:** Reverse a singly linked list using pointer updates.

**Expected example:** `[3, 2, 1]`  
**Complexity:** O(n) time, O(1) extra space.

```python
class Node:
    def __init__(self, value, next=None):
        self.value, self.next = value, next

def reverse_linked_list(head):
    previous = None
    current = head
    while current:
        following = current.next
        current.next = previous
        previous, current = current, following
    return previous

def to_list(head):
    out = []
    while head:
        out.append(head.value); head = head.next
    return out

head = Node(1, Node(2, Node(3)))
print(to_list(reverse_linked_list(head)))
assert to_list(reverse_linked_list(None)) == []
```

```python
def has_cycle(head):
    slow = fast = head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        if slow is fast:
            return True
    return False

one, two, three = Node(1), Node(2), Node(3)
one.next, two.next, three.next = two, three, two
print(has_cycle(one))
assert not has_cycle(Node(1))
```

### 60 · Queue using two stacks
**Problem:** Implement amortized O(1) enqueue and dequeue.

**Expected example:** `1 2`  
**Complexity:** Amortized O(1) per operation.

```python
class TwoStackQueue:
    def __init__(self):
        self.inbox, self.outbox = [], []
    def enqueue(self, value):
        self.inbox.append(value)
    def dequeue(self):
        if not self.outbox:
            while self.inbox:
                self.outbox.append(self.inbox.pop())
        if not self.outbox:
            raise IndexError("dequeue from empty queue")
        return self.outbox.pop()

q = TwoStackQueue()
q.enqueue(1); q.enqueue(2)
print(q.dequeue(), q.dequeue())
assert q.inbox == [] and q.outbox == []
```

### 62 · Binary search
**Problem:** Search a sorted list iteratively.

**Expected example:** `3`  
**Complexity:** O(log n) time, O(1) space.

```python
def binary_search(values, target):
    left, right = 0, len(values) - 1
    while left <= right:
        mid = left + (right - left) // 2
        if values[mid] == target: return mid
        if values[mid] < target: left = mid + 1
        else: right = mid - 1
    return -1

print(binary_search([1, 3, 5, 7, 9], 7))
assert binary_search([], 2) == -1
```

### 63 · First occurrence in sorted list
**Problem:** Return the first index of target, or -1.

**Expected example:** `1`  
**Complexity:** O(log n) time.

```python
def first_occurrence(values, target):
    left, right, answer = 0, len(values) - 1, -1
    while left <= right:
        mid = (left + right) // 2
        if values[mid] >= target:
            if values[mid] == target: answer = mid
            right = mid - 1
        else:
            left = mid + 1
    return answer

print(first_occurrence([1, 2, 2, 2, 5], 2))
assert first_occurrence([1, 3], 2) == -1
```

### 64 · Lower bound insertion position
**Problem:** Find the first index where target could be inserted to keep sorted order.

**Expected example:** `1 2`  
**Complexity:** O(log n) time.

```python
def lower_bound(values, target):
    left, right = 0, len(values)
    while left < right:
        mid = (left + right) // 2
        if values[mid] < target: left = mid + 1
        else: right = mid
    return left

print(lower_bound([1, 3, 3, 7], 3), lower_bound([1, 3, 7], 5))
assert lower_bound([], 5) == 0
```

### 65 · Bubble sort
**Problem:** Implement bubble sort and stop early if already sorted.

**Expected example:** `[1, 2, 4, 5, 8]`  
**Complexity:** Worst-case O(n²), O(n) due to copy.

```python
def bubble_sort(values):
    arr = values[:]
    for end in range(len(arr) - 1, 0, -1):
        swapped = False
        for i in range(end):
            if arr[i] > arr[i + 1]:
                arr[i], arr[i + 1] = arr[i + 1], arr[i]
                swapped = True
        if not swapped: break
    return arr

print(bubble_sort([5, 1, 4, 2, 8]))
assert bubble_sort([]) == []
```

### 66 · Merge sort
**Problem:** Implement merge sort recursively.

**Expected example:** `[1, 2, 3, 5, 8, 9]`  
**Complexity:** O(n log n) time, O(n) auxiliary space.

```python
def merge_sort(values):
    if len(values) <= 1:
        return values[:]
    mid = len(values) // 2
    left, right = merge_sort(values[:mid]), merge_sort(values[mid:])
    out, i, j = [], 0, 0
    while i < len(left) and j < len(right):
        if left[i] <= right[j]: out.append(left[i]); i += 1
        else: out.append(right[j]); j += 1
    return out + left[i:] + right[j:]

print(merge_sort([8, 3, 5, 1, 9, 2]))
assert merge_sort([1, 1]) == [1, 1]
```

### 68 · Sort colors / three values
**Problem:** Sort a list containing only 0, 1 and 2 in place using Dutch national flag.

**Expected example:** `[0, 0, 1, 1, 2, 2]`  
**Complexity:** O(n) time, O(1) extra space.

```python
def sort_colors(nums):
    low = mid = 0
    high = len(nums) - 1
    while mid <= high:
        if nums[mid] == 0:
            nums[low], nums[mid] = nums[mid], nums[low]
            low += 1; mid += 1
        elif nums[mid] == 1:
            mid += 1
        elif nums[mid] == 2:
            nums[mid], nums[high] = nums[high], nums[mid]
            high -= 1
        else:
            raise ValueError("values must be 0, 1, or 2")
    return nums

print(sort_colors([2, 0, 2, 1, 1, 0]))
assert sort_colors([]) == []
```

### 69 · Merge overlapping intervals
**Problem:** Merge overlapping [start, end] intervals.

**Expected example:** `[[1, 6], [8, 10], [15, 18]]`  
**Complexity:** O(n log n) time.

```python
def merge_intervals(intervals):
    if not intervals:
        return []
    ordered = sorted(intervals)
    merged = [ordered[0][:]]
    for start, end in ordered[1:]:
        if start <= merged[-1][1]:
            merged[-1][1] = max(merged[-1][1], end)
        else:
            merged.append([start, end])
    return merged

print(merge_intervals([[1, 3], [2, 6], [8, 10], [15, 18]]))
assert merge_intervals([]) == []
```

### 73 · Longest substring without repeats
**Problem:** Return the length of the longest substring with unique characters.

**Expected example:** `3`  
**Complexity:** O(n) time, O(k) space.

```python
def longest_unique_substring(text):
    last_seen, left, best = {}, 0, 0
    for right, ch in enumerate(text):
        if ch in last_seen and last_seen[ch] >= left:
            left = last_seen[ch] + 1
        last_seen[ch] = right
        best = max(best, right - left + 1)
    return best

print(longest_unique_substring("abcabcbb"))
assert longest_unique_substring("") == 0
```

### 74 · Minimum-size subarray sum
**Problem:** For positive integers, return the shortest subarray length with sum at least target, or 0.

**Expected example:** `2`  
**Complexity:** O(n) time; assumes positive values.

```python
def min_subarray_len(target, nums):
    left, total, best = 0, 0, float("inf")
    for right, value in enumerate(nums):
        total += value
        while total >= target:
            best = min(best, right - left + 1)
            total -= nums[left]
            left += 1
    return 0 if best == float("inf") else best

print(min_subarray_len(7, [2, 3, 1, 2, 4, 3]))
assert min_subarray_len(100, [1, 2]) == 0
```

### 75 · Maximum average subarray
**Problem:** Find the maximum average among all windows of size k.

**Expected example:** `12.75`  
**Complexity:** O(n) time, O(1) extra space.

```python
def max_average(nums, k):
    if not 1 <= k <= len(nums):
        raise ValueError("invalid k")
    window = sum(nums[:k])
    best = window
    for i in range(k, len(nums)):
        window += nums[i] - nums[i - k]
        best = max(best, window)
    return best / k

print(max_average([1, 12, -5, -6, 50, 3], 4))
assert max_average([5], 1) == 5
```

### 77 · Count subarrays with sum k

### 78 · Container with most water

```python
def max_container_area(heights):
    left, right, best = 0, len(heights) - 1, 0
    while left < right:
        best = max(best, (right - left) * min(heights[left], heights[right]))
        if heights[left] < heights[right]: left += 1
        else: right -= 1
    return best

print(max_container_area([1, 8, 6, 2, 5, 4, 8, 3, 7]))
assert max_container_area([]) == 0
```

### 79 · 3Sum unique triplets
**Problem:** Return unique triplets summing to zero.

**Expected example:** `[[-1, -1, 2], [-1, 0, 1]]`  
**Complexity:** O(n²) time after sorting.

```python
def three_sum(nums):
    nums = sorted(nums)
    result = []
    for i in range(len(nums) - 2):
        if i and nums[i] == nums[i - 1]: continue
        left, right = i + 1, len(nums) - 1
        while left < right:
            total = nums[i] + nums[left] + nums[right]
            if total == 0:
                result.append([nums[i], nums[left], nums[right]])
                left += 1; right -= 1
                while left < right and nums[left] == nums[left - 1]: left += 1
                while left < right and nums[right] == nums[right + 1]: right -= 1
            elif total < 0: left += 1
            else: right -= 1
    return result

print(three_sum([-1, 0, 1, 2, -1, -4]))
assert three_sum([]) == []
```

### 80 · Smallest window containing all target characters
**Problem:** Return the shortest substring containing target characters with multiplicity, or empty string.

**Expected example:** `BANC`  
**Complexity:** O(n + m) time, O(k) space.

```python
from collections import Counter

def min_window(text, target):
    if not target: return ""
    need = Counter(target)
    missing = len(target)
    left = start = 0
    best = float("inf")
    for right, ch in enumerate(text, 1):
        if need[ch] > 0: missing -= 1
        need[ch] -= 1
        while missing == 0:
            if right - left < best:
                start, best = left, right - left
            need[text[left]] += 1
            if need[text[left]] > 0: missing += 1
            left += 1
    return "" if best == float("inf") else text[start:start + best]

print(min_window("ADOBECODEBANC", "ABC"))
assert min_window("abc", "") == ""
```

### 82 · Recursive binary search
**Problem:** Implement binary search recursively.

**Expected example:** `2`  
**Complexity:** O(log n) time and recursion space.

```python
def recursive_binary_search(nums, target, left=0, right=None):
    if right is None: right = len(nums) - 1
    if left > right: return -1
    mid = (left + right) // 2
    if nums[mid] == target: return mid
    if nums[mid] < target:
        return recursive_binary_search(nums, target, mid + 1, right)
    return recursive_binary_search(nums, target, left, mid - 1)

print(recursive_binary_search([1, 3, 5, 7], 5))
assert recursive_binary_search([], 1) == -1
```

### 85 · Combination sum
**Problem:** Find combinations summing to target; candidates are positive and may be reused.

**Expected example:** `[[2, 2, 3], [7]]`  
**Complexity:** Exponential in worst case.

```python
def combination_sum(candidates, target):
    candidates = sorted(set(candidates))
    if any(x <= 0 for x in candidates):
        raise ValueError("candidates must be positive")
    result = []
    def backtrack(start, remaining, path):
        if remaining == 0:
            result.append(path[:]); return
        for i in range(start, len(candidates)):
            value = candidates[i]
            if value > remaining: break
            path.append(value)
            backtrack(i, remaining - value, path)
            path.pop()
    backtrack(0, target, [])
    return result

print(combination_sum([2, 3, 6, 7], 7))
assert combination_sum([2], 1) == []
```

### 86 · N-Queens count
**Problem:** Count solutions to the N-Queens problem using sets for occupied columns and diagonals.

**Expected example:** `2`  
**Complexity:** Backtracking; exponential worst case.

```python
def n_queens_count(n):
    cols, diag1, diag2 = set(), set(), set()
    count = 0
    def place(row):
        nonlocal count
        if row == n:
            count += 1; return
        for col in range(n):
            if col in cols or row - col in diag1 or row + col in diag2:
                continue
            cols.add(col); diag1.add(row - col); diag2.add(row + col)
            place(row + 1)
            cols.remove(col); diag1.remove(row - col); diag2.remove(row + col)
    place(0)
    return count

print(n_queens_count(4))
assert n_queens_count(1) == 1
```

### 87 · Word search in a grid
**Problem:** Return whether a word exists by adjacent horizontal/vertical cells without cell reuse.

**Expected example:** `True`  
**Complexity:** O(rows·cols·4^L) upper bound; board is restored after search.

```python
def word_exists(board, word):
    if not word: return True
    if not board or not board[0]: return False
    rows, cols = len(board), len(board[0])
    def dfs(r, c, i):
        if i == len(word): return True
        if r < 0 or c < 0 or r >= rows or c >= cols or board[r][c] != word[i]:
            return False
        ch = board[r][c]
        board[r][c] = None
        found = (dfs(r+1,c,i+1) or dfs(r-1,c,i+1) or
                 dfs(r,c+1,i+1) or dfs(r,c-1,i+1))
        board[r][c] = ch
        return found
    return any(dfs(r, c, 0) for r in range(rows) for c in range(cols))

grid = [list("ABCE"), list("SFCS"), list("ADEE")]
print(word_exists(grid, "ABCCED"))
assert word_exists(grid, "SEE")
```

### 88 · Parentheses generation
**Problem:** Generate all balanced strings with n pairs of parentheses.

**Expected example:** `['((()))', '(()())', '(())()', '()(())', '()()()']`  
**Complexity:** Catalan-number output growth.

```python
def generate_parentheses(n):
    result = []
    def build(path, opened, closed):
        if len(path) == 2 * n:
            result.append(path); return
        if opened < n: build(path + "(", opened + 1, closed)
        if closed < opened: build(path + ")", opened, closed + 1)
    build("", 0, 0)
    return result

print(generate_parentheses(3))
assert generate_parentheses(0) == [""]
```

### 89 · Maze path existence
**Problem:** Find if a path exists from top-left to bottom-right; 0 is open, 1 is blocked.

**Expected example:** `True`  
**Complexity:** O(rows·cols) states.

```python
def maze_path_exists(maze):
    if not maze or not maze[0]: return False
    rows, cols = len(maze), len(maze[0])
    seen = set()
    def dfs(r, c):
        if not (0 <= r < rows and 0 <= c < cols): return False
        if maze[r][c] != 0 or (r, c) in seen: return False
        if (r, c) == (rows - 1, cols - 1): return True
        seen.add((r, c))
        return dfs(r+1,c) or dfs(r-1,c) or dfs(r,c+1) or dfs(r,c-1)
    return dfs(0, 0)

print(maze_path_exists([[0, 1, 0], [0, 0, 0], [1, 1, 0]]))
assert not maze_path_exists([[0, 1], [1, 0]])
```

### 90 · Tower of Hanoi moves
**Problem:** Return the sequence of moves for n disks.

**Expected example:** `[('A', 'B'), ('A', 'C'), ('B', 'C')]`  
**Complexity:** 2ⁿ - 1 moves.

```python
def hanoi(n, source="A", auxiliary="B", target="C"):
    if n < 0: raise ValueError("n must be non-negative")
    moves = []
    def solve(k, src, aux, dst):
        if k == 0: return
        solve(k - 1, src, dst, aux)
        moves.append((src, dst))
        solve(k - 1, aux, src, dst)
    solve(n, source, auxiliary, target)
    return moves

print(hanoi(2))
assert len(hanoi(3)) == 7
```

## 9. Trees, graphs and heaps (91–105)

### 91 · Binary tree preorder traversal
**Problem:** Return root-left-right values.

**Expected example:** `[1, 2, 3]`  
**Complexity:** O(n) time; O(h) call stack.

```python
class TreeNode:
    def __init__(self, val, left=None, right=None):
        self.val, self.left, self.right = val, left, right

def preorder(root):
    if root is None: return []
    return [root.val] + preorder(root.left) + preorder(root.right)

root = TreeNode(1, TreeNode(2), TreeNode(3))
print(preorder(root))
assert preorder(None) == []
```

### 92 · Binary tree inorder traversal
**Problem:** Return left-root-right values.

**Expected example:** `[1, 2, 3]`  
**Complexity:** O(n) time.

```python
def inorder(root):
    if root is None: return []
    return inorder(root.left) + [root.val] + inorder(root.right)

root = TreeNode(2, TreeNode(1), TreeNode(3))
print(inorder(root))
assert inorder(None) == []
```

### 93 · Tree level-order traversal
**Problem:** Return values grouped by depth.

**Expected example:** `[[1], [2, 3], [4]]`  
**Complexity:** O(n) time and space.

```python
from collections import deque

def level_order(root):
    if not root: return []
    queue, levels = deque([root]), []
    while queue:
        level = []
        for _ in range(len(queue)):
            node = queue.popleft()
            level.append(node.val)
            if node.left: queue.append(node.left)
            if node.right: queue.append(node.right)
        levels.append(level)
    return levels

root = TreeNode(1, TreeNode(2, TreeNode(4)), TreeNode(3))
print(level_order(root))
assert level_order(None) == []
```

### 94 · Tree height
**Problem:** Return height measured in number of nodes on the longest root-leaf path.

**Expected example:** `3`  
**Complexity:** O(n) time, O(h) recursion stack.

```python
def tree_height(root):
    if root is None: return 0
    return 1 + max(tree_height(root.left), tree_height(root.right))

root = TreeNode(1, TreeNode(2), TreeNode(3, TreeNode(4)))
print(tree_height(root))
assert tree_height(None) == 0
```

### 95 · Validate binary search tree
**Problem:** Validate strict BST ordering using lower and upper bounds.

**Expected example:** `True`  
**Complexity:** O(n) time, O(h) stack.

```python
def is_valid_bst(root):
    def check(node, low, high):
        if node is None: return True
        if not low < node.val < high: return False
        return check(node.left, low, node.val) and check(node.right, node.val, high)
    return check(root, float("-inf"), float("inf"))

root = TreeNode(2, TreeNode(1), TreeNode(3))
print(is_valid_bst(root))
assert not is_valid_bst(TreeNode(5, TreeNode(1), TreeNode(4)))
```

### 96 · Lowest common ancestor in binary tree
**Problem:** Find LCA assuming both target values exist and values are unique.

**Expected example:** `5`  
**Complexity:** O(n) time, O(h) stack.

```python
def lowest_common_ancestor(root, p, q):
    if root is None or root.val in (p, q): return root
    left = lowest_common_ancestor(root.left, p, q)
    right = lowest_common_ancestor(root.right, p, q)
    if left and right: return root
    return left or right

root = TreeNode(3, TreeNode(5, TreeNode(6), TreeNode(2)), TreeNode(1))
print(lowest_common_ancestor(root, 6, 2).val)
assert lowest_common_ancestor(root, 6, 2).val == 5
```

### 97 · Graph BFS
**Problem:** Traverse a graph from a start node.

**Expected example:** `['A', 'B', 'C', 'D']`  
**Complexity:** O(V + E).

```python
from collections import deque

def bfs(graph, start):
    seen, order = {start}, []
    queue = deque([start])
    while queue:
        node = queue.popleft()
        order.append(node)
        for neighbor in graph.get(node, []):
            if neighbor not in seen:
                seen.add(neighbor); queue.append(neighbor)
    return order

graph = {"A": ["B", "C"], "B": ["D"], "C": ["D"], "D": []}
print(bfs(graph, "A"))
assert bfs({}, "X") == ["X"]
```

### 98 · Graph DFS
**Problem:** Traverse a graph iteratively using depth-first search.

**Expected example:** `['A', 'B', 'D', 'C']`  
**Complexity:** O(V + E).

```python
def dfs(graph, start):
    seen, order, stack = set(), [], [start]
    while stack:
        node = stack.pop()
        if node in seen: continue
        seen.add(node); order.append(node)
        stack.extend(reversed(graph.get(node, [])))
    return order

graph = {"A": ["B", "C"], "B": ["D"], "C": [], "D": []}
print(dfs(graph, "A"))
assert dfs({}, "Z") == ["Z"]
```

### 99 · Detect cycle in directed graph
**Problem:** Use three-state DFS: unseen, visiting, complete.

**Expected example:** `True`  
**Complexity:** O(V + E); recursive depth can be an issue on very deep graphs.

```python
def has_directed_cycle(graph):
    state = {}
    def visit(node):
        if state.get(node) == 1: return True
        if state.get(node) == 2: return False
        state[node] = 1
        for neighbor in graph.get(node, []):
            if visit(neighbor): return True
        state[node] = 2
        return False
    nodes = set(graph)
    for neighbors in graph.values(): nodes.update(neighbors)
    return any(visit(node) for node in nodes if state.get(node, 0) == 0)

print(has_directed_cycle({"A": ["B"], "B": ["C"], "C": ["A"]}))
assert not has_directed_cycle({"A": ["B"], "B": []})
```

### 100 · Topological sort
**Problem:** Return a topological ordering using Kahn's algorithm; reject cycles.

**Expected example:** `One valid order: ['shop', 'cook', 'eat']`  
**Complexity:** O(V + E); exact valid order may vary with node iteration order.

```python
from collections import deque

def topological_sort(graph):
    nodes = set(graph)
    for neighbors in graph.values(): nodes.update(neighbors)
    indegree = {node: 0 for node in nodes}
    for node, neighbors in graph.items():
        for neighbor in neighbors: indegree[neighbor] += 1
    queue = deque(node for node in nodes if indegree[node] == 0)
    result = []
    while queue:
        node = queue.popleft(); result.append(node)
        for neighbor in graph.get(node, []):
            indegree[neighbor] -= 1
            if indegree[neighbor] == 0: queue.append(neighbor)
    if len(result) != len(nodes):
        raise ValueError("graph contains a cycle")
    return result

order = topological_sort({"cook": ["eat"], "shop": ["cook"], "eat": []})
print(order)
assert order.index("shop") < order.index("cook") < order.index("eat")
```

### 101 · Shortest path in unweighted graph
**Problem:** Return the shortest path between start and target using BFS.

**Expected example:** `['A', 'B', 'D'] or ['A', 'C', 'D']`  
**Complexity:** O(V + E).

```python
from collections import deque

def shortest_path(graph, start, target):
    queue = deque([start])
    parent = {start: None}
    while queue:
        node = queue.popleft()
        if node == target: break
        for neighbor in graph.get(node, []):
            if neighbor not in parent:
                parent[neighbor] = node
                queue.append(neighbor)
    if target not in parent: return None
    path, node = [], target
    while node is not None:
        path.append(node); node = parent[node]
    return path[::-1]

graph = {"A": ["B", "C"], "B": ["D"], "C": ["D"], "D": []}
print(shortest_path(graph, "A", "D"))
assert shortest_path(graph, "X", "Y") is None
```

### 102 · Dijkstra shortest distances
**Problem:** Find shortest distances in a graph with non-negative edge weights.

**Expected example:** `{'A': 0, 'C': 1, 'B': 3}`  
**Complexity:** O((V + E) log V) with a binary heap.

```python
import heapq

def dijkstra(graph, start):
    distances = {start: 0}
    heap = [(0, start)]
    while heap:
        distance, node = heapq.heappop(heap)
        if distance != distances[node]: continue
        for neighbor, weight in graph.get(node, []):
            if weight < 0: raise ValueError("negative weights are unsupported")
            candidate = distance + weight
            if candidate < distances.get(neighbor, float("inf")):
                distances[neighbor] = candidate
                heapq.heappush(heap, (candidate, neighbor))
    return distances

graph = {"A": [("B", 4), ("C", 1)], "C": [("B", 2)], "B": []}
print(dijkstra(graph, "A"))
assert dijkstra({"A": []}, "A")["A"] == 0
```

### 103 · Connected components in undirected graph
**Problem:** Count connected components, including isolated vertices.

**Expected example:** `2`  
**Complexity:** O(V + E).

```python
def connected_components(n, edges):
    graph = {i: [] for i in range(n)}
    for a, b in edges:
        graph[a].append(b); graph[b].append(a)
    seen, count = set(), 0
    for node in range(n):
        if node in seen: continue
        count += 1
        stack = [node]; seen.add(node)
        while stack:
            current = stack.pop()
            for neighbor in graph[current]:
                if neighbor not in seen:
                    seen.add(neighbor); stack.append(neighbor)
    return count

print(connected_components(5, [(0, 1), (1, 2), (3, 4)]))
assert connected_components(0, []) == 0
```

### 104 · Kth-largest using a heap
**Problem:** Find kth-largest with a min-heap of size k.

**Expected example:** `5`  
**Complexity:** O(n log k) time, O(k) space.

```python
import heapq

def kth_largest_heap(nums, k):
    if not 1 <= k <= len(nums): raise ValueError("k out of range")
    heap = []
    for value in nums:
        heapq.heappush(heap, value)
        if len(heap) > k: heapq.heappop(heap)
    return heap[0]

print(kth_largest_heap([3, 2, 1, 5, 6, 4], 2))
assert kth_largest_heap([1], 1) == 1
```

### 105 · Merge k sorted lists
**Problem:** Merge sorted lists using a min-heap.

**Expected example:** `[1, 2, 3, 4, 5, 6, 7, 9]`  
**Complexity:** O(N log k), N total items and k lists.

```python
import heapq

def merge_k_sorted(lists):
    heap = []
    for list_index, values in enumerate(lists):
        if values:
            heapq.heappush(heap, (values[0], list_index, 0))
    result = []
    while heap:
        value, list_index, index = heapq.heappop(heap)
        result.append(value)
        next_index = index + 1
        if next_index < len(lists[list_index]):
            heapq.heappush(heap, (lists[list_index][next_index], list_index, next_index))
    return result

print(merge_k_sorted([[1, 4, 7], [2, 5], [3, 6, 9]]))
assert merge_k_sorted([]) == []
```

## 10. Dynamic programming and greedy algorithms (106–115)

### 106 · Climbing stairs
**Problem:** Count ways to climb n steps taking 1 or 2 steps.

**Expected example:** `8`  
**Complexity:** O(n) time, O(1) space.

```python
def climb_stairs(n):
    if n < 0: return 0
    a, b = 1, 1
    for _ in range(n):
        a, b = b, a + b
    return a

print(climb_stairs(5))
assert climb_stairs(0) == 1
```

### 107 · Coin change minimum coins
**Problem:** Return minimum number of coins to make amount, or -1 if impossible.

**Expected example:** `3`  
**Complexity:** O(amount × number_of_coins) time, O(amount) space.

```python
def coin_change(coins, amount):
    if amount < 0 or any(c <= 0 for c in coins):
        raise ValueError("amount must be non-negative and coins positive")
    dp = [amount + 1] * (amount + 1)
    dp[0] = 0
    for value in range(1, amount + 1):
        for coin in coins:
            if coin <= value:
                dp[value] = min(dp[value], dp[value - coin] + 1)
    return dp[amount] if dp[amount] <= amount else -1

print(coin_change([1, 2, 5], 11))
assert coin_change([2], 3) == -1
```

### 108 · 0/1 knapsack
**Problem:** Maximize value with capacity; each item can be used at most once.

**Expected example:** `9`  
**Complexity:** O(n × capacity) time, O(capacity) space.

```python
def knapsack(weights, values, capacity):
    if len(weights) != len(values) or capacity < 0:
        raise ValueError("invalid input")
    dp = [0] * (capacity + 1)
    for weight, value in zip(weights, values):
        if weight <= 0: raise ValueError("weights must be positive")
        for cap in range(capacity, weight - 1, -1):
            dp[cap] = max(dp[cap], dp[cap - weight] + value)
    return dp[capacity]

print(knapsack([2, 3, 4], [4, 5, 6], 5))
assert knapsack([], [], 0) == 0
```

### 109 · Longest increasing subsequence
**Problem:** Return the length of a strictly increasing subsequence in O(n log n).

**Expected example:** `4`  
**Complexity:** O(n log n) time, O(n) space.

```python
from bisect import bisect_left

def lis_length(nums):
    tails = []
    for value in nums:
        index = bisect_left(tails, value)
        if index == len(tails): tails.append(value)
        else: tails[index] = value
    return len(tails)

print(lis_length([10, 9, 2, 5, 3, 7, 101, 18]))
assert lis_length([]) == 0
```

### 110 · Longest common subsequence
**Problem:** Return LCS length for two strings.

**Expected example:** `3`  
**Complexity:** O(nm) time, O(m) space.

```python
def lcs_length(a, b):
    previous = [0] * (len(b) + 1)
    for char_a in a:
        current = [0]
        for j, char_b in enumerate(b, 1):
            if char_a == char_b:
                current.append(previous[j - 1] + 1)
            else:
                current.append(max(previous[j], current[-1]))
        previous = current
    return previous[-1]

print(lcs_length("abcde", "ace"))
assert lcs_length("", "abc") == 0
```

### 111 · Edit distance
**Problem:** Compute Levenshtein distance with insert, delete and replace operations.

**Expected example:** `3`  
**Complexity:** O(nm) time, O(m) space.

```python
def edit_distance(a, b):
    previous = list(range(len(b) + 1))
    for i, ca in enumerate(a, 1):
        current = [i]
        for j, cb in enumerate(b, 1):
            cost = 0 if ca == cb else 1
            current.append(min(current[-1] + 1, previous[j] + 1, previous[j - 1] + cost))
        previous = current
    return previous[-1]

print(edit_distance("kitten", "sitting"))
assert edit_distance("", "abc") == 3
```

### 112 · House robber
**Problem:** Maximize sum of non-adjacent values.

**Expected example:** `12`  
**Complexity:** O(n) time, O(1) space.

```python
def house_robber(nums):
    take = skip = 0
    for value in nums:
        take, skip = skip + value, max(take, skip)
    return max(take, skip)

print(house_robber([2, 7, 9, 3, 1]))
assert house_robber([]) == 0
```

### 113 · Activity selection
**Problem:** Select the maximum number of non-overlapping activities represented by (start, end).

**Expected example:** `[(1, 3), (3, 5), (5, 9)]`  
**Complexity:** O(n log n) due to sorting.

```python
def select_activities(activities):
    ordered = sorted(activities, key=lambda item: item[1])
    selected, end_time = [], float("-inf")
    for start, end in ordered:
        if start >= end_time:
            selected.append((start, end))
            end_time = end
    return selected

print(select_activities([(1, 3), (2, 4), (3, 5), (0, 7), (5, 9)]))
assert select_activities([]) == []
```

### 114 · Jump game
**Problem:** Each value is the maximum jump length; determine if the last index is reachable.

**Expected example:** `True False`  
**Complexity:** O(n) time, O(1) space.

```python
def can_jump(nums):
    farthest = 0
    for i, jump in enumerate(nums):
        if i > farthest: return False
        farthest = max(farthest, i + jump)
        if farthest >= len(nums) - 1: return True
    return not nums

print(can_jump([2, 3, 1, 1, 4]), can_jump([3, 2, 1, 0, 4]))
assert can_jump([])
```

### 115 · Minimum number of meeting rooms
**Problem:** Find minimum rooms needed for half-open intervals [start, end).

**Expected example:** `2`  
**Complexity:** O(n log n) time.

```python
import heapq

def min_meeting_rooms(intervals):
    if not intervals: return 0
    intervals = sorted(intervals)
    active_ends = []
    for start, end in intervals:
        if active_ends and active_ends[0] <= start:
            heapq.heappop(active_ends)
        heapq.heappush(active_ends, end)
    return len(active_ends)

print(min_meeting_rooms([(0, 30), (5, 10), (15, 20)]))
assert min_meeting_rooms([]) == 0
```

### 116 · Validate incoming records
**Problem:** Validate a list of records; return valid records and count rejected ones.

**Expected example:** `([{'user_id': 'u1', 'score': 0.8}], 1)`  
**Complexity / production note:** O(n). In production add schema versioning, metrics, and quarantine.

```python
def validate_records(rows):
    valid, rejected = [], 0
    for row in rows:
        user_id, score = row.get("user_id"), row.get("score")
        if not isinstance(user_id, str) or not user_id.strip():
            rejected += 1; continue
        if isinstance(score, bool) or not isinstance(score, (int, float)):
            rejected += 1; continue
        if not 0 <= score <= 1:
            rejected += 1; continue
        valid.append({"user_id": user_id.strip(), "score": float(score)})
    return valid, rejected

print(validate_records([{"user_id": "u1", "score": 0.8}, {"user_id": "", "score": 0.2}]))
assert validate_records([]) == ([], 0)
```

### 117 · Retry a transient operation
**Problem:** Retry only selected transient exceptions with bounded attempts.

**Expected example:** `ok 3`  
**Complexity / production note:** Add exponential backoff, jitter, deadline, and idempotency in production.

```python
def retry_operation(operation, attempts=3):
    if attempts < 1: raise ValueError("attempts must be positive")
    for attempt in range(attempts):
        try:
            return operation()
        except (TimeoutError, ConnectionError):
            if attempt == attempts - 1:
                raise

calls = {"count": 0}
def flaky():
    calls["count"] += 1
    if calls["count"] < 3: raise TimeoutError("temporary")
    return "ok"

print(retry_operation(flaky), calls["count"])
assert calls["count"] == 3
```

### 118 · Idempotency key store
**Problem:** Demonstrate deduplicating repeated requests in one process.

**Expected example:** `{'job_id': 1} twice`  
**Complexity / production note:** Demo only: production needs durable atomic storage, concurrency safety and retention policy.

```python
class IdempotencyStore:
    def __init__(self):
        self.results = {}
    def run_once(self, key, operation):
        if key in self.results:
            return self.results[key]
        result = operation()
        self.results[key] = result
        return result

store = IdempotencyStore()
counter = {"n": 0}
def create_job():
    counter["n"] += 1
    return {"job_id": counter["n"]}

print(store.run_once("request-1", create_job))
print(store.run_once("request-1", create_job))
assert counter["n"] == 1
```

### 119 · Process records in batches
**Problem:** Yield fixed-size batches from an iterable without loading everything at once.

**Expected example:** `[[0, 1, 2], [3, 4, 5], [6]]`  
**Complexity / production note:** O(n) time, O(size) extra memory.

```python
from itertools import islice

def batched(iterable, size):
    if size <= 0: raise ValueError("size must be positive")
    iterator = iter(iterable)
    while batch := list(islice(iterator, size)):
        yield batch

print(list(batched(range(7), 3)))
assert list(batched([], 2)) == []
```

### 120 · Top-level timing decorator
**Problem:** Create a decorator that prints a function's elapsed time while preserving metadata.

**Expected example:** `Prints a timing line and result: 499500`  
**Complexity / production note:** Timing output varies. Production systems should usually use structured metrics/tracing.

```python
from functools import wraps
from time import perf_counter

def timed(func):
    @wraps(func)
    def wrapper(*args, **kwargs):
        start = perf_counter()
        try:
            return func(*args, **kwargs)
        finally:
            elapsed = perf_counter() - start
            print(f"{func.__name__} took {elapsed:.6f}s")
    return wrapper

@timed
def compute_total(values):
    return sum(values)

print("result:", compute_total(range(1000)))
assert compute_total.__name__ == "compute_total"
```
