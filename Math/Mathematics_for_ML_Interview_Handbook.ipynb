## 1. Mathematical notation and prerequisites

| Symbol | Meaning | Common ML use |
|---|---|---|
| $x, y$ | Scalar values | One feature and target |
| $\mathbf{x}$ | Vector | One observation or parameter vector |
| $X$ | Matrix / design matrix | Dataset: rows are samples, columns are features |
| $X^\top$ | Transpose | Covariance, least squares, gradients |
| $w, \theta$ | Model parameters | Linear/logistic regression, neural networks |
| $\hat y$ | Prediction | Model output |
| $\mathbb{E}[X]$ | Expected value | Mean under a probability distribution |
| $\mathrm{Var}(X)$ | Variance | Spread, uncertainty, standardization |
| $\nabla_\theta L$ | Gradient of loss | Gradient descent and backpropagation |
| $P(A\mid B)$ | Conditional probability | Bayesian models, diagnosis, classification |
| $\lambda$ | Regularization strength or eigenvalue depending on context | Ridge/Lasso; spectral methods |
| $n, d$ | Samples and features | Complexity, statistical estimation |

### Scalars, vectors, matrices, and tensors
- **Scalar:** one number, such as a learning rate.
- **Vector:** ordered numbers, such as model weights.
- **Matrix:** a rectangular array, such as a feature matrix.
- **Tensor:** a generalized multi-dimensional array, such as a batch of images.

Shape discipline matters: if $X \in \mathbb{R}^{n \times d}$, there are $n$ samples and $d$ features; for $w \in \mathbb{R}^{d}$, $Xw \in \mathbb{R}^{n}$.

A function maps inputs to outputs, \(f:\mathcal{X}	o\mathcal{Y}\). A model is a parameterized function \(f_	heta(x)\); training chooses \(	heta\) so predictions match observed targets. Linear regression uses \(f(x)=w^	op x+b\); a neural network composes many functions.

<details>
<summary><strong>Q2. What is a linear function versus an affine function?</strong></summary>

**Answer / theory**

A linear map satisfies additivity and homogeneity and passes through the origin. \(f(x)=wx\) is linear; \(f(x)=wx+b\) is affine when \(b
e0\). In ML, the intercept makes a model affine, even though it is often casually called linear regression.

**When and where it is used:** Linear regression, dense layers, feature maps.

</details>

<details>
<summary><strong>Q3. Why are logarithms useful in machine learning?</strong></summary>

**Answer / theory**

Logs turn products into sums: \(\log(ab)=\log a+\log b\), make tiny probabilities numerically manageable, and convert likelihood products into log-likelihood sums. Log loss heavily penalizes confident wrong predictions.

**When and where it is used:** Maximum likelihood, cross-entropy, Naive Bayes, perplexity.

</details>

<details>
<summary><strong>Q4. What is an exponential function and where is it used?</strong></summary>

**Answer / theory**

The exponential \(e^x\) grows rapidly and is used to turn log scores into positive quantities. Softmax uses exponentials to create normalized class probabilities; exponential-family distributions use it in probability density functions.

**When and where it is used:** Softmax, Poisson models, exponential-family distributions.

</details>

<details>
<summary><strong>Q5. What is a derivative intuitively?</strong></summary>

**Answer / theory**

A derivative is the instantaneous rate of change. \(f'(x)=\lim_{h	o0}[f(x+h)-f(x)]/h\). It tells how a small parameter change affects a function and is the foundation of gradient-based learning.

**When and where it is used:** Gradient descent, sensitivity analysis, backpropagation.

</details>

<details>
<summary><strong>Q6. What is a logarithmic scale and why can it help?</strong></summary>

**Answer / theory**

A log scale displays multiplicative changes more naturally. A move from 1 to 10 is the same log-distance as 10 to 100. It is useful when values span orders of magnitude, such as loss, latency, or model parameter counts.

**When and where it is used:** Learning curves, metrics, heavy-tailed data visualization.

</details>

<details>
<summary><strong>Q7. What is Euclidean distance?</strong></summary>

**Answer / theory**

For vectors \(x,y\), \(d_2(x,y)=\sqrt{\sum_i(x_i-y_i)^2}\). It measures straight-line distance and is sensitive to feature scales; standardize features when units differ substantially.

**When and where it is used:** KNN, K-means, clustering, nearest-neighbor retrieval.

</details>

<details>
<summary><strong>Q8. What is Manhattan distance?</strong></summary>

**Answer / theory**

\(d_1(x,y)=\sum_i|x_i-y_i|\). It measures axis-aligned path length and can be more robust to individual coordinate differences than squared Euclidean distance, though it still depends on feature scaling.

**When and where it is used:** KNN, grid-like spaces, robust distance comparisons.

</details>

<details>
<summary><strong>Q9. What is cosine similarity?</strong></summary>

**Answer / theory**

\(\cos(x,y)=x^	op y/(\|x\|_2\|y\|_2)\). It measures angle rather than magnitude. It is undefined for zero vectors, which must be handled explicitly.

**When and where it is used:** Text embeddings, semantic search, recommendation.

</details>

<details>
<summary><strong>Q10. What does dimensionality mean?</strong></summary>

**Answer / theory**

A vector's dimension is its number of components; a dataset's feature dimension is usually \(d\). Increasing \(d\) can increase compute and data needs and make distance-based methods less discriminative: this is part of the curse of dimensionality.

**When and where it is used:** Feature engineering, KNN, PCA, high-dimensional modeling.

</details>

## 3. Linear algebra

Read the core idea first, then open each question to test recall. Answers explain the concept, formula, algorithm connection, and practical caveat.

<details>
<summary><strong>Q11. What is a vector and what do its dimensions mean in ML?</strong></summary>

**Answer / theory**

A vector is an ordered collection of numbers. A feature vector often represents one example; a parameter vector stores model weights. The order must match the feature schema, or predictions can silently become wrong.

**When and where it is used:** Linear/logistic regression, neural network parameters.

</details>

<details>
<summary><strong>Q12. What is a dot product?</strong></summary>

**Answer / theory**

For \(x,y\in\mathbb{R}^d\), \(x^\top y=\sum_i x_i y_i\). It measures alignment and computes weighted sums. A linear model predicts \(\hat y=x^\top w+b\).

**When and where it is used:** Linear layers, attention scores, similarity, linear regression.

</details>

<details>
<summary><strong>Q13. What is a matrix transpose?</strong></summary>

**Answer / theory**

Transpose swaps rows and columns: \((X^\top)_{ij}=X_{ji}\). If \(X\) is \(n\times d\), then \(X^\top\) is \(d\times n\). It appears in covariance and least-squares equations.

**When and where it is used:** Least squares, covariance, gradient derivations.

</details>

<details>
<summary><strong>Q14. When is matrix multiplication valid?</strong></summary>

**Answer / theory**

If \(A\) has shape \(m\times k\) and \(B\) has shape \(k\times n\), then \(AB\) has shape \(m\times n\). The inner dimensions must match. Matrix multiplication is not generally commutative: \(AB\ne BA\).

**When and where it is used:** Dense layers, attention, regression, transformations.

</details>

<details>
<summary><strong>Q15. What is the identity matrix?</strong></summary>

**Answer / theory**

The identity matrix \(I\) has ones on its diagonal and zeros elsewhere, and \(AI=IA=A\) when dimensions are compatible. It is used in linear systems and regularization expressions such as \(X^\top X+\lambda I\).

**When and where it is used:** Ridge regression, matrix algebra, optimization.

</details>

<details>
<summary><strong>Q16. What is matrix rank?</strong></summary>

**Answer / theory**

Rank is the number of linearly independent rows or columns. A rank-deficient design matrix has redundant features and can make the ordinary least-squares solution non-unique or numerically unstable.

**When and where it is used:** Linear regression diagnostics, dimensionality reduction.

</details>

<details>
<summary><strong>Q17. What is linear independence?</strong></summary>

**Answer / theory**

Vectors are linearly independent if no nonzero coefficient combination sums to zero. Redundant columns in a design matrix create multicollinearity and can make individual coefficients hard to identify.

**When and where it is used:** Regression, feature selection, basis construction.

</details>

<details>
<summary><strong>Q18. What is a matrix inverse, and why not always compute it?</strong></summary>

**Answer / theory**

An inverse \(A^{-1}\) satisfies \(A^{-1}A=I\) and exists only for square nonsingular matrices. In numerical code, solve \(Ax=b\) with a linear solver rather than explicitly computing \(A^{-1}b\): it is typically more stable and efficient.

**When and where it is used:** Least squares, linear systems, numerical optimization.

</details>

<details>
<summary><strong>Q19. What are eigenvalues and eigenvectors?</strong></summary>

**Answer / theory**

For square \(A\), \(Av=\lambda v\) means direction \(v\) is scaled by eigenvalue \(\lambda\). Eigenvectors expose invariant directions; eigenvalues quantify scaling along them.

**When and where it is used:** PCA, spectral clustering, stability analysis.

</details>

<details>
<summary><strong>Q20. What is singular value decomposition (SVD)?</strong></summary>

**Answer / theory**

Any real matrix can be decomposed as \(X=U\Sigma V^\top\). Singular values describe the strength of orthogonal directions. SVD is robust for low-rank approximation and is often preferable to eigendecomposition of \(X^\top X\).

**When and where it is used:** PCA, compression, denoising, least squares.

</details>

<details>
<summary><strong>Q21. What is positive semidefinite (PSD)?</strong></summary>

**Answer / theory**

A symmetric matrix \(A\) is PSD when \(z^\top A z\ge0\) for all \(z\); equivalently, its eigenvalues are nonnegative. Covariance matrices and many kernel Gram matrices are PSD.

**When and where it is used:** Covariance, kernel methods, convex quadratic objectives.

</details>

<details>
<summary><strong>Q22. What is a covariance matrix?</strong></summary>

**Answer / theory**

For centered data \(X_c\), sample covariance is \(S=X_c^\top X_c/(n-1)\). Diagonal entries are variances; off-diagonal entries are pairwise covariances. Its shape is \(d\times d\).

**When and where it is used:** PCA, portfolio risk, feature dependence.

</details>

<details>
<summary><strong>Q23. What is an orthogonal matrix?</strong></summary>

**Answer / theory**

A square matrix \(Q\) is orthogonal if \(Q^\top Q=I\). Its columns are orthonormal, and it preserves Euclidean lengths and angles. Orthogonal transformations are useful for stable coordinate changes.

**When and where it is used:** QR decomposition, rotations, numerical linear algebra.

</details>

<details>
<summary><strong>Q24. What is a norm?</strong></summary>

**Answer / theory**

A norm measures vector magnitude. \(L_1=\sum_i|x_i|\), \(L_2=\sqrt{\sum_i x_i^2}\), and \(L_\infty=\max_i|x_i|\). Different norms induce different notions of size and regularization.

**When and where it is used:** Regularization, distances, gradient clipping.

</details>

<details>
<summary><strong>Q25. What is a condition number?</strong></summary>

**Answer / theory**

For a nonsingular matrix, \(\kappa(A)=\|A\|\|A^{-1}\|\). A large condition number means small input or rounding errors can produce large solution errors. Correlated features often make regression ill-conditioned.

**When and where it is used:** Numerical stability, regression, optimization.

</details>

## 4. Calculus and optimization

Read the core idea first, then open each question to test recall. Answers explain the concept, formula, algorithm connection, and practical caveat.

<details>
<summary><strong>Q26. What is a partial derivative?</strong></summary>

**Answer / theory**

A partial derivative measures change with respect to one variable while holding others fixed. For \(L(w_1,w_2)\), \(\partial L/\partial w_1\) is the local sensitivity to \(w_1\).

**When and where it is used:** Multivariable models, gradient computation.

</details>

<details>
<summary><strong>Q27. What is a gradient?</strong></summary>

**Answer / theory**

The gradient \(
abla_\theta L\) stacks all partial derivatives. It points in the direction of steepest local increase under the Euclidean metric, so gradient descent steps in the negative direction.

**When and where it is used:** Linear/logistic regression, neural networks.

</details>

<details>
<summary><strong>Q28. What is the chain rule?</strong></summary>

**Answer / theory**

For \(y=f(g(x))\), \(dy/dx=f'(g(x))g'(x)\). For nested neural-network layers, derivatives multiply through the computation graph; backpropagation applies this efficiently.

**When and where it is used:** Backpropagation, composite losses, deep learning.

</details>

<details>
<summary><strong>Q29. What is a Hessian matrix?</strong></summary>

**Answer / theory**

The Hessian contains second partial derivatives: \(H_{ij}=\partial^2L/(\partial\theta_i\partial\theta_j)\). It describes local curvature and is used by Newton-type methods and theoretical convergence analysis.

**When and where it is used:** Newton optimization, curvature diagnostics.

</details>

<details>
<summary><strong>Q30. What is gradient descent?</strong></summary>

**Answer / theory**

Update parameters by \(	heta_{t+1}=\theta_t-\eta\nabla L(\theta_t)\), where \(\eta\) is the learning rate. Too large a step can diverge; too small a step can converge slowly.

**When and where it is used:** Linear/logistic regression, neural networks.

</details>

<details>
<summary><strong>Q31. What is batch versus stochastic versus mini-batch gradient descent?</strong></summary>

**Answer / theory**

Batch uses all training examples per update; stochastic uses one; mini-batch uses a subset. Mini-batches balance compute efficiency, gradient noise, and hardware utilization.

**When and where it is used:** Deep learning training, large datasets.

</details>

<details>
<summary><strong>Q32. What is the learning rate?</strong></summary>

**Answer / theory**

The learning rate controls update size. It is not a universal constant: suitable values depend on objective scaling, model, optimizer, batch size, and schedule. Learning-rate schedules and warm-up can improve stability.

**When and where it is used:** All gradient-based training.

</details>

<details>
<summary><strong>Q33. What is a convex function?</strong></summary>

**Answer / theory**

A function \(f\) is convex when \(f(tx+(1-t)y)\le tf(x)+(1-t)f(y)\) for \(t\in[0,1]\). Every local minimum of a convex function is global. Squared-error linear regression is convex in its parameters.

**When and where it is used:** Linear regression, logistic regression, convex optimization.

</details>

<details>
<summary><strong>Q34. What are local and global minima?</strong></summary>

**Answer / theory**

A local minimum is no worse than nearby points; a global minimum is no worse than every feasible point. Non-convex neural-network losses can have many stationary points, so initialization and optimization dynamics matter.

**When and where it is used:** Neural network training, optimization theory.

</details>

<details>
<summary><strong>Q35. What is regularization?</strong></summary>

**Answer / theory**

Regularization adds a penalty to the empirical loss, e.g. \(L(w)+\lambda\|w\|_2^2\). It discourages overly complex parameter values and can improve generalization; \(\lambda\) is selected using validation data.

**When and where it is used:** Ridge/Lasso, neural networks, inverse problems.

</details>

<details>
<summary><strong>Q36. What is the bias-variance trade-off?</strong></summary>

**Answer / theory**

Expected prediction error can be decomposed conceptually into bias, variance, and irreducible noise under common squared-error assumptions. Simpler models may underfit (high bias); overly flexible models may overfit (high variance).

**When and where it is used:** Model selection, regularization, ensembles.

</details>

<details>
<summary><strong>Q37. What is a saddle point?</strong></summary>

**Answer / theory**

A stationary point has zero gradient; it may be a minimum, maximum, or saddle. At a saddle, curvature is positive in some directions and negative in others. Zero gradient does not guarantee a minimum.

**When and where it is used:** Deep-learning optimization.

</details>

<details>
<summary><strong>Q38. What is a Lagrange multiplier?</strong></summary>

**Answer / theory**

For constrained optimization, form \(\mathcal L(x,\lambda)=f(x)+\lambda g(x)\) for equality constraint \(g(x)=0\), then solve stationarity and feasibility conditions. Inequality constraints require KKT conditions.

**When and where it is used:** SVM margins, constrained estimation, resource allocation.

</details>

<details>
<summary><strong>Q39. What is the KKT framework?</strong></summary>

**Answer / theory**

Karush–Kuhn–Tucker conditions combine stationarity, primal feasibility, dual feasibility, and complementary slackness. Under suitable convexity and regularity assumptions, they characterize optimal solutions.

**When and where it is used:** Support vector machines, constrained convex optimization.

</details>

<details>
<summary><strong>Q40. What is automatic differentiation?</strong></summary>

**Answer / theory**

Automatic differentiation applies chain-rule operations to elementary computational steps, calculating derivatives accurate to floating-point precision up to numerical error. It differs from symbolic differentiation and finite-difference approximation.

**When and where it is used:** PyTorch/JAX/TensorFlow backpropagation.

</details>

## 5. Probability and random variables

Read the core idea first, then open each question to test recall. Answers explain the concept, formula, algorithm connection, and practical caveat.

<details>
<summary><strong>Q41. What is probability?</strong></summary>

**Answer / theory**

Probability quantifies uncertainty on a scale from 0 to 1. A probability model encodes assumptions about possible outcomes and supports inference and decision-making under uncertainty.

**When and where it is used:** Classification, Bayesian inference, risk modeling.

</details>

<details>
<summary><strong>Q42. What is a random variable?</strong></summary>

**Answer / theory**

A random variable maps outcomes to numerical values. It may be discrete (counts) or continuous (measurements). Its distribution describes probabilities of values or ranges.

**When and where it is used:** Statistical learning, simulation, probabilistic models.

</details>

<details>
<summary><strong>Q43. What is conditional probability?</strong></summary>

**Answer / theory**

\(P(A\mid B)=P(A\cap B)/P(B)\) for \(P(B)>0\). It updates probability when evidence \(B\) is known.

**When and where it is used:** Bayes classifiers, diagnostics, probabilistic prediction.

</details>

<details>
<summary><strong>Q44. State Bayes' theorem.</strong></summary>

**Answer / theory**

\(P(A\mid B)=P(B\mid A)P(A)/P(B)\). Posterior is proportional to likelihood times prior; the evidence term normalizes across hypotheses.

**When and where it is used:** Naive Bayes, Bayesian inference, spam detection.

</details>

<details>
<summary><strong>Q45. What are independent events?</strong></summary>

**Answer / theory**

Events \(A,B\) are independent when \(P(A\cap B)=P(A)P(B)\), equivalently \(P(A\mid B)=P(A)\) when defined. Independence is a strong assumption and should not be confused with zero correlation in general.

**When and where it is used:** Naive Bayes assumptions, sampling, experimental design.

</details>

<details>
<summary><strong>Q46. What are expectation and variance?</strong></summary>

**Answer / theory**

\(E[X]\) is the probability-weighted mean. \(Var(X)=E[(X-E[X])^2]=E[X^2]-E[X]^2\). Expectation describes location; variance describes spread.

**When and where it is used:** Loss expectations, uncertainty, standardization.

</details>

<details>
<summary><strong>Q47. What is covariance versus correlation?</strong></summary>

**Answer / theory**

Covariance measures joint variation and depends on units; correlation \(\rho=Cov(X,Y)/(\sigma_X\sigma_Y)\) is standardized to [-1,1]. Correlation captures linear association, not causation.

**When and where it is used:** Feature analysis, PCA, multicollinearity checks.

</details>

<details>
<summary><strong>Q48. What is the law of large numbers?</strong></summary>

**Answer / theory**

Under suitable assumptions, sample averages converge to the expected value as sample size grows. It supports estimation by averaging but does not imply that every finite sample is representative.

**When and where it is used:** Monte Carlo estimation, A/B tests, model evaluation.

</details>

<details>
<summary><strong>Q49. What is the central limit theorem?</strong></summary>

**Answer / theory**

Under common conditions, standardized sums or means approach a normal distribution as sample size grows, even if the population is not normal. Dependence, heavy tails, and small samples can complicate this approximation.

**When and where it is used:** Confidence intervals, sampling distributions, inference.

</details>

<details>
<summary><strong>Q50. What is a Bernoulli distribution?</strong></summary>

**Answer / theory**

A Bernoulli variable takes 1 with probability \(p\) and 0 with probability \(1-p\); its mean is \(p\) and variance \(p(1-p)\).

**When and where it is used:** Binary outcomes, logistic regression likelihood.

</details>

<details>
<summary><strong>Q51. What is a Binomial distribution?</strong></summary>

**Answer / theory**

For \(n\) independent Bernoulli trials with success probability \(p\), \(X\sim Binomial(n,p)\), \(E[X]=np\), and \(Var(X)=np(1-p)\).

**When and where it is used:** Conversion counts, click-through events.

</details>

<details>
<summary><strong>Q52. What is a Gaussian/Normal distribution?</strong></summary>

**Answer / theory**

A normal distribution is parameterized by mean \(\mu\) and variance \(\sigma^2\), with a symmetric bell-shaped density. It is central to measurement noise models and many inference approximations, but real data need not be normal.

**When and where it is used:** Gaussian Naive Bayes, linear regression assumptions, noise models.

</details>

<details>
<summary><strong>Q53. What is a Poisson distribution?</strong></summary>

**Answer / theory**

For event counts over a fixed exposure, \(X\sim Poisson(\lambda)\) has mean and variance \(\lambda\). It assumes a particular count process; overdispersion may call for alternatives such as negative binomial models.

**When and where it is used:** Count prediction, arrivals, incident rates.

</details>

<details>
<summary><strong>Q54. What is a likelihood?</strong></summary>

**Answer / theory**

Likelihood \(L(\theta\mid data)=p(data\mid\theta)\) treats observed data as fixed and parameters as variable. It is not itself generally a probability distribution over parameters.

**When and where it is used:** Maximum likelihood, regression, classification.

</details>

<details>
<summary><strong>Q55. What is maximum likelihood estimation?</strong></summary>

**Answer / theory**

MLE selects \(\hat\theta=\arg\max_\theta p(data\mid\theta)\), usually by maximizing log-likelihood. It can be biased in finite samples and can fail under model misspecification.

**When and where it is used:** Linear/logistic regression, probabilistic models.

</details>

<details>
<summary><strong>Q56. What is a prior, likelihood, and posterior?</strong></summary>

**Answer / theory**

A prior encodes uncertainty before observing data; likelihood measures how well parameters explain data; posterior combines both via Bayes' theorem. The prior's influence typically decreases with informative data, though not always.

**When and where it is used:** Bayesian regression, medical risk, uncertainty estimates.

</details>

<details>
<summary><strong>Q57. What is entropy of a random variable?</strong></summary>

**Answer / theory**

\(H(X)=-\sum_x p(x)\log p(x)\). Entropy measures uncertainty or average information, with log base 2 giving bits and natural logs giving nats.

**When and where it is used:** Decision trees, coding, information theory.

</details>

## 6. Statistics and inference

Read the core idea first, then open each question to test recall. Answers explain the concept, formula, algorithm connection, and practical caveat.

<details>
<summary><strong>Q58. What is a population versus a sample?</strong></summary>

**Answer / theory**

A population is the target set or data-generating process; a sample is the observed subset. Statistical inference uses samples to estimate population properties and quantify uncertainty.

**When and where it is used:** Model evaluation, survey sampling, experimentation.

</details>

<details>
<summary><strong>Q59. What is sample mean?</strong></summary>

**Answer / theory**

\(ar x=\sum_i x_i/n\). It estimates a population mean and is sensitive to outliers. The sample mean is unbiased for the population mean under independent identically distributed sampling with finite mean.

**When and where it is used:** EDA, imputation, estimation.

</details>

<details>
<summary><strong>Q60. What is sample variance and why use n−1?</strong></summary>

**Answer / theory**

\(s^2=\sum_i(x_i-\bar x)^2/(n-1)\). The \(n-1\) denominator corrects finite-sample bias for estimating population variance when the mean is estimated from the same sample.

**When and where it is used:** Standard error, confidence intervals, feature scaling.

</details>

<details>
<summary><strong>Q61. What is standard deviation?</strong></summary>

**Answer / theory**

Standard deviation is the square root of variance and returns to the original measurement units. It measures spread, but is sensitive to outliers.

**When and where it is used:** Standardization, monitoring, anomaly detection.

</details>

<details>
<summary><strong>Q62. What is a z-score?</strong></summary>

**Answer / theory**

\(z=(x-\mu)/\sigma\). It measures distance from the mean in standard-deviation units. If \(\mu,\sigma\) are estimated from data, be careful to fit them only on training data.

**When and where it is used:** Outlier screening, standardization, anomaly detection.

</details>

<details>
<summary><strong>Q63. What is a confidence interval?</strong></summary>

**Answer / theory**

A confidence-interval procedure has a long-run coverage property under its assumptions. A 95% frequentist interval does not mean there is a 95% probability that a fixed parameter lies in this particular computed interval.

**When and where it is used:** A/B testing, uncertainty reporting.

</details>

<details>
<summary><strong>Q64. What is a p-value?</strong></summary>

**Answer / theory**

A p-value is the probability, assuming the null and model assumptions, of observing a test statistic at least as extreme as the one observed. It is not the probability that the null is true.

**When and where it is used:** Hypothesis testing, controlled experiments.

</details>

<details>
<summary><strong>Q65. What are Type I and Type II errors?</strong></summary>

**Answer / theory**

Type I error rejects a true null; Type II error fails to reject a false null. Significance level \(lpha\) controls Type I error under the test assumptions; power \(1-eta\) is the chance of detecting a specified alternative.

**When and where it is used:** Medical screening, A/B testing, fraud detection.

</details>

<details>
<summary><strong>Q66. What is a statistical hypothesis test?</strong></summary>

**Answer / theory**

A test specifies null and alternative hypotheses, a test statistic, a reference distribution, and a decision rule. Interpret results alongside effect size, uncertainty, assumptions, and practical significance.

**When and where it is used:** Experiment analysis, product changes, scientific claims.

</details>

<details>
<summary><strong>Q67. What is the difference between correlation and causation?</strong></summary>

**Answer / theory**

Correlation is association; causation asks whether changing one variable changes another under a defined intervention. Confounding, selection bias, and reverse causality can create misleading associations.

**When and where it is used:** Feature interpretation, causal inference, business experiments.

</details>

<details>
<summary><strong>Q68. What is sampling bias?</strong></summary>

**Answer / theory**

Sampling bias occurs when inclusion in the observed data systematically differs from the target population or process. It can invalidate generalization even when the sample is large.

**When and where it is used:** Training data collection, surveys, fairness audits.

</details>

<details>
<summary><strong>Q69. What is multicollinearity?</strong></summary>

**Answer / theory**

Predictors are highly linearly dependent. Predictions may remain useful, but individual coefficient estimates become unstable and their standard errors can grow. Check correlation, condition numbers, and variance inflation factors.

**When and where it is used:** Linear regression, coefficient interpretation.

</details>

<details>
<summary><strong>Q70. What is a bootstrap?</strong></summary>

**Answer / theory**

Bootstrap repeatedly resamples observations (usually with replacement) to approximate a statistic's sampling distribution. It can estimate uncertainty when analytic formulas are difficult, but does not fix biased or nonrepresentative data.

**When and where it is used:** Confidence intervals for metrics, model comparisons.

</details>

<details>
<summary><strong>Q71. What is cross-validation?</strong></summary>

**Answer / theory**

Cross-validation splits data into training/validation folds to estimate generalization. Use grouped or time-aware splits when examples are dependent; never tune on the final test set.

**When and where it is used:** Hyperparameter tuning, model selection.

</details>

<details>
<summary><strong>Q72. What is data leakage?</strong></summary>

**Answer / theory**

Leakage occurs when training or selection uses information unavailable at prediction time, such as future data or preprocessing fitted on the full dataset. It produces optimistic validation scores.

**When and where it is used:** All supervised learning pipelines.

</details>

## 7. Information theory and ML loss functions

Read the core idea first, then open each question to test recall. Answers explain the concept, formula, algorithm connection, and practical caveat.

<details>
<summary><strong>Q73. What is information gain?</strong></summary>

**Answer / theory**

Information gain is the reduction in entropy after a split: \(IG=H(Y)-\sum_j p_j H(Y\mid child_j)\). Decision trees use it to select splits that reduce target uncertainty.

**When and where it is used:** ID3/C4.5-style trees, feature split selection.

</details>

<details>
<summary><strong>Q74. What is Gini impurity?</strong></summary>

**Answer / theory**

\(Gini=1-\sum_k p_k^2\). It is zero for a pure node and larger when classes are mixed. CART often uses Gini impurity for classification splits.

**When and where it is used:** Decision trees, random forests.

</details>

<details>
<summary><strong>Q75. What is cross-entropy?</strong></summary>

**Answer / theory**

For true distribution \(p\) and predicted \(q\), \(H(p,q)=-\sum_i p_i\log q_i\). In classification it penalizes assigning low probability to the true class and is a standard probabilistic loss.

**When and where it is used:** Logistic regression, softmax classifiers, neural networks.

</details>

<details>
<summary><strong>Q76. What is binary cross-entropy?</strong></summary>

**Answer / theory**

\(L=-[y\log p+(1-y)\log(1-p)]\), averaged over examples. Clip probabilities or use stable logits-based implementations to avoid log(0) and numerical instability.

**When and where it is used:** Binary classification, logistic regression.

</details>

<details>
<summary><strong>Q77. What is mean squared error (MSE)?</strong></summary>

**Answer / theory**

\(MSE=\frac1n\sum_i(y_i-\hat y_i)^2\). It is smooth and penalizes large errors strongly, but is sensitive to outliers.

**When and where it is used:** Linear regression, regression neural networks.

</details>

<details>
<summary><strong>Q78. What is mean absolute error (MAE)?</strong></summary>

**Answer / theory**

\(MAE=\frac1n\sum_i|y_i-\hat y_i|\). It is more robust to large residuals than MSE, but has a kink at zero and a different optimization profile.

**When and where it is used:** Regression with outliers, interpretable error reporting.

</details>

<details>
<summary><strong>Q79. What is Huber loss?</strong></summary>

**Answer / theory**

Huber loss is quadratic for small residuals and linear for large residuals, controlled by threshold \(\delta\). It combines local smoothness with reduced outlier sensitivity.

**When and where it is used:** Robust regression.

</details>

<details>
<summary><strong>Q80. What is KL divergence?</strong></summary>

**Answer / theory**

\(D_{KL}(P\|Q)=\sum_x P(x)\log[P(x)/Q(x)]\). It measures the extra coding cost of using \(Q\) when the truth is \(P\). It is asymmetric and not a metric.

**When and where it is used:** Variational inference, distillation, distribution matching.

</details>

<details>
<summary><strong>Q81. What is perplexity in language modeling?</strong></summary>

**Answer / theory**

Perplexity is often \(\exp(\text{average negative log-likelihood})\) using natural logarithms. Lower is better on the same tokenization and evaluation data, but scores across different tokenizers are not directly comparable.

**When and where it is used:** Language-model evaluation.

</details>

<details>
<summary><strong>Q82. What is a margin in classification?</strong></summary>

**Answer / theory**

A margin measures how far a prediction lies on the correct side of a decision boundary. Hinge loss \(\max(0,1-yf(x))\) encourages a margin of at least one in a suitable label convention.

**When and where it is used:** Support vector machines, margin-based learning.

</details>

## 8. Algorithm-specific mathematics

Read the core idea first, then open each question to test recall. Answers explain the concept, formula, algorithm connection, and practical caveat.

<details>
<summary><strong>Q83. Derive the linear regression prediction equation.</strong></summary>

**Answer / theory**

With design matrix \(X\in\mathbb{R}^{n\times d}\), weights \(w\in\mathbb{R}^d\), and targets \(y\in\mathbb{R}^n\), predictions are \(\hat y=Xw+b\). For centered data and full column rank, OLS is \(\hat w=(X^\top X)^{-1}X^\top y\); in code, prefer `np.linalg.lstsq`.

**When and where it is used:** Linear regression, baseline forecasting.

</details>

<details>
<summary><strong>Q84. Why does logistic regression use the sigmoid?</strong></summary>

**Answer / theory**

The sigmoid \(\sigma(z)=1/(1+e^{-z})\) maps real-valued logits to (0,1), enabling probability interpretation for binary labels. The model is linear in log-odds, not in probability.

**When and where it is used:** Binary classification, calibrated probabilistic baselines.

</details>

<details>
<summary><strong>Q85. What is softmax?</strong></summary>

**Answer / theory**

\(p_k=e^{z_k}/\sum_j e^{z_j}\) converts class logits to probabilities summing to one. Subtracting \(\max(z)\) before exponentiation preserves probabilities and improves numerical stability.

**When and where it is used:** Multiclass classification, attention normalization.

</details>

<details>
<summary><strong>Q86. What is the K-means objective?</strong></summary>

**Answer / theory**

K-means minimizes within-cluster squared distances: \(\sum_i\|x_i-\mu_{c_i}\|_2^2\). It alternates assigning points to nearest centers and updating centers. It favors roughly spherical clusters and is scale-sensitive.

**When and where it is used:** Unsupervised segmentation, clustering.

</details>

<details>
<summary><strong>Q87. What is PCA optimizing?</strong></summary>

**Answer / theory**

PCA finds orthogonal directions maximizing projected variance, equivalently minimizing squared reconstruction error for a fixed-rank linear projection. Center features first; scale them when units differ and variance scale should not dominate.

**When and where it is used:** Compression, denoising, visualization, preprocessing.

</details>

<details>
<summary><strong>Q88. What is the SVM maximum-margin objective?</strong></summary>

**Answer / theory**

A linear soft-margin SVM commonly minimizes \(\frac12\|w\|^2+C\sum_i\max(0,1-y_i(w^\top x_i+b))\). The norm term encourages a wider margin; \(C\) trades margin size against training violations.

**When and where it is used:** Binary classification, high-dimensional feature spaces.

</details>

<details>
<summary><strong>Q89. What is the Naive Bayes conditional-independence assumption?</strong></summary>

**Answer / theory**

Given class \(Y\), features are treated as conditionally independent: \(P(x\mid Y)=\prod_j P(x_j\mid Y)\). This simplifying assumption is often false, yet can work well for text and sparse features.

**When and where it is used:** Spam filtering, text classification.

</details>

<details>
<summary><strong>Q90. What is a neural-network activation function?</strong></summary>

**Answer / theory**

An activation introduces nonlinearity. Without nonlinear activations, stacked affine layers collapse to one affine transformation. ReLU is \(\max(0,x)\); sigmoid and tanh can saturate and yield small gradients.

**When and where it is used:** MLPs, CNNs, transformers.

</details>

<details>
<summary><strong>Q91. What is backpropagation mathematically?</strong></summary>

**Answer / theory**

Backpropagation applies the chain rule to compute derivatives of a scalar loss with respect to each parameter by reusing intermediate derivatives from the computational graph.

**When and where it is used:** Training neural networks efficiently.

</details>

<details>
<summary><strong>Q92. What is attention scaled dot-product?</strong></summary>

**Answer / theory**

Attention is \(softmax(QK^\top/\sqrt{d_k})V\). The scale controls the magnitude of dot-product logits as key dimension grows, helping avoid overly sharp softmax distributions.

**When and where it is used:** Transformers, sequence modeling, multimodal models.

</details>

<details>
<summary><strong>Q93. What is the bias-variance role of ensembles?</strong></summary>

**Answer / theory**

Bagging averages models trained on resampled data to reduce variance; boosting adds learners sequentially to correct residual patterns and can reduce bias. Both can overfit or be miscalibrated if misused.

**When and where it is used:** Random forests, gradient boosting, XGBoost-style systems.

</details>

<details>
<summary><strong>Q94. What is a kernel trick?</strong></summary>

**Answer / theory**

A kernel \(k(x,z)\) represents an inner product in an implicit feature space without explicitly constructing that mapping. Valid kernel matrices are typically positive semidefinite.

**When and where it is used:** Kernel SVM, Gaussian processes.

</details>

<details>
<summary><strong>Q95. What is the Bellman equation?</strong></summary>

**Answer / theory**

Dynamic programming expresses a value as immediate reward plus discounted value of the next state, e.g. \(V(s)=\max_a E[r+\gamma V(s')\mid s,a]\) for an optimal value function.

**When and where it is used:** Reinforcement learning, planning.

</details>

<details>
<summary><strong>Q96. What is regularized least squares?</strong></summary>

**Answer / theory**

Ridge regression minimizes \(\|Xw-y\|_2^2+\lambda\|w\|_2^2\), shrinking coefficients and improving conditioning. Lasso uses \(\lambda\|w\|_1\), which can set some coefficients exactly to zero.

**When and where it is used:** Ridge/Lasso, high-dimensional regression.

</details>

Finite-precision rounding means \((a+b)+c\) can differ from \(a+(b+c)\). Summation order, parallel reductions, and hardware can produce small reproducibility differences.

<details>

**Answer / theory**

An algorithm is stable when small rounding errors do not cause disproportionately large errors in the result. Algebraically equivalent formulas can differ in stability, such as solving a system versus explicitly forming an inverse.

**When and where it is used:** Regression, softmax, likelihood calculations.

<details>
<summary><strong>Q100. What is overflow and underflow?</strong></summary>

**Answer / theory**

Overflow exceeds representable magnitude; underflow rounds tiny magnitudes toward zero. Stable log-sum-exp and log-domain computations help with probabilities and exponentials.

**When and where it is used:** Softmax, likelihoods, deep learning.

</details>

<details>
<summary><strong>Q101. Why standardize features?</strong></summary>

**Answer / theory**

Standardization transforms \(x_j\) to \((x_j-\mu_j)/\sigma_j\), making scales more comparable. It is often important for gradient-based models, KNN, SVMs, and PCA; fit the scaler on training data only.

**When and where it is used:** KNN, SVM, PCA, linear/logistic regression with regularization.

</details>

<details>
<summary><strong>Q102. What is vectorization?</strong></summary>

**Answer / theory**

Vectorization expresses operations over whole arrays rather than Python loops, often enabling optimized native kernels. It improves speed but can increase temporary memory usage; chunking may be necessary for large arrays.

**When and where it is used:** NumPy preprocessing, numerical kernels.

<details>
<summary><strong>Q103. What is time complexity versus space complexity?</strong></summary>

**Answer / theory**

Time complexity describes how work scales with input size; space complexity describes memory growth. An \(O(n^2)\) pairwise-distance matrix may be infeasible even when the arithmetic itself is simple.

**When and where it is used:** Nearest neighbors, pairwise similarities, scaling.

</details>

<details>
<summary><strong>Q104. What is Monte Carlo estimation?</strong></summary>

**Answer / theory**

Monte Carlo approximates expectations using random samples: \(E[f(X)]\approx\frac1N\sum_i f(x_i)\). Under independent finite-variance sampling, standard error often falls proportional to \(1/\sqrt N\).

**When and where it is used:** Bayesian inference, risk estimation, simulation.

</details>

<details>
<summary><strong>Q105. What is reproducibility in numerical experiments?</strong></summary>

Set seeds where possible, record library versions, data splits, preprocessing, hardware, and configuration. Exact bitwise reproducibility may still be unavailable across devices or parallel implementations.

**When and where it is used:** Model evaluation, scientific work, MLOps.

**Answer / theory**

Detect NaN and infinities explicitly, decide whether to impute, drop, flag, or reject records based on domain meaning, and ensure transformations do not silently propagate invalid values.

**When and where it is used:** Data validation, training and inference pipelines.

### Lab 10 — Gradient descent for one-dimensional regression

**Concept and formula:** Shows derivatives of MSE and the update rule. Learning rate and feature scale affect convergence.

### Lab 12 — Sigmoid and binary cross-entropy

**Concept and formula:** Logistic regression; the sign-aware sigmoid avoids overflow for very negative logits.

<details>
<summary><strong>Interview check: explain the idea before running the code</strong></summary>

A strong answer states (1) the mathematical operation, (2) its assumptions, (3) the algorithm or use case, and (4) a numerical or data-quality caveat.

</details>

````python
import numpy as np
def sigmoid(z):
    z = np.asarray(z, dtype=float)
    out = np.empty_like(z)
    pos = z >= 0
    out[pos] = 1 / (1 + np.exp(-z[pos]))
    ez = np.exp(z[~pos])
    out[~pos] = ez / (1 + ez)
    return out
logits = np.array([-3., 0., 3.])
p = sigmoid(logits)
y = np.array([0., 1., 1.])
eps = 1e-12
bce = -np.mean(y*np.log(np.clip(p, eps, 1)) + (1-y)*np.log(np.clip(1-p, eps, 1)))
print("probabilities:", p)
print("binary cross-entropy:", bce)
````

### Lab 13 — Numerically stable softmax

**Concept and formula:** Multiclass classifiers and transformer attention; subtracting max logits prevents overflow.

<details>
<summary><strong>Interview check: explain the idea before running the code</strong></summary>

A strong answer states (1) the mathematical operation, (2) its assumptions, (3) the algorithm or use case, and (4) a numerical or data-quality caveat.

````python
import numpy as np
def softmax(z):
    z = np.asarray(z, dtype=float)
    shifted = z - np.max(z, axis=-1, keepdims=True)
    e = np.exp(shifted)
    return e / e.sum(axis=-1, keepdims=True)
logits = np.array([[1000., 1001., 999.], [1., 2., 3.]])
p = softmax(logits)
print(p)
print("row sums:", p.sum(axis=1))
assert np.allclose(p.sum(axis=1), 1)
````

<details>
<summary><strong>Advanced interview Q1: Why is \(X^TX\) potentially a poor route to ordinary least squares?</strong></summary>

**Answer**

It squares the condition number in the 2-norm: \(\kappa(X^TX)pprox\kappa(X)^2\). Prefer QR/SVD-based least-squares solvers such as `np.linalg.lstsq` for stability.

<summary><strong>Advanced interview Q2: Why can a model have high accuracy but poor business performance?</strong></summary>

**Answer**

Accuracy ignores class prevalence, error costs, calibration, and threshold choice. Inspect precision/recall, PR-AUC where relevant, calibration, subgroup metrics, and decision utility.

<details>
<summary><strong>Advanced interview Q3: Why can correlation be near zero when variables are dependent?</strong></summary>

**Answer**

Correlation measures linear association only. For example, symmetric \(X\) and \(Y=X^2\) can have zero correlation despite deterministic dependence.

</details>

<details>
<summary><strong>Advanced interview Q4: Why do we subtract the maximum logit in softmax?</strong></summary>

**Answer**

It leaves probabilities unchanged because multiplying all exponentials by the same positive constant cancels in the normalization, while avoiding overflow.

</details>

<details>
<summary><strong>Advanced interview Q5: When should you use MAE instead of MSE?</strong></summary>

**Answer**

When large residuals should not dominate as strongly or robust absolute-error reporting matters. MSE may be preferable when Gaussian-noise assumptions or smooth gradients are important.

</details>

<details>
<summary><strong>Advanced interview Q6: Why is PCA sensitive to scaling?</strong></summary>

**Answer**

PCA maximizes variance; features measured in large units can dominate total variance. Standardize when scale differences are arbitrary, but do not automatically standardize if absolute variance is meaningful.

</details>

<details>
<summary><strong>Advanced interview Q7: What changes when observations are time-dependent?</strong></summary>

**Answer**

IID assumptions and random shuffling may be invalid. Use chronological splits, leakage-safe features, and time-series-specific uncertainty and evaluation methods.

</details>

<details>
<summary><strong>Advanced interview Q8: How would you debug exploding gradients?</strong></summary>

**Answer**

Check learning rate, loss scaling, invalid values, initialization, and sequence length; inspect gradient norms, use stable logit-based losses, consider gradient clipping, and verify gradients numerically.

</details>

<details>
<summary><strong>Advanced interview Q9: What is the difference between a parameter and a hyperparameter?</strong></summary>

**Answer**

Parameters are learned from training data, such as weights. Hyperparameters configure the learning procedure, such as learning rate, regularization strength, tree depth, or number of clusters.

</details>

<details>
<summary><strong>Advanced interview Q10: Why does regularization often improve generalization?</strong></summary>

**Answer**

It constrains effective model complexity or parameter magnitude, reducing variance at the cost of possible bias. Tune it using validation data and inspect both train and validation metrics.

</details>

<details>

</details>

<details>

It does not fix biased sampling, guarantee the model is correct, or quantify all forms of uncertainty. Coverage depends on the sampling and modeling assumptions.

</details>

<details>

Repeated decisions leak information from the test set into the workflow and make reported test performance optimistic. Keep a final untouched test set.

## 13. Production scenarios and debugging playbook

<details>
<summary><strong>Scenario 1 — Validation score is suspiciously perfect</strong></summary>

**Likely causes:** target leakage, preprocessing fitted before the split, duplicate records across train/test, future features, or test-set tuning.

**Response:** split first; fit imputation/scaling/feature selection on training data only; use group/time-aware splitting; audit feature availability at prediction time; reserve an untouched test set.

</details>

<details>
<summary><strong>Scenario 2 — Regression coefficients change dramatically after adding a feature</strong></summary>

**Likely causes:** multicollinearity, small sample size, scale differences, outliers, or a changed design matrix.

**Response:** inspect correlations and condition number; compare coefficient uncertainty and predictive performance; consider ridge regression or feature redesign; standardize if appropriate.

</details>

<details>
<summary><strong>Scenario 3 — Softmax returns NaN or infinity</strong></summary>

**Likely causes:** exponentiating large logits, invalid inputs, or zero/invalid normalization.

**Response:** subtract row-wise maximum logits; use a stable log-softmax/cross-entropy-from-logits implementation; assert all logits and outputs are finite.

</details>

<details>
<summary><strong>Scenario 4 — Training loss falls but validation loss rises</strong></summary>

**Likely causes:** overfitting, leakage in evaluation, train/validation distribution mismatch, or a noisy/small validation set.

**Response:** inspect curves and split design; regularize, simplify, augment data, or early-stop using validation metrics; do not tune on the final test set.

</details>

<details>
<summary><strong>Scenario 5 — Pairwise distance calculation exhausts memory</strong></summary>

**Likely causes:** materializing an \(n\\times m\\times d\) intermediate or full \(n\\times m\) distance matrix.

**Response:** process queries in chunks; use vectorized algebra carefully; use approximate nearest-neighbor indexes when appropriate; measure peak memory and latency.

</details>

<details>

````python
# Export the current notebook to HTML (expand/collapse) and PDF (printable).
# In Colab, the preferred path reads the live notebook; if unavailable, upload the .ipynb.
from pathlib import Path
import json, html, re, sys, subprocess, os

try:
    import nbformat
except ImportError:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "nbformat"])
    import nbformat

try:
    from bs4 import BeautifulSoup
except ImportError:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "beautifulsoup4"])
    from bs4 import BeautifulSoup

try:
    from pygments import highlight
    from pygments.lexers import PythonLexer
    from pygments.formatters import HtmlFormatter
except ImportError:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "pygments"])
    from pygments import highlight
    from pygments.lexers import PythonLexer
    from pygments.formatters import HtmlFormatter

try:
    from reportlab.lib import colors
    from reportlab.lib.pagesizes import A4
    from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
    from reportlab.lib.units import mm
    from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer, Preformatted, PageBreak
except ImportError:
    subprocess.check_call([sys.executable, "-m", "pip", "install", "-q", "reportlab"])
    from reportlab.lib import colors
    from reportlab.lib.pagesizes import A4
    from reportlab.lib.styles import getSampleStyleSheet, ParagraphStyle
    from reportlab.lib.units import mm
    from reportlab.platypus import SimpleDocTemplate, Paragraph, Spacer, Preformatted, PageBreak

notebook = None
try:
    from google.colab import _message
    live = _message.blocking_request("get_ipynb", timeout_sec=10)
    if isinstance(live, dict):
        live = live.get("ipynb", live)
    if isinstance(live, str):
        live = json.loads(live)
    notebook = nbformat.from_dict(live) if isinstance(live, dict) and "cells" in live else None
except Exception:
    notebook = None

if notebook is None:
    try:
        from google.colab import files
        print("Upload the saved .ipynb file to export its latest saved outputs.")
        uploaded = files.upload()
        ipynb_files = [name for name in uploaded if name.endswith(".ipynb")]
        if not ipynb_files:
            raise ValueError("No .ipynb file was uploaded.")
        notebook = nbformat.read(ipynb_files[0], as_version=4)
    except ImportError:
        candidates = list(Path(".").glob("*.ipynb"))
        if not candidates:
            raise RuntimeError("Could not retrieve the live notebook. Save and upload the .ipynb, or run this cell in Colab.")
        notebook = nbformat.read(candidates[0], as_version=4)

def as_text(value):
    return "".join(value) if isinstance(value, list) else str(value)

def inline_md(text):
    text = html.escape(text)
    text = re.sub(r"`([^`]+)`", r"<code>\1</code>", text)
    text = re.sub(r"\*\*(.+?)\*\*", r"<strong>\1</strong>", text)
    text = re.sub(r"\*(.+?)\*", r"<em>\1</em>", text)
    return text

def render_md(source):
    # Ensure the source is a string
    source = as_text(source)
    lines = source.splitlines()
    out, para, codebuf, in_code = [], [], [], False
    def flush():
        nonlocal para
        if para:
            out.append("<p>" + "<br>".join(inline_md(x) for x in para) + "</p>")
            para = []
    for line in lines:
        if line.strip().startswith("```"):
            flush()
            if not in_code:
                in_code, codebuf = True, []
            else:
                out.append("<pre><code>" + html.escape("\n".join(codebuf)) + "</code></pre>")
                in_code = False
            continue
        if in_code:
            codebuf.append(line); continue
        if not line.strip():
            flush(); continue
        if line.lstrip().startswith(("<details", "</details", "<summary", "</summary")):
            flush(); out.append(line); continue
        m = re.match(r"^(#{1,4})\s+(.*)", line)
        if m:
            flush(); out.append(f"<h{len(m.group(1))}>{inline_md(m.group(2))}</h{len(m.group(1))}>"); continue
        if line.strip().startswith("|"):
            flush()
            # Basic table block
            table = [line]
            continue
        if re.match(r"^\s*[-*]\s+", line):
            flush(); out.append("<p>• " + inline_md(re.sub(r"^\s*[-*]\s+", "", line)) + "</p>"); continue
        if re.match(r"^\s*\d+\.\s+", line):
            flush(); out.append("<p>" + inline_md(line.strip()) + "</p>"); continue
        para.append(line)
    flush()
    if in_code:
        out.append("<pre><code>" + html.escape("\n".join(codebuf)) + "</code></pre>")
    return "\n".join(out)

def output_html(output):
    kind = output.get("output_type", "")
    if kind == "stream":
        return "<div class='output'><b>Output</b><pre>" + html.escape(as_text(output.get("text", ""))) + "</pre></div>"
    if kind == "error":
        return "<div class='output error'><b>Error</b><pre>" + html.escape("\n".join(output.get("traceback", []))) + "</pre></div>"
    data = output.get("data", {})
    if "text/html" in data:
        return "<div class='output'>" + as_text(data["text/html"]) + "</div>"
    if "image/png" in data:
        return "<div class='output'><img style='max-width:100%' src='data:image/png;base64," + as_text(data["image/png"]) + "'></div>"
    if "text/plain" in data:
        return "<div class='output'><pre>" + html.escape(as_text(data["text/plain"])) + "</pre></div>"
    return ""

sections, toc = [], []
for idx, cell in enumerate(notebook.cells):
    if cell.cell_type == "markdown":
        soup = BeautifulSoup(render_md(cell.source), "html.parser")
        for h in soup.find_all(["h1", "h2", "h3"]):
            anchor = f"section-{len(toc)+1}"
            h["id"] = anchor
            toc.append((anchor, h.get_text(" ", strip=True), int(h.name[1])))
        sections.append("<article class='markdown'>" + str(soup) + "</article>")
    elif cell.cell_type == "code":
        code_html = highlight(as_text(cell.source), PythonLexer(), HtmlFormatter(nowrap=False))
        outputs = "".join(output_html(o) for o in cell.get("outputs", []))
        sections.append("<article class='codecard'><details open><summary>⌄ Runnable Code Cell <span class='hint'>expand / collapse</span></summary><div class='codebody'><div class='label'>PYTHON</div>" + code_html + (outputs or "<p class='muted'>No saved output.</p>") + "</div></details></article>")

toc_html = "".join(f"<li class='level{level}'><a href='#{anchor}'>{html.escape(title)}</a></li>" for anchor,title,level in toc)
css = HtmlFormatter().get_style_defs(".highlight")
html_doc = """<!doctype html><html><head><meta charset='utf-8'><meta name='viewport' content='width=device-width,initial-scale=1'><title>Mathematics for ML Handbook</title><style>""" + css + """
body{margin:0;background:#f4f7fb;color:#1f2937;font:15px/1.7 -apple-system,BlinkMacSystemFont,'Segoe UI',Arial,sans-serif}.hero{background:linear-gradient(130deg,#172554,#2563eb);color:white;padding:38px 24px}.hero div{max-width:1200px;margin:auto}.hero h1{font-size:clamp(28px,4vw,44px);line-height:1.15}.toolbar{position:sticky;top:0;background:white;padding:12px;text-align:center;border-bottom:1px solid #dbe3ef;z-index:10}.toolbar button{padding:10px 14px;margin:3px;border:0;border-radius:7px;background:#2563eb;color:white;cursor:pointer;font-weight:700}.layout{max-width:1500px;margin:22px auto;padding:0 18px;display:grid;grid-template-columns:270px minmax(0,1fr);gap:20px}aside{position:sticky;top:70px;max-height:calc(100vh - 90px);overflow:auto;background:white;border:1px solid #dbe3ef;border-radius:12px;padding:16px}aside ul{list-style:none;padding-left:0}aside li{font-size:12px;margin:6px 0}aside .level2{padding-left:9px}aside .level3{padding-left:18px}aside a{color:#334155;text-decoration:none}.markdown,.codecard{background:white;border:1px solid #dbe3ef;border-radius:12px;padding:20px;margin-bottom:15px;overflow-wrap:anywhere}.markdown h1,.markdown h2,.markdown h3{color:#172554;scroll-margin-top:80px}.markdown h2{border-bottom:1px solid #dbeafe;padding-bottom:7px}.markdown table{border-collapse:collapse;width:100%;overflow:auto;display:block}.markdown th,.markdown td{border:1px solid #dbe3ef;padding:7px}.markdown th{background:#eff6ff}.markdown pre,.highlight pre,.output pre{white-space:pre-wrap;overflow-wrap:anywhere;border-radius:8px;padding:12px;font-size:12px}.codecard summary{cursor:pointer;font-weight:800;color:#172554;list-style:none}.hint{float:right;color:#64748b;font-size:12px;font-weight:400}.label{color:#2563eb;font-size:11px;font-weight:800;letter-spacing:1px}.codebody{padding-top:12px}.highlight{background:#111827;border-radius:8px;overflow:auto}.highlight pre{background:transparent}.output{background:#f8fafc;border:1px solid #dbe3ef;padding:10px;border-radius:8px;margin-top:10px}.output pre{background:#f1f5f9}.muted{color:#64748b;font-size:12px}@media(max-width:900px){.layout{grid-template-columns:1fr}aside{position:static;max-height:280px}}@media print{@page{size:A4;margin:12mm}.hero{background:white!important;color:#172554!important;padding:5px}.toolbar,aside{display:none!important}.layout{display:block;margin:0;padding:0}.markdown,.codecard{box-shadow:none;break-inside:auto;padding:10px}details,details:not([open])>*{display:block!important}details>summary{display:none!important}.highlight pre,.output pre{font-size:7pt}
</style></head><body><header class='hero'><div><p>MATHEMATICS · MACHINE LEARNING · INTERVIEW PREPARATION</p><h1>Mathematics for Machine Learning</h1><p>Theory, formulas, expandable interview answers, runnable labs, and production scenarios</p></div></header><div class='toolbar'><button onclick='toggleAll(true)'>Expand all</button><button onclick='toggleAll(false)'>Collapse all</button><button onclick='window.print()'>Print / Save as PDF</button></div><div class='layout'><aside><h2>Contents</h2><ul>""" + toc_html + """</ul></aside><main>""" + "".join(sections) + """</main></div><script>function toggleAll(x){document.querySelectorAll('details').forEach(d=>d.open=x)}</script></body></html>"""
base = Path.cwd()
html_path = base / "Mathematics_for_ML_Interview_Handbook.html"
html_path.write_text(html_doc, encoding="utf-8")

# PDF export: readable printable text with code and saved outputs.
styles = getSampleStyleSheet()
styles.add(ParagraphStyle(name="MathCover", parent=styles["Title"], fontName="Helvetica-Bold", fontSize=23, leading=28, textColor=colors.HexColor("#172554"), alignment=1, spaceAfter=16))
styles.add(ParagraphStyle(name="MathHead", parent=styles["Heading2"], fontSize=13, leading=16, textColor=colors.HexColor("#1d4ed8"), spaceBefore=10, spaceAfter=5, keepWithNext=True))
styles.add(ParagraphStyle(name="MathBody", parent=styles["BodyText"], fontSize=8.5, leading=11, spaceAfter=5, splitLongWords=True))
styles.add(ParagraphStyle(name="MathCode", fontName="Courier", fontSize=6.3, leading=7.8, backColor=colors.HexColor("#f1f5f9"), borderColor=colors.HexColor("#dbe3ef"), borderWidth=.4, borderPadding=5, spaceBefore=3, spaceAfter=6, splitLongWords=True))
story = [Spacer(1, 30*mm), Paragraph("Mathematics for Machine Learning", styles["MathCover"]), Paragraph("Theory • Formulas • Interview Q&A • Runnable Labs", styles["MathBody"]), Spacer(1, 10*mm), Paragraph("A practical mathematics handbook for machine learning interviews and numerical model development.", styles["MathBody"]), PageBreak()]

def add_p(txt, style="MathBody"):
    txt = re.sub(r"\s+", " ", str(txt)).strip()
    if txt:
        story.append(Paragraph(html.escape(txt), styles[style]))
def add_pre(txt):
    txt = as_text(txt)
    if txt.strip():
        story.append(Preformatted(txt[:40000], styles["MathCode"]))

for cell in notebook.cells:
    if cell.cell_type == "markdown":
        soup = BeautifulSoup(render_md(cell.source), "html.parser")
        for el in soup.contents:
            if getattr(el, "name", None) is None:
                if str(el).strip(): add_p(str(el))
            elif el.name in ("h1","h2","h3","h4"):
                add_p(el.get_text(" ", strip=True), "MathHead")
            elif el.name == "pre":
                add_pre(el.get_text())
            elif el.name in ("ul","ol"):
                for j, li in enumerate(el.find_all("li", recursive=False), 1):
                    add_p(("• " if el.name=="ul" else f"{j}. ") + li.get_text(" ", strip=True))
            elif el.name == "table":
                rows = []
                for tr in el.find_all("tr"):
                    row = [c.get_text(" ", strip=True) for c in tr.find_all(["th","td"], recursive=False)]
                    if row: rows.append(row)
                if rows:
                    from reportlab.platypus import Table, TableStyle
                    ncol = max(len(r) for r in rows)
                    rows = [r + [""]*(ncol-len(r)) for r in rows]
                    widths = [168*mm/ncol]*ncol
                    t = Table(rows, repeatRows=1, colWidths=widths)
                    t.setStyle(TableStyle([("BACKGROUND",(0,0),(-1,0),colors.HexColor("#dbeafe")),("TEXTCOLOR",(0,0),(-1,0),colors.HexColor("#172554")),("FONTNAME",(0,0),(-1,0),"Helvetica-Bold"),("FONTSIZE",(0,0),(-1,-1),6.8),("LEADING",(0,0),(-1,-1),8.5),("GRID",(0,0),(-1,-1),.3,colors.HexColor("#cbd5e1")),("VALIGN",(0,0),(-1,-1),"TOP")]))
                    story.append(t); story.append(Spacer(1,5))
            else:
                txt = el.get_text(" ", strip=True)
                if txt: add_p(txt)
    elif cell.cell_type == "code":
        add_p("Runnable Code Cell", "MathHead")
        add_pre(cell.source)
        for output in cell.get("outputs", []):
            if output.get("output_type") == "stream":
                out_text = as_text(output.get("text", ""))
                if out_text.strip(): add_p("Saved output"); add_pre(out_text)
            elif output.get("output_type") == "error":
                add_p("Execution error", "MathHead"); add_pre("\n".join(output.get("traceback", [])))
            else:
                data = output.get("data", {})
                if "text/plain" in data:
                    out_text = as_text(data["text/plain"])
                    if out_text.strip(): add_p("Saved output"); add_pre(out_text)
                elif "image/png" in data:
                    add_p("[Image output retained in HTML export.]")

pdf_path = base / "Mathematics_for_ML_Interview_Handbook.pdf"
def footer(canvas, doc):
    canvas.saveState()
    if doc.page > 1:
        canvas.setFont("Helvetica", 7)
        canvas.setFillColor(colors.HexColor("#64748b"))
        canvas.drawString(16*mm, A4[1]-8*mm, "MATHEMATICS FOR MACHINE LEARNING")
        canvas.drawCentredString(A4[0]/2, 8*mm, f"Page {doc.page}")
    canvas.restoreState()
SimpleDocTemplate(str(pdf_path), pagesize=A4, leftMargin=16*mm, rightMargin=16*mm, topMargin=16*mm, bottomMargin=15*mm, title="Mathematics for Machine Learning Handbook").build(story, onFirstPage=footer, onLaterPages=footer)

# Trigger browser downloads in Colab. In a local Jupyter runtime, files remain in cwd.
try:
    from google.colab import files as colab_files
    colab_files.download(str(html_path))
    colab_files.download(str(pdf_path))
except Exception:
    print("Export complete. HTML and PDF are saved in the current working directory.")
print("HTML:", html_path)
print("PDF:", pdf_path)
````
