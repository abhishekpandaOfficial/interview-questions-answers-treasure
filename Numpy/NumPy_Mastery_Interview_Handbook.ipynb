**Formula quick reference**
- Mean: $$\mu = \frac{1}{n}\sum_{i=1}^{n}x_i$$

- Population variance: $$\sigma^2 = \frac{1}{n}\sum_{i=1}^{n}(x_i-\mu)^2$$

- Standardization: $$z = \frac{x-\mu}{\sigma}$$

- Dot product: $$a^T b = \sum_i a_i b_i$$

- Euclidean distance: $$d(x,y)=\sqrt{\sum_j(x_j-y_j)^2}$$

- Mean squared error: $$MSE=\frac{1}{n}\sum_i(y_i-\hat y_i)^2$$

- Softmax: $$p_i=\frac{e^{z_i}}{\sum_j e^{z_j}}$$

- Linear least squares: $$\hat\beta=\arg\min_\beta\|X\beta-y\|_2^2$$

## Part 1 — 120 NumPy Theory and Interview Questions

<details>
<summary><strong>Q001. What is NumPy?</strong></summary>

**Answer / theory**

NumPy (Numerical Python) is Python's foundational library for fast numerical computing. Its central object, `ndarray`, stores homogeneous, n-dimensional data and supports vectorized operations, broadcasting, linear algebra, statistics, random sampling, and efficient memory handling.

**Practical note:** Use NumPy when operations are numerical and array-oriented; use pandas when labeled/tabular data manipulation is central.
</details>

<details>
<summary><strong>Q002. Why is NumPy usually faster than Python loops?</strong></summary>

**Answer / theory**

An ndarray stores elements in a compact, typed memory buffer. Many operations execute in compiled C/Fortran loops, reducing Python interpreter overhead. Vectorization also lets libraries optimize memory access and CPU instructions.

**Practical note:** Vectorization is often faster, but temporary arrays can increase memory use; benchmark representative workloads.
</details>

<details>
<summary><strong>Q003. What is an ndarray?</strong></summary>

**Answer / theory**

An `ndarray` is an N-dimensional array with a fixed dtype, shape, and memory layout. It can represent a scalar (0-D), vector (1-D), matrix (2-D), or higher-dimensional tensor.

**Practical note:** Key attributes: `ndim`, `shape`, `size`, `dtype`, `itemsize`, and `nbytes`.
</details>

<details>
<summary><strong>Q004. What is the difference between a Python list and an ndarray?</strong></summary>

**Answer / theory**

Lists can mix object types and store references to Python objects. NumPy arrays typically hold values of one dtype in a compact buffer, enabling vectorized arithmetic and predictable memory use.

**Practical note:** Lists are flexible containers; ndarrays are better for numerical workloads.
</details>

<details>
<summary><strong>Q005. What do shape, size, and ndim mean?</strong></summary>

**Answer / theory**

`shape` is the tuple of axis lengths; `size` is the total number of elements; `ndim` is the number of axes. For shape `(2, 3, 4)`, size is `2×3×4=24` and ndim is 3.

**Practical note:** Formula: `size = product(shape)`; for a scalar, shape is `()` and size is 1.
</details>

<details>
<summary><strong>Q006. What is dtype?</strong></summary>

**Answer / theory**

`dtype` describes the element type and representation, such as `int64`, `float32`, or `bool`. Dtype affects precision, range, memory, and which operations are valid.

**Practical note:** Use explicit dtypes when memory, interoperability, or numerical precision matters.
</details>

<details>
<summary><strong>Q007. How do you create arrays?</strong></summary>

**Answer / theory**

Common constructors include `np.array`, `np.asarray`, `np.zeros`, `np.ones`, `np.full`, `np.empty`, `np.arange`, `np.linspace`, and random generators.

**Practical note:** `empty` allocates uninitialized values: fill it before reading.
</details>

<details>
<summary><strong>Q008. What is the difference between arange and linspace?</strong></summary>

**Answer / theory**

`arange(start, stop, step)` advances by a step and excludes the stop value. `linspace(start, stop, num)` creates a fixed number of evenly spaced points, including endpoints by default.

**Practical note:** For floating-point grids with a desired count, `linspace` avoids step-count ambiguity.
</details>

<details>
<summary><strong>Q009. What is vectorization?</strong></summary>

**Answer / theory**

Vectorization expresses a computation as array operations instead of explicit Python loops. For example, `x * 2 + 1` transforms every element through compiled array kernels.

**Practical note:** Vectorization improves clarity and often speed; it does not mean the algorithm has zero computational cost.
</details>

<details>
<summary><strong>Q010. What is broadcasting?</strong></summary>

**Answer / theory**

Broadcasting lets arrays with compatible shapes participate in elementwise operations without manually replicating data. Compare dimensions from the right: dimensions must match or one must be 1.

**Practical note:** Shapes `(3,1)` and `(1,4)` broadcast to `(3,4)`.
</details>

<details>
<summary><strong>Q011. What are the broadcasting rules?</strong></summary>

**Answer / theory**

Align dimensions from the trailing side. Two dimensions are compatible if they are equal or either is 1; missing leading dimensions are treated as 1. Otherwise NumPy raises a shape-mismatch error.

**Practical note:** Always reason from shapes before debugging a broadcasting exception.
</details>

<details>
<summary><strong>Q012. What is axis in NumPy?</strong></summary>

**Answer / theory**

An axis identifies a dimension. For a 2-D array, `axis=0` reduces rows down each column; `axis=1` reduces columns across each row.

**Practical note:** A reduction removes the selected axis unless `keepdims=True`.
</details>

<details>
<summary><strong>Q013. Explain axis=0 versus axis=1.</strong></summary>

**Answer / theory**

For a matrix with shape `(rows, columns)`, `sum(axis=0)` returns one sum per column, shape `(columns,)`; `sum(axis=1)` returns one sum per row, shape `(rows,)`.

**Practical note:** Use axis language rather than assuming axis 0 means 'horizontal'.
</details>

<details>
<summary><strong>Q014. What is reshaping?</strong></summary>

**Answer / theory**

`reshape` changes the dimensions while preserving element order. The total number of elements must remain constant, except one dimension may be inferred with `-1`.

**Practical note:** For 12 elements, `(3,4)`, `(2,6)`, and `(2,2,3)` are valid shapes.
</details>

<details>
<summary><strong>Q015. What is the difference between reshape and resize?</strong></summary>

**Answer / theory**

`reshape` returns a view when possible, otherwise a copy; it does not alter the original shape. `ndarray.resize` changes the array object in place under constraints, and `np.resize` can repeat data to fill a requested shape.

**Practical note:** Prefer `reshape` for predictable transformations.
</details>

<details>
<summary><strong>Q016. What does transpose do?</strong></summary>

**Answer / theory**

Transpose permutes axes. For a 2-D matrix, `.T` swaps rows and columns; for higher dimensions, `transpose(axes)` specifies the new axis order.

**Practical note:** A transpose is often a view with changed strides, not a physical data reorder.
</details>

<details>
<summary><strong>Q017. What is flatten versus ravel?</strong></summary>

**Answer / theory**

`flatten()` always returns a copy. `ravel()` returns a view when possible and otherwise a copy.

**Practical note:** Choose `ravel` when avoiding a copy matters, but do not assume it always shares memory.
</details>

<details>
<summary><strong>Q018. What is a view?</strong></summary>

**Answer / theory**

A view is a new array object that refers to the same underlying data buffer. Mutating shared elements through a view can change the original array.

**Practical note:** Use `np.shares_memory` to test memory overlap when correctness depends on it.
</details>

<details>
<summary><strong>Q019. What is a copy?</strong></summary>

**Answer / theory**

A copy owns independent data. Mutating the copy does not mutate the original array's elements.

**Practical note:** `a.copy()` is the usual explicit way to create an independent data copy.
</details>

<details>
<summary><strong>Q020. Explain basic slicing.</strong></summary>

**Answer / theory**

Basic slicing uses slices such as `a[1:4]`, `a[:, 0]`, and `a[::2]`. It commonly returns views for ndarrays, so writes can affect the original.

**Practical note:** Remember stop is exclusive: `a[start:stop:step]`.
</details>

<details>
<summary><strong>Q021. What is fancy indexing?</strong></summary>

**Answer / theory**

Integer-array indexing selects specified positions, e.g. `a[[0, 3, 5]]`. It returns a copy rather than a view in ordinary advanced-indexing cases.

**Practical note:** Repeated indices can behave unexpectedly for in-place updates; use `np.add.at` when repeated-index accumulation is required.
</details>

<details>
<summary><strong>Q022. What is boolean indexing?</strong></summary>

**Answer / theory**

A boolean mask selects values where the mask is true, e.g. `a[a > 0]`. The mask usually must match the indexed dimensions or be broadcastable under supported indexing rules.

**Practical note:** Boolean selection returns a copy of selected elements.
</details>

<details>
<summary><strong>Q023. How do you filter values?</strong></summary>

**Answer / theory**

Build a boolean condition such as `(a >= lower) & (a <= upper)`, then use `a[mask]`. Use `&`, `|`, and `~` with parentheses, not Python `and`/`or` for elementwise masks.

**Practical note:** For complex predicates, name the mask to improve readability.
</details>

<details>
<summary><strong>Q024. Why use parentheses with boolean masks?</strong></summary>

**Answer / theory**

Python operator precedence can evaluate comparisons and bitwise operators in an unintended order. Parenthesize each comparison before combining with `&` or `|`.

**Practical note:** Correct: `(a > 2) & (a < 8)`.
</details>

<details>
<summary><strong>Q025. How do you add or remove dimensions?</strong></summary>

**Answer / theory**

Use `np.newaxis`/`None`, `np.expand_dims`, or `np.squeeze`. Adding an axis changes shape, not the number of elements; squeeze removes axes of length 1.

**Practical note:** `a[:, None]` converts shape `(n,)` to `(n,1)`.
</details>

<details>
<summary><strong>Q026. What is concatenation versus stacking?</strong></summary>

**Answer / theory**

`concatenate` joins arrays along an existing axis. `stack` creates a new axis and requires the input shapes to match.

**Practical note:** For arrays of shape `(2,3)`, stacking two arrays on axis 0 gives `(2,2,3)`.
</details>

<details>
<summary><strong>Q027. What is hstack, vstack, and dstack?</strong></summary>

**Answer / theory**

These convenience functions concatenate along conventions corresponding roughly to horizontal, vertical, or depth placement. Their behavior depends on input dimensionality.

**Practical note:** For predictable shapes, `concatenate` or `stack` often communicates intent more clearly.
</details>

<details>
<summary><strong>Q028. How do you split an array?</strong></summary>

**Answer / theory**

Use `split`, `array_split`, `hsplit`, `vsplit`, or `dsplit`. `split` requires equal divisions; `array_split` permits uneven partitions.

**Practical note:** Check the output shapes, especially when the length is not divisible by the number of sections.
</details>

<details>
<summary><strong>Q029. What are universal functions (ufuncs)?</strong></summary>

**Answer / theory**

Ufuncs are elementwise operations such as `np.add`, `np.sqrt`, and `np.exp`. They support broadcasting, type handling, and often optional `out` and `where` parameters.

**Practical note:** Ufuncs are the building blocks for many vectorized computations.
</details>

<details>
<summary><strong>Q030. What are reductions?</strong></summary>

**Answer / theory**

Reductions combine values along one or more axes: `sum`, `mean`, `min`, `max`, `prod`, `std`, `var`, `any`, and `all`.

**Practical note:** Use `axis` to control aggregation and `keepdims=True` to preserve reduced dimensions.
</details>

<details>
<summary><strong>Q031. How do mean, variance, and standard deviation relate?</strong></summary>

**Answer / theory**

Population variance is the mean squared deviation: `var = mean((x - mean(x))**2)`. Standard deviation is its square root. NumPy's default `ddof=0` uses population variance; `ddof=1` gives the sample estimate.

**Practical note:** Use `ddof=1` for the common unbiased sample-variance estimator.
</details>

<details>
<summary><strong>Q032. What is argmax versus max?</strong></summary>

**Answer / theory**

`max` returns the maximum value; `argmax` returns the index of the first maximum along the selected axis.

**Practical note:** To retrieve both, use the index from `argmax` to index the array.
</details>

<details>
<summary><strong>Q033. What do percentile and quantile mean?</strong></summary>

**Answer / theory**

A quantile partitions a distribution at a probability in `[0,1]`; a percentile typically uses a 0–100 scale. `np.quantile(x, .9)` corresponds to the 90th percentile.

**Practical note:** Interpolation/method options can affect results for finite samples.
</details>

<details>
<summary><strong>Q034. How do you handle NaN?</strong></summary>

**Answer / theory**

NaN represents a missing/undefined floating-point value and propagates through many arithmetic operations. Use `np.isnan`, `np.nanmean`, `np.nansum`, or explicit masks as appropriate.

**Practical note:** Do not compare NaN using `==`; use `np.isnan(x)`.
</details>

<details>
<summary><strong>Q035. What is the difference between NaN and None?</strong></summary>

**Answer / theory**

`None` is a Python object and can force object dtype in arrays. `np.nan` is a floating-point NaN value. Integer arrays cannot directly represent NaN without changing representation or using a nullable scheme.

**Practical note:** Use a deliberate missing-data strategy rather than mixing arbitrary Python objects into numeric arrays.
</details>

<details>
<summary><strong>Q036. What are infinities and how do you test them?</strong></summary>

**Answer / theory**

`np.inf` and `-np.inf` represent positive and negative infinity. Use `np.isinf` to detect them and `np.isfinite` to select finite values.

**Practical note:** Check for infinities after division, logarithms, or unstable calculations.
</details>

<details>
<summary><strong>Q037. How do you avoid divide-by-zero warnings?</strong></summary>

**Answer / theory**

Use masks or `np.divide(a, b, out=..., where=b != 0)`. `np.errstate` can control warnings, but should not replace correct mathematical handling.

**Practical note:** Decide what zero denominator means in the application: NaN, zero, infinity, or an exception.
</details>

<details>
<summary><strong>Q038. What is dtype promotion?</strong></summary>

**Answer / theory**

NumPy chooses a result dtype based on operand dtypes and operation rules. Mixing integer and floating-point arrays often produces floating-point results; exact rules depend on NumPy version and scalar/array types.

**Practical note:** Inspect `result.dtype`; explicitly cast when precision or API contracts require it.
</details>

<details>
<summary><strong>Q039. Why can integer overflow occur?</strong></summary>

**Answer / theory**

Fixed-width integer dtypes have finite ranges. Arithmetic beyond the representable range can wrap or otherwise follow dtype-specific behavior rather than arbitrary-precision Python integer behavior.

**Practical note:** Use a wider dtype or Python integers when the range is not known.
</details>

<details>
<summary><strong>Q040. What is float precision?</strong></summary>

**Answer / theory**

Floating-point values use finite binary precision. Many decimal fractions cannot be represented exactly, so equality comparisons can be fragile.

**Practical note:** Use `np.isclose(a, b, rtol=..., atol=...)` for approximate numerical equality.
</details>

<details>
<summary><strong>Q041. What is np.isclose/allclose?</strong></summary>

**Answer / theory**

`isclose` compares elementwise within relative and absolute tolerances; `allclose` checks whether all paired elements are close.

**Practical note:** Tolerance should reflect the scale and error budget of the problem.
</details>

<details>
<summary><strong>Q042. What are strides?</strong></summary>

**Answer / theory**

Strides describe how many bytes to move in each dimension to reach the next element. They explain why transposes and slices can be views without moving data.

**Practical note:** Non-contiguous views can be slower for some operations and may require copies in downstream libraries.
</details>

<details>
<summary><strong>Q043. What is C versus Fortran memory order?</strong></summary>

**Answer / theory**

C order makes the last index vary fastest; Fortran order makes the first index vary fastest. Order matters for memory layout and some external libraries.

**Practical note:** Use `np.ascontiguousarray` or `np.asfortranarray` when a consumer requires a layout.
</details>

<details>
<summary><strong>Q044. What is a contiguous array?</strong></summary>

**Answer / theory**

An array is contiguous when its elements occupy the expected adjacent memory layout under C or Fortran ordering. Slices with steps and transposes may be non-contiguous.

**Practical note:** Check `a.flags.c_contiguous` and `a.flags.f_contiguous`.
</details>

<details>
<summary><strong>Q045. What is a structured array?</strong></summary>

**Answer / theory**

A structured dtype gives each record named fields with potentially different types, similar to a low-level record layout.

**Practical note:** Structured arrays are useful for binary formats/interoperability, but pandas is often more convenient for analytics.
</details>

<details>
<summary><strong>Q046. What is an object array?</strong></summary>

**Answer / theory**

An object-dtype array stores Python object references instead of a homogeneous primitive numeric buffer. Many operations lose the speed advantages of numeric arrays.

**Practical note:** Avoid object dtype in performance-sensitive numeric code where possible.
</details>

<details>
<summary><strong>Q047. How do you generate random numbers correctly?</strong></summary>

**Answer / theory**

Use `rng = np.random.default_rng(seed)` and methods such as `rng.integers`, `rng.normal`, and `rng.choice`. A seeded generator supports reproducible workflows.

**Practical note:** Record the seed and generator/version context for reproducibility.
</details>

<details>
<summary><strong>Q048. Why prefer default_rng over the legacy global RNG?</strong></summary>

**Answer / theory**

`default_rng` provides an explicit generator object, reducing hidden global state and making random streams easier to manage.

**Practical note:** Pass a generator into functions that need randomness.
</details>

<details>
<summary><strong>Q049. What is a seed?</strong></summary>

**Answer / theory**

A seed initializes a pseudorandom generator's state. The same seed with the same generator and environment generally reproduces the same sequence.

**Practical note:** Seeds improve reproducibility but do not make a model statistically valid or cryptographically secure.
</details>

<details>
<summary><strong>Q050. How do you sample without replacement?</strong></summary>

**Answer / theory**

Use `rng.choice(population, size=k, replace=False)`. The requested sample size cannot exceed the population size.

**Practical note:** Use replacement only when repeated selections are intended.
</details>

<details>
<summary><strong>Q051. What is a permutation?</strong></summary>

**Answer / theory**

`rng.permutation` returns a permuted copy/order, while `rng.shuffle` shuffles an array in place along its first axis.

**Practical note:** Be careful: shuffle mutates its input.
</details>

<details>
<summary><strong>Q052. How do you compute a dot product?</strong></summary>

**Answer / theory**

For vectors, `np.dot(a,b)` or `a @ b` computes the sum of pairwise products. For matrices, `@` performs matrix multiplication under linear-algebra rules.

**Practical note:** For vectors, `a·b = Σ a_i b_i`; output shape depends on operand dimensions.
</details>

<details>
<summary><strong>Q053. What is matrix multiplication versus elementwise multiplication?</strong></summary>

**Answer / theory**

`a * b` multiplies corresponding elements; `a @ b` performs linear algebra matrix multiplication. Their shape requirements differ.

**Practical note:** For matrices `(m,n)` and `(n,p)`, `@` returns `(m,p)`.
</details>

<details>
<summary><strong>Q054. What is a transpose in linear algebra?</strong></summary>

**Answer / theory**

The transpose swaps matrix rows and columns. For a real matrix, `A.T` is the transpose; for complex values, `A.conj().T` is the conjugate transpose.

**Practical note:** Many formulas for covariance, least squares, and projections use transposes.
</details>

<details>
<summary><strong>Q055. What is a norm?</strong></summary>

**Answer / theory**

A norm measures vector/matrix magnitude. `np.linalg.norm(x)` defaults to the Euclidean norm for a vector; options can specify other orders or axes.

**Practical note:** L2 norm: `sqrt(Σ x_i²)`; L1 norm: `Σ |x_i|`.
</details>

<details>
<summary><strong>Q056. How do you solve a linear system?</strong></summary>

**Answer / theory**

Use `np.linalg.solve(A, b)` for `A x = b` when `A` is square and nonsingular. It is generally preferable to explicitly computing `inv(A) @ b`.

**Practical note:** For overdetermined systems, consider `np.linalg.lstsq`.
</details>

<details>
<summary><strong>Q057. Why avoid matrix inversion for solving equations?</strong></summary>

**Answer / theory**

Explicit inversion can be less numerically stable and often does more work than a direct solve. `solve` uses optimized factorization-based methods.

**Practical note:** Use inverse only when the inverse matrix itself is required.
</details>

<details>
<summary><strong>Q058. What is determinant?</strong></summary>

**Answer / theory**

The determinant is a scalar describing properties of a square matrix, including singularity. A zero determinant indicates a singular matrix in exact arithmetic.

**Practical note:** For numerical stability, do not use determinant alone as a condition test.
</details>

<details>
<summary><strong>Q059. What are eigenvalues and eigenvectors?</strong></summary>

**Answer / theory**

An eigenvector `v` satisfies `A v = λ v`; `λ` is its eigenvalue. They reveal invariant directions and are used in stability, PCA, and dynamical systems.

**Practical note:** Use `np.linalg.eig` for general square matrices and `eigh` for real symmetric/Hermitian matrices.
</details>

<details>
<summary><strong>Q060. What is SVD?</strong></summary>

**Answer / theory**

Singular value decomposition factors a matrix as `A = U Σ Vᴴ`. Singular values quantify directions of strength and support low-rank approximation, denoising, and least squares.

**Practical note:** Use `np.linalg.svd`; `full_matrices=False` can reduce memory for rectangular matrices.
</details>

<details>
<summary><strong>Q061. What is matrix rank?</strong></summary>

**Answer / theory**

Rank is the number of linearly independent rows/columns. Numerically, rank is estimated using a tolerance based on singular values.

**Practical note:** Rank estimates can change with scaling and tolerance.
</details>

<details>
<summary><strong>Q062. What is condition number?</strong></summary>

**Answer / theory**

A condition number measures sensitivity of a problem to small input perturbations. A large condition number indicates an ill-conditioned system.

**Practical note:** `np.linalg.cond(A)` helps diagnose why small input errors cause large solution errors.
</details>

<details>
<summary><strong>Q063. How do you calculate covariance?</strong></summary>

**Answer / theory**

Covariance measures joint variation: positive covariance means variables tend to move together, negative means opposite movement. `np.cov` defaults to variables in rows (`rowvar=True`).

**Practical note:** For observations as rows and features as columns, use `np.cov(X, rowvar=False)`.
</details>

<details>
<summary><strong>Q064. What is correlation?</strong></summary>

**Answer / theory**

Correlation standardizes covariance to a dimensionless scale, commonly between -1 and 1. `np.corrcoef` estimates the correlation matrix.

**Practical note:** Correlation measures linear association, not causation.
</details>

<details>
<summary><strong>Q065. What is a histogram?</strong></summary>

**Answer / theory**

`np.histogram` bins values into intervals and returns bin counts plus edges. It is useful for distributions and feature diagnostics.

**Practical note:** Counts sum to the number of included observations unless density mode or range filtering changes interpretation.
</details>

<details>
<summary><strong>Q066. How do you normalize an array?</strong></summary>

**Answer / theory**

Normalization has multiple meanings. Min-max scaling maps values to a range; standardization subtracts mean and divides by standard deviation; vector normalization divides by a norm.

**Practical note:** Choose the formula based on the downstream task and fit any data-dependent parameters on training data only.
</details>

<details>
<summary><strong>Q067. What is min-max scaling?</strong></summary>

**Answer / theory**

For a feature `x`, min-max scaling is `(x - min(x))/(max(x)-min(x))`, usually mapping the training range to `[0,1]`.

**Practical note:** If max equals min, handle the zero denominator explicitly.
</details>

<details>
<summary><strong>Q068. What is standardization?</strong></summary>

**Answer / theory**

Standardization uses `z=(x-μ)/σ`, where μ and σ are estimated from the training data. It centers features and scales their variance.

**Practical note:** For ML, reuse training-set μ and σ on validation/test/production data.
</details>

<details>
<summary><strong>Q069. How do you compute pairwise distances?</strong></summary>

**Answer / theory**

For rows as points, Euclidean distance is `sqrt(Σ_j (x_ij-y_kj)^2)`. Broadcasting can compute pairwise differences, but intermediate arrays may be large.

**Practical note:** For large datasets, use chunking or specialized distance routines to control memory.
</details>

<details>
<summary><strong>Q070. What is a sliding window?</strong></summary>

**Answer / theory**

A sliding window creates overlapping views of local neighborhoods, useful for time series and signal processing. `sliding_window_view` can avoid copying the entire windowed array.

**Practical note:** Window views can be large logically; avoid materializing huge intermediate arrays.
</details>

<details>
<summary><strong>Q071. What is vectorized masking in real projects?</strong></summary>

**Answer / theory**

A mask represents a condition for every element and can drive filtering, replacement, or conditional selection without Python loops.

**Practical note:** `np.where(condition, x, y)` selects values elementwise.
</details>

<details>
<summary><strong>Q072. What is np.select?</strong></summary>

**Answer / theory**

`np.select` chooses from multiple result choices according to ordered conditions, with a default for unmatched entries.

**Practical note:** It can replace nested vectorized conditional logic.
</details>

<details>
<summary><strong>Q073. What is np.clip?</strong></summary>

**Answer / theory**

`np.clip(a, low, high)` bounds values below/above specified thresholds. It is useful for limiting predictions, sensor ranges, and pixel values.

**Practical note:** Clipping changes data; it should reflect a real domain constraint.
</details>

<details>
<summary><strong>Q074. What are cumulative operations?</strong></summary>

**Answer / theory**

`cumsum`, `cumprod`, `maximum.accumulate`, and related methods produce running aggregates along an axis.

**Practical note:** For a time series, cumulative sum is `S_t = Σ_{i≤t} x_i`.
</details>

<details>
<summary><strong>Q075. What is a rolling mean in NumPy?</strong></summary>

**Answer / theory**

A rolling mean averages a moving window. One approach uses `sliding_window_view`; another uses cumulative sums for efficient window sums.

**Practical note:** A cumulative-sum method can compute many window sums in O(n), after an O(n) cumulative pass.
</details>

<details>
<summary><strong>Q076. How do you sort arrays?</strong></summary>

**Answer / theory**

`np.sort` returns a sorted copy; `a.sort()` sorts in place. `argsort` returns indices that would sort the data.

**Practical note:** Choose stable sorting when equal-key ordering matters.
</details>

<details>

**Practical note:** This pattern is foundational in probabilistic models and softmax/log-likelihood calculations.

3. Check whether a `(n,)` vector was unintentionally broadcast against `(n, 1)`, producing `(n, n)`.
4. Add regression tests for one row, multiple rows, and empty inputs.
</details>

<details>
<summary><strong>Scenario 2: A numerical pipeline produces NaN or infinity</strong></summary>

1. Inspect `np.isfinite(X)` and identify the first invalid stage.
2. Check division by zero, logarithms of nonpositive values, overflow, and invalid input ranges.
3. Define an explicit policy: reject, impute, mask, or cap invalid values.
4. Do not silently replace every invalid value with zero without a domain justification.
</details>

<details>
<summary><strong>Scenario 3: Memory usage spikes after vectorization</strong></summary>

1. Estimate raw bytes with `array.nbytes`.
2. Count temporary arrays created by broadcasting and chained operations.
3. Use in-place operations only when aliasing/mutation is safe.
4. Chunk the computation or use memory mapping for larger-than-RAM data.
</details>

<details>
<summary><strong>Scenario 4: Training and production predictions differ</strong></summary>

1. Reuse the training mean/std/min/max in production.
2. Ensure dtype, feature order, units, missing-value rules, and shape are identical.
3. Version preprocessing parameters with the model.
4. Add parity tests using known examples.
</details>

<details>
<summary><strong>Scenario 5: Model metrics look suspiciously good</strong></summary>

1. Check for data leakage before normalization, feature selection, or resampling.
2. Split train/validation/test before fitting data-dependent transformations.
3. Check target/feature alignment and array broadcasting.
4. Compare against a simple baseline and inspect per-class metrics.
</details>

## Part 4 — Interview Formula Sheet

| Topic | Formula |
|---|---|
| Mean | `sum(x) / n` |
| Population variance | `mean((x - mean(x))**2)` |
| Standardization | `(x - mean) / std` |
| Min-max scaling | `(x - min) / (max - min)` |
| Dot product | `sum(a * b)` |
| Euclidean distance | `sqrt(sum((x-y)**2))` |
| MSE | `mean((y - y_pred)**2)` |
| MAE | `mean(abs(y - y_pred))` |
| Precision | `TP / (TP + FP)` |
| Recall | `TP / (TP + FN)` |
| Softmax | `exp(z_i) / sum(exp(z))` |
| Monte Carlo standard error for a sample mean | `sample_std / sqrt(n)` |
