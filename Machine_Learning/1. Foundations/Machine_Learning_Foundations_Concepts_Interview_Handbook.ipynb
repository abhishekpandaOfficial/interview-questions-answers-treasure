## 1. 🧩 The machine learning mental model


<details open style="border:1px solid #7c3aed; border-left:6px solid #7c3aed; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f5f3ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#4c1d95; font-size:1.04rem;"><b>1.1 What is machine learning?</b></summary>

**Machine learning (ML)** is a field of computing in which a system uses data to learn patterns or relationships that help it make predictions, decisions, rankings, representations, or generated outputs on previously unseen examples.

A useful conceptual abstraction is:

\[
\text{Data} + \text{Learning procedure} \longrightarrow \text{Fitted model}
\]

The fitted model is then used on new inputs:

\[
\text{New input} \longrightarrow \text{Model} \longrightarrow \text{Prediction / score / action}
\]

**Important distinction:** training is the process of estimating a model from data; inference is using the fitted model to produce outputs. A model can perform well on its training data and still fail on new data.

<div style="border:1px solid #93c5fd;background:#eff6ff;border-radius:12px;padding:14px">
<b>Think of ML as a system, not a magic model.</b><br/>
The quality of the outcome depends on problem framing, data, labels, evaluation, deployment conditions, and monitoring—not just the model family.
</div>

</details>


<details style="border:1px solid #059669; border-left:6px solid #059669; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#ecfdf5;">
<summary style="cursor:pointer; padding:14px 4px; color:#064e3b; font-size:1.04rem;"><b>1.2 The core vocabulary</b></summary>

| Term | Meaning | Example |
|---|---|---|
| **Example / observation / sample** | One record or case | One customer transaction |
| **Feature / predictor / input variable** | Information used as input | Transaction amount |
| **Target / label / outcome** | What the model should predict in supervised learning | Fraud / not fraud |
| **Dataset** | Collection of examples | Historical transactions |
| **Model** | A parameterized or structured mapping from inputs to outputs | A trained fraud scorer |
| **Parameter** | A value learned during fitting | A model coefficient or weight |
| **Hyperparameter** | A configuration chosen outside ordinary parameter fitting | Tree depth limit |
| **Training** | Fitting model parameters using training data | Learning from historical examples |
| **Inference** | Applying a fitted model to an input | Scoring a new transaction |
| **Prediction** | The model output | Predicted class or estimated amount |
| **Score** | A numerical output used for ranking or thresholding | Fraud risk score |
| **Loss** | Quantity optimized during fitting | A penalty for incorrect predictions |
| **Metric** | Quantity used to assess performance | Recall, MAE, log loss |
| **Baseline** | Simple reference performance | Predict the majority class |
| **Generalization** | Performance on genuinely unseen, relevant data | Performance next month |
| **Artifact** | Saved model or supporting object needed for use | Model file plus preprocessing pipeline |

## 2. 🎯 Learning types and problem framing


<details open style="border:1px solid #db2777; border-left:6px solid #db2777; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fdf2f8;">
<summary style="cursor:pointer; padding:14px 4px; color:#831843; font-size:1.04rem;"><b>2.1 Supervised, unsupervised, semi-supervised, and self-supervised learning</b></summary>

| Learning setup | What the data provides | Typical objective | Example |
|---|---|---|---|
| **Supervised** | Inputs paired with target labels | Predict a known outcome | Predict delivery time from past deliveries |
| **Unsupervised** | Inputs without a target label | Discover structure or summarize data | Group similar products |
| **Semi-supervised** | A small labeled set plus a larger unlabeled set | Use both sources of information | Classify images when labeling is expensive |
| **Self-supervised** | Targets constructed from the data itself | Learn useful representations or pretraining signals | Predict masked tokens |
| **Reinforcement learning** | Feedback/reward from actions in an environment | Learn a policy for sequential decisions | A simulated agent learning control |
| **Online / continual learning** | Data arrives over time and model updates may be incremental | Adapt to a changing stream | Updating a model as fresh observations arrive |

**Do not confuse learning setup with model family.** A model family may support more than one training setup, and a product may combine several setups.
</details>


<details style="border:1px solid #0d9488; border-left:6px solid #0d9488; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f0fdfa;">
<summary style="cursor:pointer; padding:14px 4px; color:#134e4a; font-size:1.04rem;"><b>2.2 Classification, regression, ranking, clustering, forecasting, and representation learning</b></summary>

- **Classification:** predict a discrete class or class probabilities. Examples: spam detection, defect category.
- **Regression:** estimate a numerical quantity. Examples: demand amount, house price.
- **Ranking:** order candidates by relevance or utility. Examples: search results, recommendations.
- **Clustering:** organize examples by similarity without a supplied target label. Examples: exploratory customer segmentation.
- **Forecasting:** predict future values using time-ordered information. Examples: hourly load, weekly demand.
- **Anomaly detection:** identify unusual observations relative to a reference pattern. Examples: sensor failures, suspicious activity.
- **Dimensionality reduction:** represent data with fewer dimensions while retaining selected structure.
- **Representation learning:** learn features or embeddings useful for downstream tasks.
- **Generation:** produce new content or samples. Often uses deep learning, but the problem framing and evaluation still matter.
- **Decision / policy learning:** select actions under objectives and constraints; the decision may depend on costs, rewards, and feedback.

**Problem framing questions:** What is the unit of prediction? What is the target? At what time is a prediction made? What information is available then? What action follows the prediction? What is the cost of each kind of error?
</details>


<details style="border:1px solid #2563eb; border-left:6px solid #2563eb; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#eff6ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#1e3a8a; font-size:1.04rem;"><b>2.3 Translate a business request into an ML task</b></summary>

A request such as “reduce customer churn” is not yet a complete ML specification.

1. **Decision:** What action will the organization take?
2. **Prediction unit:** One customer, account, subscription, or customer-month?
3. **Target definition:** What precisely counts as churn?
4. **Prediction horizon:** Churn within 7, 30, or 90 days?
5. **Observation window:** What historical period can be used as input?
6. **Availability time:** Would each feature exist at the moment the prediction is made?
7. **Intervention:** Can the business act on the score in time?
8. **Success criterion:** Is the goal recall, precision, incremental retention, profit, or another outcome?
9. **Constraints:** Latency, budget, fairness, privacy, interpretability, and operational capacity.
10. **Evaluation design:** Does the test set represent future customers and future conditions?

<div style="background:#fff7ed;border-left:5px solid #f97316;padding:12px;border-radius:8px">
<b>Interview signal:</b> Strong candidates clarify the decision and data-generating process before naming an algorithm.
</div>
</details>


<details style="border:1px solid #7c3aed; border-left:6px solid #7c3aed; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f5f3ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#4c1d95; font-size:1.04rem;"><b>2.4 Correlation, prediction, and causation</b></summary>

- **Association / correlation:** two variables vary together in the observed data.
- **Prediction:** use available information to estimate an unknown or future outcome.
- **Causation:** estimate what changes when an intervention is applied.

Predictive accuracy alone does **not** establish that a feature causes an outcome. A feature can be predictive because it is a proxy, a consequence of the target, or correlated with an unobserved factor. If the goal is to estimate intervention effects, the design must address confounding, selection, and the causal question.

</details>


## 3. 🗃️ Data foundations and dataset design


<details open style="border:1px solid #059669; border-left:6px solid #059669; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#ecfdf5;">
<summary style="cursor:pointer; padding:14px 4px; color:#064e3b; font-size:1.04rem;"><b>3.1 Data types and structures</b></summary>

| Data type | Examples | Considerations |
|---|---|---|
| Numerical, continuous | Height, temperature, spend | Units, skew, outliers, scale |
| Numerical, discrete | Counts, number of visits | Zero meaning, count distribution |
| Categorical, nominal | Country, product type | Categories, rare values, unseen categories |
| Categorical, ordinal | Low / medium / high | Meaningful order, not necessarily equal spacing |
| Binary | Yes / no | Class imbalance and error costs |
| Time series | Sensor values over time | Order, seasonality, leakage across time |
| Text | Reviews, documents | Language, length, privacy, tokenization |
| Image / video | Medical scans, clips | Resolution, labeling quality, provenance |
| Audio | Speech, sound events | Sampling rate, noise, timing |
| Graph / relational | Users and connections | Nodes, edges, dependencies, split design |
| Geospatial | Coordinates, regions | Spatial dependence, privacy, geography shift |

Data can be structured, semi-structured, or unstructured. The type determines useful validation and representation choices; it does not by itself determine the best model.
</details>


<details style="border:1px solid #ea580c; border-left:6px solid #ea580c; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fff7ed;">
<summary style="cursor:pointer; padding:14px 4px; color:#7c2d12; font-size:1.04rem;"><b>3.2 Dataset partitions: train, validation, and test</b></summary>

- **Training set:** used to fit model parameters.
- **Validation set:** used for model selection, hyperparameter decisions, threshold selection, and iteration.
- **Test set:** reserved for a final, comparatively unbiased estimate after choices are complete.

![Training, validation, and test split workflow](diagrams/data-split-workflow.svg)

The correct split depends on how observations are generated. Random splitting can be inappropriate for repeated people, grouped entities, time series, or spatially dependent observations. In such cases, use group-, time-, or domain-aware evaluation.

**Repeatedly checking the test set turns it into a de facto validation set.** If test results influence choices, the test estimate can become optimistic.
</details>


<details style="border:1px solid #db2777; border-left:6px solid #db2777; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fdf2f8;">
<summary style="cursor:pointer; padding:14px 4px; color:#831843; font-size:1.04rem;"><b>3.3 Sampling, representativeness, and label quality</b></summary>

- **Population:** the broader set about which decisions are intended.
- **Sample:** the observed subset used for analysis or training.
- **Sampling bias:** selection causes the observed sample to differ systematically from the target population.
- **Selection bias:** inclusion in the data depends on factors related to the outcome or features.
- **Label noise:** labels are incorrect, ambiguous, inconsistent, or weakly measured.
- **Class imbalance:** some target classes occur much less often than others.
- **Coverage:** how well the dataset spans relevant users, conditions, classes, and edge cases.
- **Data provenance:** where data came from, how it was collected, and what permissions govern its use.
- **Data lineage:** traceable transformations and dependencies from source to model input.

A very large dataset can still be poor if it is unrepresentative, mislabeled, duplicated, or collected under conditions unlike production.
</details>


<details style="border:1px solid #0d9488; border-left:6px solid #0d9488; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f0fdfa;">
<summary style="cursor:pointer; padding:14px 4px; color:#134e4a; font-size:1.04rem;"><b>3.4 Missing data and outliers: concepts before methods</b></summary>

**Missingness mechanisms**
- **MCAR — Missing Completely At Random:** missingness is independent of observed and unobserved values, under the formal model.
- **MAR — Missing At Random:** missingness may depend on observed variables but not on the missing value after conditioning on them.
- **MNAR — Missing Not At Random:** missingness depends on the unobserved value or other unobserved factors.

These assumptions are substantive and often cannot be proven from observed data alone.

**Outliers** may be data errors, valid rare events, or signs of a different process. Investigate provenance and domain meaning before deleting them. An outlier is not automatically noise.

**Duplicates** can distort evaluation when near-identical records cross the train/test boundary. Deduplication rules must preserve legitimate repeated events where repetition is meaningful.
</details>


<details style="border:1px solid #2563eb; border-left:6px solid #2563eb; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#eff6ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#1e3a8a; font-size:1.04rem;"><b>3.5 Data contracts and validation</b></summary>

A data contract defines expectations for fields and records, such as schema, types, valid ranges, allowed categories, freshness, null rates, uniqueness, and relationships. Validation catches upstream changes before they silently affect predictions.

Examples of checks:
- Required columns exist and have the expected types.
- Timestamps are in the correct timezone and order.
- Numeric values fall within plausible ranges.
- Categories are known or handled explicitly.
- Target labels follow a documented definition.
- Features are available at prediction time.
- Training and inference use compatible transformations.
- Privacy and usage permissions are documented.

</details>

<summary style="cursor:pointer; padding:14px 4px; color:#064e3b; font-size:1.04rem;"><b>4.2 What happens at each stage—and why?</b></summary>

| Stage | Main work | Typical deliverable | Common failure |
|---|---|---|---|
| Problem definition | Define decision, unit, target, horizon, constraints | Problem statement and acceptance criteria | Optimizing a metric unrelated to business value |
| Data discovery | Find sources, permissions, coverage, label process | Data inventory and lineage | Training on data that cannot legally or operationally be used |
| Data understanding | Profile quality, distributions, missingness, duplicates | Data quality report | Assuming the dataset is representative |
| Split design | Select random, temporal, grouped, or other valid split | Reproducible split specification | Leakage or unrealistic test conditions |
| Baseline | Compare with a simple rule or trivial predictor | Baseline metric and error profile | Claiming success without a reference point |
| Preparation | Fit preprocessing correctly within training workflow | Reusable preprocessing pipeline | Fitting transformations on held-out data |
| Model selection | Compare candidates using validation protocol | Experiment results | Repeatedly tuning to the test set |
| Error analysis | Inspect slices, failures, uncertainty, and costs | Error taxonomy and follow-up plan | Looking only at one aggregate metric |
| Final evaluation | Evaluate frozen approach on held-out test data | Final evaluation report | Reporting the best result after many test attempts |
| Packaging | Version model, transformations, dependencies, schema | Reproducible inference artifact | Training/inference mismatch |
| Deployment | Integrate into a product or decision process | Service, batch job, or embedded model | Ignoring latency, fallback, access, and rollback |
| Monitoring | Track input quality, drift, latency, failures, outcomes | Alerts and review dashboards | Assuming offline performance lasts forever |
| Maintenance | Investigate, retrain, roll back, or retire | Change record and new version | Retraining automatically without validating benefit |

</details>


<details style="border:1px solid #ea580c; border-left:6px solid #ea580c; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fff7ed;">
<summary style="cursor:pointer; padding:14px 4px; color:#7c2d12; font-size:1.04rem;"><b>4.3 Baselines and error analysis</b></summary>

A baseline is a reference point—not necessarily a weak model. Examples include:
- A domain rule already used by the business.
- Predicting a majority class.
- Predicting a historical average or last observed value.
- A simple, interpretable model.
- The existing human or operational process.

Compare candidates against the baseline using the same split, target definition, and evaluation protocol. Also compare **cost, latency, stability, explainability, maintenance burden, and failure modes**.

Error analysis means studying where and why a system fails:
- Which classes or groups have higher error?
- Are errors concentrated at particular times, geographies, or input ranges?
- Are labels ambiguous?
- Are failures caused by missing inputs or unseen categories?
- Is the metric hiding severe errors in a small but important segment?
- Does a prediction error actually change the downstream decision?

</details>


<details style="border:1px solid #db2777; border-left:6px solid #db2777; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fdf2f8;">
<summary style="cursor:pointer; padding:14px 4px; color:#831843; font-size:1.04rem;"><b>4.4 Experiment tracking and reproducibility</b></summary>

Record enough information to reconstruct why a result occurred:
- Dataset version and split definition.
- Code revision and environment/dependency versions.
- Feature and preprocessing definitions.
- Model family and configuration.
- Random seeds where relevant (seeds improve repeatability but do not guarantee determinism in every environment).
- Metrics, slice results, plots, and evaluation protocol.
- Model artifact location, owner, timestamp, and approval status.

**Reproducibility** means being able to repeat the procedure under documented conditions. **Repeatability** often refers to obtaining consistent results under the same setup; exact terminology can vary by discipline.
</details>


## 5. 📏 Evaluation, validation, and metrics


<details open style="border:1px solid #0d9488; border-left:6px solid #0d9488; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f0fdfa;">
<summary style="cursor:pointer; padding:14px 4px; color:#134e4a; font-size:1.04rem;"><b>5.1 Why one metric is rarely enough</b></summary>

A metric is a measurement of a defined property. A **loss function** is commonly optimized during training; an **evaluation metric** is used to assess a model against a goal. They may be related, but they are not necessarily identical.

Before choosing a metric, ask:
1. What decision will use the prediction?
2. Which errors are costly, and who bears those costs?
3. Are probabilities, rankings, classes, or numerical estimates needed?
4. Is the target rare or imbalanced?
5. Does performance need to be comparable across groups or time?
6. Are there operational constraints such as review capacity or latency?

</details>


<details style="border:1px solid #2563eb; border-left:6px solid #2563eb; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#eff6ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#1e3a8a; font-size:1.04rem;"><b>5.2 Classification metric vocabulary</b></summary>

| Term | Meaning |
|---|---|
| **True positive (TP)** | Positive case predicted positive |
| **False positive (FP)** | Negative case predicted positive |
| **True negative (TN)** | Negative case predicted negative |
| **False negative (FN)** | Positive case predicted negative |
| **Accuracy** | Fraction of all predictions that are correct |
| **Precision** | Among predicted positives, fraction that are truly positive |
| **Recall / sensitivity / true positive rate** | Among actual positives, fraction found |
| **Specificity / true negative rate** | Among actual negatives, fraction correctly rejected |
| **F1 score** | Harmonic mean of precision and recall |
| **ROC curve** | True positive rate versus false positive rate across thresholds |
| **AUC-ROC** | Ranking discrimination summarized by area under ROC curve |
| **Precision–recall curve** | Precision versus recall across thresholds |
| **Average precision / PR-AUC variants** | Summary of precision–recall performance; exact convention matters |
| **Log loss / cross-entropy** | Penalizes probabilistic predictions, especially confident incorrect ones |
| **Calibration** | Agreement between predicted probabilities and observed event frequencies |
| **Decision threshold** | Cutoff that converts a score/probability into a class or action |

**Threshold matters:** a score model and a thresholded decision are different things. Moving the threshold typically changes the trade-off between false positives and false negatives.

**Imbalanced data caution:** high accuracy can be nearly meaningless if a rare positive class is missed. Choose metrics based on the intended decision and compare to appropriate baselines.
</details>


<details style="border:1px solid #7c3aed; border-left:6px solid #7c3aed; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f5f3ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#4c1d95; font-size:1.04rem;"><b>5.3 Regression metric vocabulary</b></summary>

- **MAE (Mean Absolute Error):** average absolute prediction error; often easier to interpret in target units.
- **MSE (Mean Squared Error):** average squared error; gives larger errors disproportionate influence.
- **RMSE (Root Mean Squared Error):** square root of MSE; expressed in target units.
- **\(R^2\):** compares residual variation with a reference based on the target mean under the standard definition; it is not a universal measure of “percent accuracy.”
- **Median absolute error:** robust summary of typical absolute error.
- **MAPE (Mean Absolute Percentage Error):** percentage-based error; problematic when actual values are zero or near zero and can behave poorly for some scales.
- **Pinball / quantile loss:** evaluates a specified conditional quantile.
- **Residual:** observed target minus predicted target; residual patterns can reveal bias, nonlinearity, heteroscedasticity, or data issues.

A metric must be interpreted in the context of units, target distribution, decision costs, and the baseline. A lower error number is not meaningful unless the evaluation setup is valid.
</details>


<details style="border:1px solid #059669; border-left:6px solid #059669; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#ecfdf5;">
<summary style="cursor:pointer; padding:14px 4px; color:#064e3b; font-size:1.04rem;"><b>5.4 Ranking, retrieval, and top-k metrics</b></summary>

When the system orders candidates rather than simply classifying one example, relevant concepts include:
- **Precision@k:** fraction of the top \(k\) results that are relevant.
- **Recall@k:** fraction of all relevant items retrieved within the top \(k\).
- **Mean reciprocal rank (MRR):** emphasizes the position of the first relevant result.
- **Normalized discounted cumulative gain (NDCG):** accounts for relevance grades and discounts lower-ranked results.
- **Coverage and diversity:** whether the system exposes a useful range of candidates rather than repeatedly showing a narrow set.

Ranking evaluation must define relevance, candidate sets, user/context boundaries, and the point at which results are judged.
</details>


<details style="border:1px solid #ea580c; border-left:6px solid #ea580c; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fff7ed;">
<summary style="cursor:pointer; padding:14px 4px; color:#7c2d12; font-size:1.04rem;"><b>5.5 Validation strategies</b></summary>

| Strategy | Appropriate when | Main caution |
|---|---|---|
| Random holdout | Examples are sufficiently independent and deployment resembles random sampling | Can leak entities or future information |
| K-fold cross-validation | Data is limited and observations can be split into suitable folds | Repeated entities/time dependence may invalidate folds |
| Stratified split / folds | Preserving class proportions is useful for classification | Does not solve group or temporal leakage |
| Group split | Multiple rows belong to the same person, customer, device, or entity | Group definition must match deployment |
| Time-based split | Future performance is the goal | Preserve time order; account for temporal gaps if needed |
| Rolling / expanding evaluation | Repeatedly evaluate across forecast origins | More expensive; design must reflect operational retraining |
| Spatial or domain holdout | Generalization to new locations or domains matters | A random split can overstate transfer performance |

Cross-validation is not a magic guarantee. It is only as trustworthy as the split design, data independence assumptions, and tuning procedure.
</details>


<details style="border:1px solid #db2777; border-left:6px solid #db2777; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fdf2f8;">
<summary style="cursor:pointer; padding:14px 4px; color:#831843; font-size:1.04rem;"><b>5.6 Confidence intervals, uncertainty, and statistical significance</b></summary>

- **Point estimate:** one estimated metric value.
- **Uncertainty interval:** a range intended to express uncertainty under stated assumptions.
- **Confidence interval:** a frequentist procedure with a long-run coverage interpretation.
- **Credible interval:** a Bayesian interval interpreted using a posterior distribution.
- **Statistical significance:** evidence evaluated under a specified statistical model and null hypothesis; it does not automatically imply practical importance.
- **Practical significance:** whether the size of the effect matters operationally.
- **Multiple comparisons:** trying many candidates or metrics can increase the chance of selecting a result that looks unusually good by chance.
- **Subgroup uncertainty:** small groups may have noisy metrics; show denominators and uncertainty, not only point estimates.

A tiny metric improvement may not justify extra latency, cost, complexity, or operational risk.
</details>


<details style="border:1px solid #0d9488; border-left:6px solid #0d9488; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f0fdfa;">
<summary style="cursor:pointer; padding:14px 4px; color:#134e4a; font-size:1.04rem;"><b>5.7 Calibration, thresholding, and decision costs</b></summary>

A model that outputs a probability of 0.8 is calibrated if, among comparable predictions near 0.8, the event occurs about 80% of the time. Discrimination and calibration are distinct: a model may rank cases well but have inaccurate probabilities.

A threshold should be selected using the intended action, false-positive/false-negative costs, review capacity, and constraints—not by habit. If costs vary by context, one global threshold may not be optimal or appropriate. Threshold selection must use validation data, not the final test set.
</details>


## 6. 🛡️ Generalization, overfitting, leakage, and robustness


<details open style="border:1px solid #2563eb; border-left:6px solid #2563eb; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#eff6ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#1e3a8a; font-size:1.04rem;"><b>6.1 Underfitting, overfitting, and generalization</b></summary>

- **Underfitting:** the learned system fails to capture patterns needed for the task; both training and held-out performance may be poor.
- **Overfitting:** the system adapts too closely to training data, including noise or incidental patterns, so held-out performance is worse.
- **Generalization gap:** difference between training performance and performance on unseen data, interpreted with the metric's direction in mind.
- **Bias–variance trade-off:** conceptual framework describing the tension between systematic error from restrictive assumptions and sensitivity to sample variation.
- **Regularization:** techniques or constraints that discourage undesirable complexity or parameter values.
- **Early stopping:** stop training based on validation behavior to reduce unnecessary fitting.
- **Data augmentation:** create label-preserving or task-valid variations to improve robustness or effective data diversity; validity depends on the domain.

More complexity is not always better. The goal is reliable performance under the conditions where the system will be used.
</details>


<details style="border:1px solid #7c3aed; border-left:6px solid #7c3aed; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f5f3ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#4c1d95; font-size:1.04rem;"><b>6.2 Data leakage: one of the most important interview topics</b></summary>

**Leakage** occurs when training or model selection uses information that would not legitimately be available for the intended prediction, or when evaluation information improperly influences the model.

Common forms:
- **Target leakage:** a feature directly or indirectly reveals the target or a downstream consequence of it.
- **Temporal leakage:** future information enters features or training evaluation.
- **Preprocessing leakage:** transformations are fit using validation/test data when they should be fit only on training data.
- **Entity leakage:** records for the same person/device/entity appear across partitions in a way that makes evaluation unrealistically easy.
- **Duplicate leakage:** identical or near-identical samples cross split boundaries.
- **Evaluation leakage:** repeated test-set use influences choices.
- **Human/process leakage:** manual decisions based on held-out labels seep into the modeling process.

**Prevention:** define prediction time; use realistic split design; fit preprocessing only within the training folds; preserve an untouched final test set; audit feature timestamps and lineage; deduplicate carefully; document all test-set access.
</details>


<details style="border:1px solid #059669; border-left:6px solid #059669; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#ecfdf5;">
<summary style="cursor:pointer; padding:14px 4px; color:#064e3b; font-size:1.04rem;"><b>6.3 Distribution shift and drift</b></summary>

- **Covariate / input shift:** the input distribution changes.
- **Label shift:** class proportions change under a defined shift model.
- **Concept drift:** the relationship between inputs and target changes over time.
- **Domain shift:** training and deployment come from different environments, locations, populations, or data-generation processes.
- **Data quality drift:** schema, missingness, freshness, units, or values change.
- **Prediction drift:** output distribution changes; this is a signal to investigate, not proof of model degradation.
- **Performance drift:** measured outcomes worsen once reliable labels become available.

Monitoring must distinguish an input-distribution change from actual deterioration in outcomes. Drift detectors can generate false alarms; stable input distributions also do not guarantee stable performance.
</details>


<details style="border:1px solid #ea580c; border-left:6px solid #ea580c; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fff7ed;">
<summary style="cursor:pointer; padding:14px 4px; color:#7c2d12; font-size:1.04rem;"><b>6.4 Robustness, uncertainty, and out-of-distribution inputs</b></summary>

- **Robustness:** performance remains acceptable under relevant perturbations or operating conditions.
- **Out-of-distribution (OOD):** an input differs materially from the distribution or conditions represented in training.
- **Aleatoric uncertainty:** irreducible uncertainty associated with the observation process or data.
- **Epistemic uncertainty:** uncertainty associated with limited knowledge or model/data coverage; it may reduce with better evidence.
- **Abstention / selective prediction:** the system declines to make a prediction or routes a case to a human when confidence or validity is insufficient.
- **Fallback:** predefined behavior when the model or its dependencies are unavailable or unsafe to use.

A confidence score is not automatically a calibrated probability, and a high score does not prove that an input is in-distribution.
</details>


<details style="border:1px solid #db2777; border-left:6px solid #db2777; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fdf2f8;">
<summary style="cursor:pointer; padding:14px 4px; color:#831843; font-size:1.04rem;"><b>6.5 Explainability and interpretability</b></summary>

- **Interpretability:** how directly a human can understand a model or its behavior.
- **Global explanation:** describes behavior across a dataset or population.
- **Local explanation:** describes a particular prediction.
- **Feature importance:** a measure of a feature's contribution under a specified method; different methods answer different questions.
- **Counterfactual explanation:** describes how an input might need to change to alter an outcome, subject to feasibility constraints.
- **Model-agnostic explanation:** method that can be applied to different model families, often with computational or approximation trade-offs.

Explanations are not automatically causal, faithful, stable, or complete. Validate that they are useful for the decision and audience.
</details>


## 7. 🧭 Broad algorithm-family map — no algorithm implementations

This chapter is a **navigation map only**. The individual algorithms, equations, coding examples, tuning, and algorithm-specific interview Q&A belong in the later notebooks.


<details open style="border:1px solid #0d9488; border-left:6px solid #0d9488; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f0fdfa;">
<summary style="cursor:pointer; padding:14px 4px; color:#134e4a; font-size:1.04rem;"><b>7.1 Families and the kinds of problems they often fit</b></summary>

| Family / approach | Consider it when… | Strengths to evaluate | Common cautions |
|---|---|---|---|
| Linear models | A transparent baseline or approximately additive relationship is useful | Simplicity, speed, coefficients can be inspectable | May miss complex nonlinear patterns without suitable representation |
| Tree-based models | Tabular relationships and interactions matter | Flexible decision boundaries; limited scaling requirements in many implementations | Can overfit; probabilities and extrapolation need care |
| Ensembles | Combining multiple predictors may improve predictive performance | Often strong on structured/tabular data | More complexity, latency, and explanation burden |
| Kernel methods | Similarity-based nonlinear decision boundaries are useful on manageable datasets | Flexible similarity representations | Scaling and memory can be challenging at large sample counts |
| Instance-based methods | Local similarity is meaningful and the dataset is manageable | Intuitive local comparisons | Inference cost and distance sensitivity |
| Probabilistic / generative statistical models | Distributional assumptions and uncertainty matter | Explicit assumptions can aid reasoning | Results depend on whether assumptions fit the data |
| Clustering approaches | Unlabeled grouping is useful for exploration or downstream workflows | Can surface structure | Clusters may not correspond to meaningful business segments |
| Dimensionality reduction | Compact representation or visualization is needed | Can reveal structure and reduce dimensions | Information can be lost; visual separation can mislead |
| Time-series / forecasting approaches | Temporal order, trend, seasonality, and forecast horizon matter | Explicit temporal framing | Leakage, changing regimes, and forecast-horizon evaluation |
| Deep neural networks | Complex, high-dimensional inputs or learned representations justify the cost | Flexible representation learning | Data/compute needs, debugging, latency, and governance |
| Embedding + retrieval systems | Similarity search over text, images, or items is needed | Semantic retrieval and candidate generation | Embedding quality, indexing, freshness, and retrieval evaluation |
| Anomaly-detection approaches | Rare or unknown abnormal cases matter | Useful when positive labels are scarce | “Unusual” does not necessarily mean harmful |
| Reinforcement-learning approaches | Sequential actions affect future states and rewards | Explicit decision-making under feedback | Exploration risk, reward design, simulation-to-real gaps |
| Causal inference approaches | The goal is to estimate intervention effects | Answers causal questions when assumptions/design are defensible | Predictive fit alone is insufficient; assumptions matter |

</details>


<details style="border:1px solid #2563eb; border-left:6px solid #2563eb; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#eff6ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#1e3a8a; font-size:1.04rem;"><b>7.2 How to choose a family without jumping straight to a model</b></summary>

Use this decision sequence:

1. **Clarify the task:** classification, regression, ranking, forecasting, discovery, generation, or action selection?
2. **Check the data regime:** tabular, text, image, audio, time series, graph, or multimodal?
3. **Set a baseline:** what simple rule or existing process must be beaten?
4. **Match validation to deployment:** future, new users, new regions, or randomly sampled cases?
5. **List constraints:** data volume, labels, latency, cost, interpretability, privacy, and compute.
6. **Compare candidate families fairly:** same data splits, target definition, metrics, and preprocessing boundaries.
7. **Inspect errors and slices:** aggregate performance can hide failure modes.
8. **Choose the simplest system that meets requirements:** include maintenance and operational costs.
9. **Confirm it changes a decision for the better:** offline scores are evidence, not the final business outcome.

</details>


<details style="border:1px solid #7c3aed; border-left:6px solid #7c3aed; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f5f3ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#4c1d95; font-size:1.04rem;"><b>7.3 When ML may not be the right answer</b></summary>

Prefer a rule, database query, optimization method, experiment, or process change when:
- A deterministic rule solves the problem more reliably.
- There is no defensible target or no trustworthy way to measure success.
- The available data does not cover the intended use.
- The problem is fundamentally causal but only predictive evidence is available.
- Errors are too costly and cannot be controlled with safeguards.
- A model cannot act in time or does not change the decision.
- Maintenance, governance, or data costs exceed likely benefit.
- A simpler solution is easier to validate, explain, and operate.

</details>


## 8. 🧰 Feature engineering and preprocessing concepts


<details open style="border:1px solid #059669; border-left:6px solid #059669; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#ecfdf5;">
<summary style="cursor:pointer; padding:14px 4px; color:#064e3b; font-size:1.04rem;"><b>8.1 What preprocessing means</b></summary>

Preprocessing transforms raw data into a representation suitable for the learning and inference process. It may include validation, missing-value handling, scaling, encoding, text normalization, date/time extraction, aggregation, image transforms, or other domain-specific operations.

**Critical rule:** any preprocessing operation that learns values from data—such as means, scales, category vocabularies, or feature-selection decisions—must be fitted using training data only and then applied consistently to validation, test, and production inputs.

</details>


<details style="border:1px solid #ea580c; border-left:6px solid #ea580c; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fff7ed;">
<summary style="cursor:pointer; padding:14px 4px; color:#7c2d12; font-size:1.04rem;"><b>8.2 Feature engineering, selection, and representation</b></summary>

- **Feature engineering:** construct useful inputs using domain knowledge or transformations.
- **Feature selection:** choose a subset of available features.
- **Feature extraction:** derive a representation from raw data.
- **Encoding:** represent categories or structured values numerically.
- **Scaling / normalization:** change numeric scale or distribution according to a defined procedure.
- **Interaction feature:** represent a relationship between two or more inputs.
- **Aggregation:** summarize events over a defined entity and time window.
- **Embedding:** dense vector representation intended to capture useful structure or similarity.
- **Feature store:** infrastructure for managing reusable feature definitions and values, often across training and inference.

Features must be available at prediction time, semantically valid, legally usable, and computed consistently.
</details>


<details style="border:1px solid #db2777; border-left:6px solid #db2777; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fdf2f8;">
<summary style="cursor:pointer; padding:14px 4px; color:#831843; font-size:1.04rem;"><b>8.3 Why pipelines matter</b></summary>

A pipeline bundles the transformation steps and the predictive model into one reproducible workflow. It reduces training/serving skew, makes cross-validation safer, and supports versioning and testing.

A production test should verify:
- Expected input schema and validation behavior.
- Missing and unknown category handling.
- Identical preprocessing semantics between training and inference.
- Stable output schema and sensible output ranges.
- Failure handling, logging, latency, and dependency versions.
- Compatibility of the saved artifact with the deployment runtime.

</details>


<details style="border:1px solid #0d9488; border-left:6px solid #0d9488; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f0fdfa;">
<summary style="cursor:pointer; padding:14px 4px; color:#134e4a; font-size:1.04rem;"><b>8.4 Dimensionality, sparsity, and the curse of dimensionality</b></summary>

- **Dimensionality:** number of features or coordinates in a representation.
- **Sparse representation:** most entries are zero or absent; common in bag-of-words and high-cardinality encodings.
- **Curse of dimensionality:** as dimensions grow, data becomes sparse relative to the space and many intuitive distance/statistical behaviors become less reliable without sufficient data or structure.
- **Multicollinearity:** predictors are strongly related, which can complicate coefficient interpretation and estimation in some settings.
- **High cardinality:** a categorical field has many distinct values, raising concerns about memory, generalization, and rare categories.

More features are not automatically better. Feature utility depends on signal, noise, data volume, stability, and leakage risk.
</details>


## 9. 🚀 Deployment, MLOps, and monitoring


<details open style="border:1px solid #2563eb; border-left:6px solid #2563eb; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#eff6ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#1e3a8a; font-size:1.04rem;"><b>9.1 Batch, online, edge, and embedded inference</b></summary>

| Pattern | Description | Common considerations |
|---|---|---|
| **Batch inference** | Score many records on a schedule | Throughput, freshness, retries, reproducibility |
| **Online / real-time inference** | Respond to individual requests | Latency, concurrency, availability, timeout behavior |
| **Streaming inference** | Score events as they arrive | Ordering, state, late events, backpressure |
| **Edge inference** | Run close to the data source | Memory, power, hardware support, update strategy |
| **Embedded inference** | Model is part of an application or device | Packaging, compatibility, safety, lifecycle |

Deployment architecture should follow the product's latency, freshness, reliability, cost, and privacy requirements—not trendiness.
</details>


<details style="border:1px solid #7c3aed; border-left:6px solid #7c3aed; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f5f3ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#4c1d95; font-size:1.04rem;"><b>9.2 What MLOps adds</b></summary>

**MLOps** combines software engineering, data engineering, ML workflows, and operational practices so models can be built, evaluated, deployed, monitored, and maintained reliably.

Core capabilities:
- Data and model versioning.
- Automated validation and reproducible training.
- Experiment tracking and model registry.
- Tests for data, features, model behavior, and inference API.
- CI/CD or controlled release workflows.
- Deployment strategies such as canary or shadow evaluation where appropriate.
- Monitoring, alerting, audit trails, and rollback.
- Access controls, secrets management, and artifact provenance.
- Retraining triggers and explicit approval policies.

Automation does not remove the need for judgment. A retrained model should not be promoted just because a scheduled job completed.
</details>


<details style="border:1px solid #059669; border-left:6px solid #059669; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#ecfdf5;">
<summary style="cursor:pointer; padding:14px 4px; color:#064e3b; font-size:1.04rem;"><b>9.3 Monitoring layers</b></summary>

1. **System health:** availability, latency, throughput, errors, memory, compute cost.
2. **Input data quality:** schema, missingness, ranges, freshness, unseen categories.
3. **Data/prediction drift:** distributional changes that may warrant investigation.
4. **Model performance:** metrics once reliable ground truth arrives.
5. **Business outcomes:** whether the deployed decision improves the intended result.
6. **Safety and fairness:** harmful failure modes, group-level differences, policy violations.
7. **Human workflow:** overrides, review queues, escalation rates, user feedback.

Monitor by time and relevant slices. Define alert thresholds, owners, and response playbooks before launch.
</details>


<details style="border:1px solid #ea580c; border-left:6px solid #ea580c; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fff7ed;">
<summary style="cursor:pointer; padding:14px 4px; color:#7c2d12; font-size:1.04rem;"><b>9.4 Model registry, versioning, and rollback</b></summary>

A model registry or equivalent inventory records model versions, metadata, artifacts, evaluation results, approval status, and deployment stage. A release should be traceable to its data, code, configuration, and evaluation.

**Rollback** returns service to a known-good version or safe fallback. It is an essential operational control, not an admission of failure. Model, data, and feature versions may all need to be coordinated.
</details>


<details style="border:1px solid #db2777; border-left:6px solid #db2777; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fdf2f8;">
<summary style="cursor:pointer; padding:14px 4px; color:#831843; font-size:1.04rem;"><b>9.5 Training-serving skew</b></summary>

Training-serving skew occurs when the feature values or transformations used during training differ from those used at inference. Causes include different code paths, time-window definitions, default values, category mappings, timezone handling, stale features, or dependency changes.

Mitigations include shared transformation code, pipeline packaging, feature definitions with versioning, parity tests, schema contracts, and monitoring of real production inputs.
</details>


## 10. ⚖️ Responsible ML: fairness, privacy, security, and governance


<details open style="border:1px solid #0d9488; border-left:6px solid #0d9488; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f0fdfa;">
<summary style="cursor:pointer; padding:14px 4px; color:#134e4a; font-size:1.04rem;"><b>10.1 Fairness and subgroup evaluation</b></summary>

Fairness is not a single universal metric. The appropriate analysis depends on the context, harm model, legal requirements, base rates, decision process, and affected groups.

Useful practices:
- Define the harm and affected populations.
- Evaluate relevant groups and intersections where sample sizes support meaningful analysis.
- Report group denominators and uncertainty.
- Inspect error types, access, calibration, and downstream impact.
- Review whether labels encode historical inequities or uneven measurement.
- Document trade-offs; fairness criteria can conflict under some conditions.
- Establish human oversight, appeal, and remediation pathways when appropriate.

A model with strong average performance can still cause unacceptable harm for a subgroup.
</details>


<details style="border:1px solid #2563eb; border-left:6px solid #2563eb; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#eff6ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#1e3a8a; font-size:1.04rem;"><b>10.2 Privacy, consent, and data minimization</b></summary>

- **Data minimization:** collect and retain only what is necessary for a defined purpose.
- **Purpose limitation:** use data consistently with the disclosed and authorized purpose.
- **De-identification:** remove or transform identifiers; this does not guarantee that re-identification is impossible.
- **Access control:** limit data and model access by role and need.
- **Retention policy:** define how long data and artifacts are kept.
- **Privacy risk review:** assess sensitive attributes, linkage risk, memorization, and disclosure risk where relevant.

Legal permission, consent, and contractual terms are separate from whether data is technically accessible. Sensitive data needs heightened care.
</details>


<details style="border:1px solid #7c3aed; border-left:6px solid #7c3aed; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f5f3ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#4c1d95; font-size:1.04rem;"><b>10.3 Security threats and model misuse</b></summary>

Relevant threats can include:
- **Data poisoning:** malicious or corrupted training examples influence learning.
- **Adversarial inputs:** crafted inputs cause unexpected model behavior.
- **Model extraction:** attempts to reconstruct a model through its outputs.
- **Membership inference:** attempts to infer whether a record was in training data.
- **Prompt injection / instruction manipulation:** relevant to systems that process untrusted natural-language content or use tools.
- **Supply-chain risk:** compromised packages, artifacts, datasets, or dependencies.
- **Unauthorized access:** exposure of datasets, weights, logs, or prediction endpoints.

Security controls include threat modeling, access restriction, dependency pinning and scanning, data validation, rate limits, audit logs, output controls, and incident response.
</details>


<details style="border:1px solid #059669; border-left:6px solid #059669; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#ecfdf5;">
<summary style="cursor:pointer; padding:14px 4px; color:#064e3b; font-size:1.04rem;"><b>10.4 Documentation and governance</b></summary>

A responsible model record should describe intended use, prohibited or unsupported use, data sources and limitations, evaluation design, known failure modes, performance by relevant slices, human oversight, privacy/security considerations, owners, version, and review date.

Governance is a lifecycle activity. Documentation should change when the data, model, use case, risk profile, or deployment environment changes.
</details>


## 11. 🗣️ Interview answer frameworks and common traps


<details open style="border:1px solid #ea580c; border-left:6px solid #ea580c; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fff7ed;">
<summary style="cursor:pointer; padding:14px 4px; color:#7c2d12; font-size:1.04rem;"><b>11.1 A strong 60-second answer structure</b></summary>

For a conceptual question, answer in this order:

1. **Definition:** What is it?
2. **Purpose:** Why do we need it?
3. **Mechanism:** How does it work conceptually?
4. **Trade-off:** What does it improve, and what can it cost?
5. **Use case:** Where is it appropriate?
6. **Failure mode:** What common mistake or limitation should be watched for?

This keeps answers precise while demonstrating engineering judgment.
</details>


<details style="border:1px solid #db2777; border-left:6px solid #db2777; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fdf2f8;">
<summary style="cursor:pointer; padding:14px 4px; color:#831843; font-size:1.04rem;"><b>11.2 Common interview traps</b></summary>

- Saying “accuracy is high” without a baseline, split design, or class balance.
- Treating correlation as causation.
- Using a random split when the deployment question is temporal or group-based.
- Fitting preprocessing before the split.
- Selecting a threshold using test labels.
- Treating drift as proof that performance has worsened.
- Claiming a feature-importance chart explains causality.
- Assuming the most complex model is best.
- Reporting one aggregate metric without error analysis.
- Ignoring latency, cost, monitoring, rollback, and ownership.
- Assuming that removing explicit sensitive fields eliminates all fairness risk.
- Treating an offline metric as proof of business impact.
</details>


<details style="border:1px solid #0d9488; border-left:6px solid #0d9488; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f0fdfa;">
<summary style="cursor:pointer; padding:14px 4px; color:#134e4a; font-size:1.04rem;"><b>11.3 Explain a project like an ML practitioner</b></summary>

Use this narrative:
- **Problem:** who needed what decision, and why?
- **Data:** source, target definition, time horizon, limitations.
- **Design:** split strategy and leakage prevention.
- **Baseline:** what was the reference point?
- **Evaluation:** metrics, uncertainty, subgroup and error analysis.
- **Trade-offs:** quality versus cost, latency, interpretability, and risk.
- **Deployment:** how the model was served and integrated.
- **Monitoring:** what was measured after launch?
- **Impact:** what changed in the real workflow, with evidence?
- **Learning:** what would you change next time?

Do not invent business impact. Distinguish measured outcomes from expectations or hypotheses.
</details>


## 12. 🎤 Expandable interview question bank

> Each question opens independently. Answers are designed as concise foundations; expand them into examples and trade-offs during interview practice.


<details style="border:1px solid #2563eb; border-left:6px solid #2563eb; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#eff6ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#1e3a8a; font-size:1.04rem;"><b>Q1. What is machine learning?</b></summary>

A discipline in which systems learn useful patterns from data to produce predictions, decisions, rankings, representations, or generated outputs. The key test is performance on relevant unseen cases, not merely fitting the training set.

</details>


<details style="border:1px solid #7c3aed; border-left:6px solid #7c3aed; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f5f3ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#4c1d95; font-size:1.04rem;"><b>Q2. How is AI different from ML?</b></summary>

Artificial intelligence is the broader goal or field of building systems that perform tasks associated with intelligence. Machine learning is one major approach within AI; not every AI system must learn from data.

</details>


<details style="border:1px solid #059669; border-left:6px solid #059669; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#ecfdf5;">
<summary style="cursor:pointer; padding:14px 4px; color:#064e3b; font-size:1.04rem;"><b>Q3. How is deep learning different from machine learning?</b></summary>

Deep learning is a family of ML methods based on multi-layer neural networks. It is especially useful for learned representations in high-dimensional domains, but can require substantial data, compute, and operational complexity.

</details>


<details style="border:1px solid #ea580c; border-left:6px solid #ea580c; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fff7ed;">
<summary style="cursor:pointer; padding:14px 4px; color:#7c2d12; font-size:1.04rem;"><b>Q4. What is the difference between training and inference?</b></summary>

Training estimates model parameters from data. Inference applies a fitted model to new inputs. Their compute, latency, privacy, and reliability requirements may differ.

</details>


<details style="border:1px solid #db2777; border-left:6px solid #db2777; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fdf2f8;">
<summary style="cursor:pointer; padding:14px 4px; color:#831843; font-size:1.04rem;"><b>Q5. What is a feature?</b></summary>

An input variable or representation used by a model. A valid feature must be available at prediction time and should not leak the target or future information.

</details>


<details style="border:1px solid #0d9488; border-left:6px solid #0d9488; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f0fdfa;">
<summary style="cursor:pointer; padding:14px 4px; color:#134e4a; font-size:1.04rem;"><b>Q6. What is a label or target?</b></summary>

The outcome the supervised learning system is trained to predict. It needs a precise definition, measurement process, time horizon, and documented limitations.

</details>


<details style="border:1px solid #2563eb; border-left:6px solid #2563eb; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#eff6ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#1e3a8a; font-size:1.04rem;"><b>Q7. What is the difference between a parameter and a hyperparameter?</b></summary>

Parameters are learned as part of fitting; hyperparameters are configuration choices that govern the model or training procedure and are typically selected using validation.

</details>


<details style="border:1px solid #7c3aed; border-left:6px solid #7c3aed; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f5f3ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#4c1d95; font-size:1.04rem;"><b>Q8. What is a baseline and why is it important?</b></summary>

A baseline is a simple or existing approach used as a reference. It reveals whether added ML complexity delivers meaningful improvement.

</details>


<details style="border:1px solid #059669; border-left:6px solid #059669; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#ecfdf5;">
<summary style="cursor:pointer; padding:14px 4px; color:#064e3b; font-size:1.04rem;"><b>Q9. What is generalization?</b></summary>

The ability to perform well on new examples from the intended deployment setting. A random holdout only estimates this well when its sampling assumptions match the real use case.

</details>


<details style="border:1px solid #ea580c; border-left:6px solid #ea580c; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fff7ed;">
<summary style="cursor:pointer; padding:14px 4px; color:#7c2d12; font-size:1.04rem;"><b>Q10. What is supervised learning?</b></summary>

Learning from input-target pairs to predict a defined outcome, such as a class or numerical value.

</details>


<details style="border:1px solid #db2777; border-left:6px solid #db2777; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fdf2f8;">
<summary style="cursor:pointer; padding:14px 4px; color:#831843; font-size:1.04rem;"><b>Q11. What is unsupervised learning?</b></summary>

Learning structure from data without a supplied target label, for example clustering or dimensionality reduction. Discovered structure still needs interpretation and validation.

</details>


<details style="border:1px solid #0d9488; border-left:6px solid #0d9488; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f0fdfa;">
<summary style="cursor:pointer; padding:14px 4px; color:#134e4a; font-size:1.04rem;"><b>Q12. What is semi-supervised learning?</b></summary>

Learning using a mixture of labeled and unlabeled data, often when obtaining labels is expensive.

</details>


<details style="border:1px solid #2563eb; border-left:6px solid #2563eb; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#eff6ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#1e3a8a; font-size:1.04rem;"><b>Q13. What is self-supervised learning?</b></summary>

Creating a learning signal from the data itself, often to pretrain representations that can be adapted to downstream tasks.

</details>


<details style="border:1px solid #7c3aed; border-left:6px solid #7c3aed; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f5f3ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#4c1d95; font-size:1.04rem;"><b>Q14. What is reinforcement learning?</b></summary>

Learning a policy through interactions with an environment and feedback such as rewards. It is suited to sequential decisions where actions affect future states.

</details>


<details style="border:1px solid #059669; border-left:6px solid #059669; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#ecfdf5;">
<summary style="cursor:pointer; padding:14px 4px; color:#064e3b; font-size:1.04rem;"><b>Q15. Classification versus regression?</b></summary>

Classification predicts discrete classes or class probabilities; regression estimates numerical quantities. The target definition and downstream decision determine the framing.

</details>


<details style="border:1px solid #ea580c; border-left:6px solid #ea580c; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fff7ed;">
<summary style="cursor:pointer; padding:14px 4px; color:#7c2d12; font-size:1.04rem;"><b>Q16. Clustering versus classification?</b></summary>

Clustering discovers groups without supplied class labels; classification learns to predict defined labels from labeled examples.

</details>


<details style="border:1px solid #db2777; border-left:6px solid #db2777; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fdf2f8;">
<summary style="cursor:pointer; padding:14px 4px; color:#831843; font-size:1.04rem;"><b>Q17. Prediction versus causal inference?</b></summary>

Prediction estimates an outcome under observed patterns. Causal inference asks what would change under an intervention and needs a defensible causal design and assumptions.

</details>


<details style="border:1px solid #0d9488; border-left:6px solid #0d9488; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f0fdfa;">
<summary style="cursor:pointer; padding:14px 4px; color:#134e4a; font-size:1.04rem;"><b>Q18. Why is problem framing so important?</b></summary>

It defines the prediction unit, target, horizon, available information, decision, cost of errors, and success criterion. A poorly framed task can produce a technically accurate but useless model.

</details>


<details style="border:1px solid #2563eb; border-left:6px solid #2563eb; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#eff6ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#1e3a8a; font-size:1.04rem;"><b>Q19. What is data leakage?</b></summary>

Information that should not be available to the model or model-selection process improperly influences fitting or evaluation, making performance look better than it will be in deployment.

</details>


<details style="border:1px solid #7c3aed; border-left:6px solid #7c3aed; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f5f3ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#4c1d95; font-size:1.04rem;"><b>Q20. What is target leakage?</b></summary>

A feature reveals the target directly or indirectly, often because it is recorded after the outcome or is a consequence of the target.

</details>


<details style="border:1px solid #059669; border-left:6px solid #059669; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#ecfdf5;">
<summary style="cursor:pointer; padding:14px 4px; color:#064e3b; font-size:1.04rem;"><b>Q21. What is train-test leakage?</b></summary>

Held-out information influences training or model selection, or related records cross partitions in a way that makes the test set unrealistically easy.

</details>


<details style="border:1px solid #ea580c; border-left:6px solid #ea580c; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fff7ed;">
<summary style="cursor:pointer; padding:14px 4px; color:#7c2d12; font-size:1.04rem;"><b>Q22. Why split before preprocessing?</b></summary>

To prevent held-out data from influencing learned preprocessing values such as means, scales, category vocabularies, or feature-selection choices. Fit transformations on training data and apply them to held-out data.

</details>


<details style="border:1px solid #db2777; border-left:6px solid #db2777; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fdf2f8;">
<summary style="cursor:pointer; padding:14px 4px; color:#831843; font-size:1.04rem;"><b>Q23. When should a time-based split be used?</b></summary>

When deployment predicts future outcomes and temporal order matters. Training on the past and testing on later periods better reflects that operational question.

</details>


<details style="border:1px solid #0d9488; border-left:6px solid #0d9488; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f0fdfa;">
<summary style="cursor:pointer; padding:14px 4px; color:#134e4a; font-size:1.04rem;"><b>Q24. When is a group split necessary?</b></summary>

When multiple records share an entity or dependency—such as a patient, customer, device, or household—and the intended test is generalization to unseen groups.

</details>


<details style="border:1px solid #2563eb; border-left:6px solid #2563eb; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#eff6ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#1e3a8a; font-size:1.04rem;"><b>Q25. What is cross-validation?</b></summary>

A procedure that repeatedly partitions data into training and validation folds to estimate performance or compare configurations. It does not repair a bad split design or eliminate leakage automatically.

</details>


<details style="border:1px solid #7c3aed; border-left:6px solid #7c3aed; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f5f3ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#4c1d95; font-size:1.04rem;"><b>Q26. Why keep a final test set?</b></summary>

To obtain a final estimate after model and threshold choices are complete. Repeatedly using the test set to guide choices contaminates its role.

</details>


<details style="border:1px solid #059669; border-left:6px solid #059669; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#ecfdf5;">
<summary style="cursor:pointer; padding:14px 4px; color:#064e3b; font-size:1.04rem;"><b>Q27. What is overfitting?</b></summary>

When a model adapts too closely to training data, including noise or incidental patterns, and performs worse on relevant unseen data.

</details>


<details style="border:1px solid #ea580c; border-left:6px solid #ea580c; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fff7ed;">
<summary style="cursor:pointer; padding:14px 4px; color:#7c2d12; font-size:1.04rem;"><b>Q28. What is underfitting?</b></summary>

When a model or training process is too limited to capture useful patterns, often producing poor training and held-out performance.

</details>


<details style="border:1px solid #db2777; border-left:6px solid #db2777; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fdf2f8;">
<summary style="cursor:pointer; padding:14px 4px; color:#831843; font-size:1.04rem;"><b>Q29. Explain the bias-variance trade-off.</b></summary>

It describes a balance between systematic error from restrictive assumptions and sensitivity to sample variation. The best balance depends on data, task, and deployment conditions.

</details>


<details style="border:1px solid #0d9488; border-left:6px solid #0d9488; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f0fdfa;">
<summary style="cursor:pointer; padding:14px 4px; color:#134e4a; font-size:1.04rem;"><b>Q30. What is regularization?</b></summary>

A constraint or penalty that discourages undesirable complexity or parameter values, often improving generalization at the cost of some training fit.

</details>


<details style="border:1px solid #2563eb; border-left:6px solid #2563eb; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#eff6ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#1e3a8a; font-size:1.04rem;"><b>Q31. What is class imbalance?</b></summary>

A target distribution in which some classes are much less common than others. It affects baseline interpretation, metric choice, sampling decisions, and uncertainty.

</details>


<details style="border:1px solid #7c3aed; border-left:6px solid #7c3aed; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f5f3ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#4c1d95; font-size:1.04rem;"><b>Q32. Why can accuracy be misleading?</b></summary>

If a class is rare, predicting the majority class can yield high accuracy while failing to identify the cases that matter. Use metrics aligned to error costs and compare against a baseline.

</details>


<details style="border:1px solid #059669; border-left:6px solid #059669; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#ecfdf5;">
<summary style="cursor:pointer; padding:14px 4px; color:#064e3b; font-size:1.04rem;"><b>Q33. Precision versus recall?</b></summary>

Precision asks how many predicted positives are truly positive; recall asks how many actual positives are found. Their relative importance depends on the cost of false positives and false negatives.

</details>


<details style="border:1px solid #ea580c; border-left:6px solid #ea580c; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fff7ed;">
<summary style="cursor:pointer; padding:14px 4px; color:#7c2d12; font-size:1.04rem;"><b>Q34. What is a confusion matrix?</b></summary>

A table comparing predicted and actual classes, typically showing true positives, false positives, true negatives, and false negatives. It helps reveal error types at a chosen threshold.

</details>


<details style="border:1px solid #db2777; border-left:6px solid #db2777; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fdf2f8;">
<summary style="cursor:pointer; padding:14px 4px; color:#831843; font-size:1.04rem;"><b>Q35. ROC-AUC versus PR-AUC?</b></summary>

ROC-AUC summarizes ranking discrimination using true-positive and false-positive rates; precision-recall summaries focus on positive-class retrieval and can be more informative under strong imbalance. Exact PR summary conventions vary.

</details>


<details style="border:1px solid #0d9488; border-left:6px solid #0d9488; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f0fdfa;">
<summary style="cursor:pointer; padding:14px 4px; color:#134e4a; font-size:1.04rem;"><b>Q36. What is calibration?</b></summary>

Agreement between predicted probabilities and observed event frequencies. Calibration is distinct from ranking ability and matters when decisions depend on probability meaning.

</details>


<details style="border:1px solid #2563eb; border-left:6px solid #2563eb; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#eff6ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#1e3a8a; font-size:1.04rem;"><b>Q37. What is a decision threshold?</b></summary>

A cutoff that converts a score or probability into a class or action. It should reflect costs, constraints, and validation evidence, not an arbitrary default.

</details>


<details style="border:1px solid #7c3aed; border-left:6px solid #7c3aed; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f5f3ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#4c1d95; font-size:1.04rem;"><b>Q38. What is MAE?</b></summary>

Mean absolute error, the average magnitude of errors in target units. It is relatively easy to interpret and is less dominated by very large errors than squared-error metrics.

</details>


<details style="border:1px solid #059669; border-left:6px solid #059669; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#ecfdf5;">
<summary style="cursor:pointer; padding:14px 4px; color:#064e3b; font-size:1.04rem;"><b>Q39. What is RMSE?</b></summary>

The square root of mean squared error, expressed in target units and more sensitive to large errors than MAE.

</details>


<details style="border:1px solid #ea580c; border-left:6px solid #ea580c; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fff7ed;">
<summary style="cursor:pointer; padding:14px 4px; color:#7c2d12; font-size:1.04rem;"><b>Q40. What does R-squared mean?</b></summary>

Under its standard definition, it compares residual variation with variation around the target mean. It is not a universal accuracy percentage and can be negative on held-out data.

</details>


<details style="border:1px solid #db2777; border-left:6px solid #db2777; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fdf2f8;">
<summary style="cursor:pointer; padding:14px 4px; color:#831843; font-size:1.04rem;"><b>Q41. Why is MAPE sometimes unsuitable?</b></summary>

It is undefined or unstable when actual values are zero or near zero and can behave asymmetrically. The metric should fit the scale and business meaning of the target.

</details>


<details style="border:1px solid #0d9488; border-left:6px solid #0d9488; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f0fdfa;">
<summary style="cursor:pointer; padding:14px 4px; color:#134e4a; font-size:1.04rem;"><b>Q42. What is a baseline for forecasting?</b></summary>

A simple forecast such as last value, seasonal last value, or historical average, chosen to reflect the series and forecast horizon. A complex model should demonstrate value beyond it.

</details>


<details style="border:1px solid #2563eb; border-left:6px solid #2563eb; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#eff6ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#1e3a8a; font-size:1.04rem;"><b>Q43. What is data drift?</b></summary>

A change in input data distribution or data quality over time. It is a reason to investigate, not proof by itself that model performance has deteriorated.

</details>


<details style="border:1px solid #7c3aed; border-left:6px solid #7c3aed; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f5f3ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#4c1d95; font-size:1.04rem;"><b>Q44. What is concept drift?</b></summary>

A change over time in the relationship between inputs and target, so patterns learned earlier may no longer hold.

</details>


<details style="border:1px solid #059669; border-left:6px solid #059669; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#ecfdf5;">
<summary style="cursor:pointer; padding:14px 4px; color:#064e3b; font-size:1.04rem;"><b>Q45. What is domain shift?</b></summary>

A mismatch between training and deployment environments, such as different regions, populations, devices, or acquisition processes.

</details>


<details style="border:1px solid #ea580c; border-left:6px solid #ea580c; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fff7ed;">
<summary style="cursor:pointer; padding:14px 4px; color:#7c2d12; font-size:1.04rem;"><b>Q46. What is OOD data?</b></summary>

Out-of-distribution data differs materially from the conditions represented in training. The system may need detection, abstention, fallback, or additional evaluation.

</details>


<details style="border:1px solid #db2777; border-left:6px solid #db2777; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fdf2f8;">
<summary style="cursor:pointer; padding:14px 4px; color:#831843; font-size:1.04rem;"><b>Q47. What is aleatoric uncertainty?</b></summary>

Uncertainty arising from inherent variability or noise in the observation process, which may remain even with more data.

</details>


<details style="border:1px solid #0d9488; border-left:6px solid #0d9488; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f0fdfa;">
<summary style="cursor:pointer; padding:14px 4px; color:#134e4a; font-size:1.04rem;"><b>Q48. What is epistemic uncertainty?</b></summary>

Uncertainty associated with limited knowledge or data coverage; in principle, additional informative evidence may reduce it.

</details>


<details style="border:1px solid #2563eb; border-left:6px solid #2563eb; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#eff6ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#1e3a8a; font-size:1.04rem;"><b>Q49. What is feature engineering?</b></summary>

Creating or transforming input representations using domain knowledge or data-driven procedures to improve suitability for the task.

</details>


<details style="border:1px solid #7c3aed; border-left:6px solid #7c3aed; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f5f3ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#4c1d95; font-size:1.04rem;"><b>Q50. What is feature selection?</b></summary>

Choosing a subset of inputs. It can reduce cost or complexity, but selection must be performed within the training/validation procedure to avoid leakage.

</details>


<details style="border:1px solid #059669; border-left:6px solid #059669; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#ecfdf5;">
<summary style="cursor:pointer; padding:14px 4px; color:#064e3b; font-size:1.04rem;"><b>Q51. What is an embedding?</b></summary>

A vector representation intended to encode useful structure, such as semantic similarity. Its usefulness depends on the task, training data, and evaluation.

</details>


<details style="border:1px solid #ea580c; border-left:6px solid #ea580c; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fff7ed;">
<summary style="cursor:pointer; padding:14px 4px; color:#7c2d12; font-size:1.04rem;"><b>Q52. What is the curse of dimensionality?</b></summary>

As dimensionality grows, data becomes sparse relative to the space and many distance or estimation behaviors become less reliable without enough data or useful structure.

</details>


<details style="border:1px solid #db2777; border-left:6px solid #db2777; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fdf2f8;">
<summary style="cursor:pointer; padding:14px 4px; color:#831843; font-size:1.04rem;"><b>Q53. What is multicollinearity?</b></summary>

Strong relationships among predictors that can make some coefficient estimates unstable or difficult to interpret, even when prediction remains usable.

</details>


<details style="border:1px solid #0d9488; border-left:6px solid #0d9488; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f0fdfa;">
<summary style="cursor:pointer; padding:14px 4px; color:#134e4a; font-size:1.04rem;"><b>Q54. Why use a pipeline?</b></summary>

To package preprocessing and prediction consistently, reduce training-serving skew, and support repeatable validation, testing, and deployment.

</details>


<details style="border:1px solid #2563eb; border-left:6px solid #2563eb; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#eff6ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#1e3a8a; font-size:1.04rem;"><b>Q55. What is training-serving skew?</b></summary>

A mismatch between transformations or feature values during training and inference, which can degrade real-world performance.

</details>


<details style="border:1px solid #7c3aed; border-left:6px solid #7c3aed; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f5f3ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#4c1d95; font-size:1.04rem;"><b>Q56. What is MLOps?</b></summary>

Practices and infrastructure for reproducible ML development, testing, deployment, monitoring, versioning, governance, and maintenance.

</details>


<details style="border:1px solid #059669; border-left:6px solid #059669; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#ecfdf5;">
<summary style="cursor:pointer; padding:14px 4px; color:#064e3b; font-size:1.04rem;"><b>Q57. What should be monitored in production?</b></summary>

System health, input quality, drift signals, model performance when labels arrive, business outcomes, subgroup harms, and human workflow indicators.

</details>


<details style="border:1px solid #ea580c; border-left:6px solid #ea580c; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fff7ed;">
<summary style="cursor:pointer; padding:14px 4px; color:#7c2d12; font-size:1.04rem;"><b>Q58. What is a model registry?</b></summary>

A versioned inventory of model artifacts and their metadata, evaluations, approvals, and deployment status.

</details>


<details style="border:1px solid #db2777; border-left:6px solid #db2777; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fdf2f8;">
<summary style="cursor:pointer; padding:14px 4px; color:#831843; font-size:1.04rem;"><b>Q59. Why is rollback important?</b></summary>

It allows a service to return to a known-good model or safe fallback when a release causes problems.

</details>


<details style="border:1px solid #0d9488; border-left:6px solid #0d9488; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f0fdfa;">
<summary style="cursor:pointer; padding:14px 4px; color:#134e4a; font-size:1.04rem;"><b>Q60. What is shadow deployment?</b></summary>

A candidate model receives production-like inputs and is evaluated without controlling the live decision, allowing comparison before promotion.

</details>


<details style="border:1px solid #2563eb; border-left:6px solid #2563eb; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#eff6ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#1e3a8a; font-size:1.04rem;"><b>Q61. What is a canary release?</b></summary>

A release initially exposed to a limited portion of traffic or users, with monitoring and rollback criteria before broader rollout.

</details>


<details style="border:1px solid #7c3aed; border-left:6px solid #7c3aed; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f5f3ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#4c1d95; font-size:1.04rem;"><b>Q62. What is explainability?</b></summary>

Methods for communicating aspects of model behavior or individual predictions. Explanations must be checked for fidelity, stability, and relevance to the audience.

</details>


<details style="border:1px solid #059669; border-left:6px solid #059669; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#ecfdf5;">
<summary style="cursor:pointer; padding:14px 4px; color:#064e3b; font-size:1.04rem;"><b>Q63. Does feature importance prove causality?</b></summary>

No. Feature importance describes contribution under a particular method and model; it does not by itself establish the effect of intervening on a feature.

</details>


<details style="border:1px solid #ea580c; border-left:6px solid #ea580c; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fff7ed;">
<summary style="cursor:pointer; padding:14px 4px; color:#7c2d12; font-size:1.04rem;"><b>Q64. What is a counterfactual explanation?</b></summary>

A description of how an input might need to change to alter a prediction, subject to feasibility and domain constraints.

</details>


<details style="border:1px solid #db2777; border-left:6px solid #db2777; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fdf2f8;">
<summary style="cursor:pointer; padding:14px 4px; color:#831843; font-size:1.04rem;"><b>Q65. What is fairness in ML?</b></summary>

A context-dependent set of requirements for avoiding unjustified disparities or harms. It needs a defined harm model, affected groups, metrics, and governance.

</details>


<details style="border:1px solid #0d9488; border-left:6px solid #0d9488; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f0fdfa;">
<summary style="cursor:pointer; padding:14px 4px; color:#134e4a; font-size:1.04rem;"><b>Q66. Why can fairness metrics conflict?</b></summary>

Different fairness criteria can be mathematically incompatible under certain base-rate and error conditions. The appropriate trade-off depends on context, law, and harm analysis.

</details>


<details style="border:1px solid #2563eb; border-left:6px solid #2563eb; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#eff6ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#1e3a8a; font-size:1.04rem;"><b>Q67. Does removing sensitive attributes guarantee fairness?</b></summary>

No. Other variables can act as proxies, and historical labels or processes may encode disparities. Evaluate outcomes and mechanisms directly.

</details>


<details style="border:1px solid #7c3aed; border-left:6px solid #7c3aed; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f5f3ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#4c1d95; font-size:1.04rem;"><b>Q68. What is data minimization?</b></summary>

Collecting and retaining only the data necessary for a defined, authorized purpose, reducing exposure and privacy risk.

</details>


<details style="border:1px solid #059669; border-left:6px solid #059669; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#ecfdf5;">
<summary style="cursor:pointer; padding:14px 4px; color:#064e3b; font-size:1.04rem;"><b>Q69. What is data poisoning?</b></summary>

Corrupting or manipulating training data to influence model behavior. Defense includes provenance, validation, access controls, and monitoring.

</details>


<details style="border:1px solid #ea580c; border-left:6px solid #ea580c; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fff7ed;">
<summary style="cursor:pointer; padding:14px 4px; color:#7c2d12; font-size:1.04rem;"><b>Q70. What is an adversarial input?</b></summary>

An input crafted or unusual in a way that triggers incorrect or unsafe model behavior. Relevant defenses depend on the domain and threat model.

</details>


<details style="border:1px solid #db2777; border-left:6px solid #db2777; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fdf2f8;">
<summary style="cursor:pointer; padding:14px 4px; color:#831843; font-size:1.04rem;"><b>Q71. What is model drift versus data drift?</b></summary>

Data drift concerns input distributions; model or performance drift usually refers to changing predictive behavior or measured outcomes. Definitions vary, so monitoring systems should specify each signal precisely.

</details>


<details style="border:1px solid #0d9488; border-left:6px solid #0d9488; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f0fdfa;">
<summary style="cursor:pointer; padding:14px 4px; color:#134e4a; font-size:1.04rem;"><b>Q72. Why track experiments?</b></summary>

To connect a result to data, code, configuration, environment, metrics, and artifacts, enabling fair comparison and reproducibility.

</details>


<details style="border:1px solid #2563eb; border-left:6px solid #2563eb; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#eff6ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#1e3a8a; font-size:1.04rem;"><b>Q73. What does reproducibility require?</b></summary>

Documented data versions, code, environment, split, configuration, and evaluation protocol. Exact determinism may still be limited by hardware or nondeterministic operations.

</details>


<details style="border:1px solid #7c3aed; border-left:6px solid #7c3aed; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f5f3ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#4c1d95; font-size:1.04rem;"><b>Q74. Why is the simplest adequate model often preferred?</b></summary>

It may reduce latency, cost, maintenance, explanation burden, and failure surface while meeting performance requirements. Simplicity is a constraint-aware choice, not a rule to avoid complexity.

</details>


<details style="border:1px solid #059669; border-left:6px solid #059669; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#ecfdf5;">
<summary style="cursor:pointer; padding:14px 4px; color:#064e3b; font-size:1.04rem;"><b>Q75. How do you know if ML is needed?</b></summary>

Compare against rules and existing processes, confirm reliable data and measurable outcomes, and verify that predictions can improve a real decision enough to justify lifecycle cost and risk.

</details>


<details style="border:1px solid #ea580c; border-left:6px solid #ea580c; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fff7ed;">
<summary style="cursor:pointer; padding:14px 4px; color:#7c2d12; font-size:1.04rem;"><b>Q76. What is human-in-the-loop ML?</b></summary>

A workflow in which people review, correct, approve, or override model outputs. It needs clear responsibilities, usable explanations, and monitoring of both human and model errors.

</details>


<details style="border:1px solid #db2777; border-left:6px solid #db2777; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fdf2f8;">
<summary style="cursor:pointer; padding:14px 4px; color:#831843; font-size:1.04rem;"><b>Q77. What is a model card?</b></summary>

Documentation of a model's intended use, evaluation, limitations, risks, and other relevant characteristics. It complements rather than replaces data governance and operational documentation.

</details>


<details style="border:1px solid #0d9488; border-left:6px solid #0d9488; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f0fdfa;">
<summary style="cursor:pointer; padding:14px 4px; color:#134e4a; font-size:1.04rem;"><b>Q78. What is data lineage?</b></summary>

Traceability of data sources and transformations through the pipeline, supporting debugging, audit, and reproducibility.

</details>


<details style="border:1px solid #2563eb; border-left:6px solid #2563eb; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#eff6ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#1e3a8a; font-size:1.04rem;"><b>Q79. What is a data contract?</b></summary>

A documented agreement about data schema, semantics, quality, freshness, and expectations between producers and consumers.

</details>


<details style="border:1px solid #7c3aed; border-left:6px solid #7c3aed; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f5f3ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#4c1d95; font-size:1.04rem;"><b>Q80. Why can a strong offline metric fail in production?</b></summary>

The deployment population, behavior, input availability, latency, integration, decision policy, or feedback loops may differ from offline evaluation; the metric may also be poorly aligned to business value.

</details>


<details style="border:1px solid #059669; border-left:6px solid #059669; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#ecfdf5;">
<summary style="cursor:pointer; padding:14px 4px; color:#064e3b; font-size:1.04rem;"><b>Q81. What is a feedback loop in ML?</b></summary>

Model predictions influence actions or data collection, which then influence future training data and outcomes. This can reinforce biases or change the data distribution.

</details>


<details style="border:1px solid #ea580c; border-left:6px solid #ea580c; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fff7ed;">
<summary style="cursor:pointer; padding:14px 4px; color:#7c2d12; font-size:1.04rem;"><b>Q82. What is the difference between a metric and a KPI?</b></summary>

A metric quantifies a defined property. A KPI is a metric selected as a key indicator of progress toward a goal; not every model metric is a business KPI.

</details>


<details style="border:1px solid #db2777; border-left:6px solid #db2777; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#fdf2f8;">
<summary style="cursor:pointer; padding:14px 4px; color:#831843; font-size:1.04rem;"><b>Q83. How should you handle unknown categories at inference?</b></summary>

Define and test an explicit policy, such as a safe unknown-category representation or a controlled error. Never assume production categories always match training.

</details>


<details style="border:1px solid #0d9488; border-left:6px solid #0d9488; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f0fdfa;">
<summary style="cursor:pointer; padding:14px 4px; color:#134e4a; font-size:1.04rem;"><b>Q84. Why document limitations?</b></summary>

To communicate where evidence is weak, unsupported uses, known failure modes, and the conditions under which the system needs review or fallback.

</details>


<details style="border:1px solid #2563eb; border-left:6px solid #2563eb; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#eff6ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#1e3a8a; font-size:1.04rem;"><b>Q85. What should an ML project report include?</b></summary>

Problem and target definition, data provenance, split design, baseline, metrics and uncertainty, error/slice analysis, trade-offs, deployment plan, monitoring, risks, and measured impact.

</details>


<details style="border:1px solid #7c3aed; border-left:6px solid #7c3aed; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#f5f3ff;">
<summary style="cursor:pointer; padding:14px 4px; color:#4c1d95; font-size:1.04rem;"><b>Q86. What is the biggest misconception about ML?</b></summary>

That selecting and training an algorithm is the whole project. Reliable ML requires sound framing, data, valid evaluation, deployment integration, monitoring, and governance.

</details>


## 13. 📚 Glossary and final revision checklist


<details open style="border:1px solid #059669; border-left:6px solid #059669; border-radius:14px; padding:0 16px 14px; margin:14px 0; background:#ecfdf5;">
<summary style="cursor:pointer; padding:14px 4px; color:#064e3b; font-size:1.04rem;"><b>13.1 Quick glossary</b></summary>

| Term | Short definition |
|---|---|
| Acceptance criteria | Conditions that must be met before a system is considered fit for use |
| Artifact | Versioned output needed to reproduce or serve a model |
| Batch inference | Predictions produced for groups of records on a schedule |
| Calibration | Whether predicted probabilities align with observed frequencies |
| Concept drift | Change in the input-target relationship over time |
| Data contract | Agreement on data structure, semantics, and quality |
| Data lineage | Traceability of sources and transformations |
| Data leakage | Improper access to information during training or evaluation |
| Deployment | Integration of a model into a real workflow or product |
| Distribution shift | Difference between data-generating conditions across environments or time |
| Feature | Model input variable or representation |
| Generalization | Performance on unseen relevant examples |
| Ground truth | The best available reference label or outcome measurement |
| Inference | Applying a fitted model to inputs |
| Label noise | Incorrect, ambiguous, or inconsistent target labels |
| Metric | Measurement used to evaluate a property |
| Model registry | Versioned catalog of model artifacts and their status |
| Monitoring | Ongoing checks of system, data, model, and outcome behavior |
| Pipeline | Ordered transformations and model steps packaged as a workflow |
| Precision | Fraction of predicted positives that are truly positive |
| Recall | Fraction of actual positives identified |
| Reproducibility | Ability to reconstruct a result from documented inputs and procedure |
| Target | Outcome a supervised model is intended to predict |
| Threshold | Cutoff used to turn a score into a class or action |
| Training-serving skew | Mismatch between training and inference feature processing |
| Validation set | Data used to compare candidates and make tuning decisions |
</details>
